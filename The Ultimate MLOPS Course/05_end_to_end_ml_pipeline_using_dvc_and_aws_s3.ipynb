# ***`End to End ML Pipeline using DVC AWS S3`***

---

1. Create a github repository and clone it.
2. Create a conda environment in it and install required dependencies.
3. Add an `experiments` folder that will contain the `data` and `notebook` related to the project.
4. Create a new folder `src` that will contain all the components of the projects.
5. In that folder add data_ingestion.py that will contain data ingestion code.
    
    - Run this file using
        ```bash
        python src\data_ingestion.py
        ``` 

6. Now we need to make data_preprocessing.py file that will contain code to preprocess the data.

    - Run this file using
        ```bash
        python src\data_preprocessing.py
        ``` 

7. Now we need to make feature_engineering.py file that will contain code to enginear features of the data.

    - Run this file using
        ```bash
        python src\feature_engineering.py
        ``` 

8. Now we need to make model_building.py file that will contain code to build model from the data.

    - Run this file using
        ```bash
        python src\model_building.py
        ``` 

9. Now we need to make model_evaluation.py file that will contain code to eveluate the model on the data.

    - Run this file using
        ```bash
        python src\model_evaluation.py
        ``` 

10. Add `data`, `models`, `reports` directories in `.gitignore` file. 

11. **Setting up a DVC Pipeline (without params):**

    - Create a `dvc.yaml` file and add stages in it.
    - Initialize dvc by
        
        ```bash
        dvc init
        ``` 
    - Then do this to test the pipeline automation. (check dvc dag) 
        
        ```bash
        dvc repro
        ``` 
    - If you want to visualize the dvc pipeline

        ```bash
        dvc dag
        ```
    -  Now we need to `git add .`
    -  `git commit -m "pipeline configured using dvc without params"`
    -  `git push origin main`

    - That.  


12.  **Setting up a DVC Pipeline (with params):**



- Create a ***params.yaml*** file and add model params in it.
    
    - Add params setup in your code.
    
    - Now do ***dvc repro*** again to test pipeline along with params.
    
    - Now we need to 
        ```bash
        git add .
        ```
        ```bash
        git commit -m "pipeline configured using dvc with params"
        ```

        ```bash
        git push origin main
        ``` 


13. ***Experiments with DVC***

    - For this first install
        
        ```bash
        pip install dvclive
        ```   
    - Add dcvlive in your code in `model_evaluation.py`
    
    - Run this command to run and track an experiment using ***DVC***. It will create a new `dvc.yaml`(if does not exist) and `dvclive` directory.(Each run will be considered as experiment.)
        
        ```bash
        dvc exp run
        ```
    - To see the experiments
        
        ```bash
        dvc exp show
        ``` 
    - To reproduce an experiment to get better result
        
        ```bash
        dvc exp apply (experiment name)
        ``` 
    - To remove an experiment

        ```bash
        dvc exp remove (exp name)
        ```
    


    


14. ***Adding a Remote S3 Storage to DVC***

    - Login to AWS console.
    - Create an IAM User.
      - Go to Users and Create a User.
      - Name the user.
      - In set permissions tab Select `Attach policies directly`.
        - Select  `AdministratorAccess` and Next.
        - Just Select Create User.
        - Go the User and Generate Access Key and Select `Command line Interface`.
        - Just download CSV which has aws credentials.
          
    - Create S3 (Enter unique name and create)
    - Now in order to work with dvc and aws you need to install these 2 libraries
        ```bash
        pip install dvc[aws]
        ```
        
        ```bash
        pip install awscli
        ```
    
    - Now we need to set up aws for this first use this command

        ```bash
        aws configure
        ``` 

        - Provide the downloaded `Access key ID` and `Secret access key`.
        
    - Now we need to set aws s3 bucket for this

        ```bash
        dvc remote add -d dvcstore s3://adil-mlops-dvc-s3
        ``` 

    - Now we can push the experiment with best result and it's relevent data. In our case

         ```bash
        dvc exp apply inert-seed
        ```

        - Now we just need to 

        ```bash
        dvc commit
        ``` 
        and 

        ```bash
        dvc push
        ``` 


    - Now we need to finally
        ```bash
        git add .
        ```
        ```bash
        git commit -m "pipeline configured using dvc with params"
        ```

        ```bash
        git push origin main
        ``` 


15. ***Always Remember to remove or delete aws resources***

- Delete `s3 bucket` and `IAM user`.
