# ***`Data Versioning using DVC ✅`***

---

1. Create a **`code.py`**. Add the following code in this

    ```python
    import pandas as pd
    import os

    # Create a sample DataFrame with column names
    data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['New York', 'Los Angeles', 'Chicago']
        }

    df = pd.DataFrame(data)

    # Ensure the "data" directory exists at the root level
    data_dir = 'data'
    os.makedirs(data_dir, exist_ok=True)

    # Define the file path
    file_path = os.path.join(data_dir, 'sample_data.csv')

    # Save the DataFrame to a CSV file, including column names
    df.to_csv(file_path, index=False)

    print(f"CSV file saved to {file_path}")
    ```

2. Run the file using this command in the terminal:
    
    ```bash
    python code.py
    ```

    - This will create a folder `data` and save `sample_data.csv` in that. Now we need to push this code into gothub in order to track changes.


3. Now in order to initialize dvc for Data Versioning use this command:
    
    ```bash
    dvc init
    ```

    - This will create 2 files. 
        1.  `.dvc` folder
        2.  `.gitignore` file

4. Now create a folder to store different versions of data using this command
    
    ```bash
    mkdir S3
    ```

5. Now we need to add a remote origin in order to store the data versioning, for this use this command:

    ```bash
    dvc remote add -d myremote S3
    ```

6. Now we need to specify the folder `data` which will be referenced for data versioning, mean everything will be tracked in `data` directory.

    ```bash
    dvc add data/
    ```

    - Now here we will recive the error. Now it will ask to do: ("git rm -r --cached 'data'" and "git commit -m "stop tracking data"")
    
    - Because initially we were tracking data/ folder from git so now we remove it for DVC to handle.

    ```bash
    git rm -r --cached data
    ```

    and 

    ```bash
    git commit -m "stop tracking data"
    ```

7. Again we do "dvc add data/" for dvc tracking and it will creates `data.dvc`.

    ```bash
    dvc add data/
    ```

    - To track the changes with git, run:

        ```bash
        git add data.dvc .gitignore
        ```

    - Now we need to add and commit the changes using git. 

8. Now we need to commit and push canges using dvc
    
    ```bash
    dvc commit
    ```
    ```bash
    dvc push
    ```

    - This is going to make 2 folders in `S3` folder. 1 will contain `unique id` of data and other will contain `version of data`. 

9. Now once again, we need to do a git add-commit-push to mark this stage as first version of data, since new data version has been tracked.

    ```bash
    git add .
    ```
    
    ```bash
    git commit -m "first version of data tracked"
    ```
    

    

10. Let's make another change in code and add another data point in code.py

    ```python
    import pandas as pd
    import os

    # Create a sample DataFrame with column names
    data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['New York', 'Los Angeles', 'Chicago']
        }

    df = pd.DataFrame(data)

    # # Adding new row to df for V2
    new_row_loc = {'Name': 'GF1', 'Age': 20, 'City': 'City1'}
    df.loc[len(df.index)] = new_row_loc

    # Ensure the "data" directory exists at the root level
    data_dir = 'data'
    os.makedirs(data_dir, exist_ok=True)

    # Define the file path
    file_path = os.path.join(data_dir, 'sample_data.csv')

    # Save the DataFrame to a CSV file, including column names
    df.to_csv(file_path, index=False)

    print(f"CSV file saved to {file_path}")
    ```

    - Now let's run our code file again:
        
        ```bash
        python code.py
        ```
    -  Now we need to check changes using
    
        ```bash
        dvc status
        ```
    - We need to commit and push changes once again

        ```bash
        dvc commit
        ```
        ```bash
        dvc push
        ```
    - This is going to make another 2 folders in `S3` folder. 1 will contain `unique id` of data and other will contain `version of data`. 

    - Now once again, we need to do a git add-commit-push to mark this stage as first version of data, since new data version has been tracked.

        ```bash
        git add .
        ```
        
        ```bash
        git commit -m "Second version of data tracked"
        ```

        ```bash
        git push origin main
        ```

12. Now we can go to specific version of data;

    - First we need `sha id` of specific commit which we can get using

        ```bash
        git log --oneline
        ```
    - Now in order to navigate to specific code just copy show id and navigate it using 
        ```bash
        git checkout 51b466e
        ``` 
    - Now in order to pull that data just type this you will get specific version of data.

        ```bash
        dvc pull
        ```
