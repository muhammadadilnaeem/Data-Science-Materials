# 🐍 ***`Object-Oriented Programming (OOP) in Python`***  
---

## ✅ **What is Object-Oriented Programming (OOP)?**  
📌 **OOP** is a programming paradigm that organizes code into **objects** — entities that combine **data** (attributes) and **behavior** (methods).  

### 🔹 **Key Principles of OOP:**
1. **Encapsulation**  
2. **Abstraction**  
3. **Inheritance**  
4. **Polymorphism**  

---

## 📊 **1. Key OOP Concepts and Terminologies**  

---

### 🔹 **1. Class**
A **class** is a blueprint for creating objects. It defines **attributes** (variables) and **methods** (functions) that describe the behavior of the object.  

**📋 Example: Class Definition**

```python
class Car:
    # Constructor Method (called automatically during object creation)
    def __init__(self, brand, model, year):
        self.brand = brand  # Attribute
        self.model = model
        self.year = year
    
    # Method
    def display_info(self):
        print(f"{self.brand} {self.model} ({self.year})")
```


---

### 🔹 **2. Object (Instance)**
An **object** is an **instance** of a class. It represents a real-world entity.  

**📋 Example: Creating Objects**

```python
# Creating Objects
car1 = Car("Toyota", "Corolla", 2020)
car2 = Car("Honda", "Civic", 2022)

# Accessing Methods
car1.display_info()  # Output: Toyota Corolla (2020)
car2.display_info()  # Output: Honda Civic (2022)
```


---

### 🔹 **3. `__init__()` Method (Constructor)**
The **constructor** method (`__init__()`) is automatically called when an object is created. It’s used to initialize object attributes.  

**📋 Example: Constructor Usage**

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

dog1 = Dog("Buddy", "Golden Retriever")
print(dog1.name)   # Output: Buddy
print(dog1.breed)  # Output: Golden Retriever
```


---

### 🔹 **4. `self` Keyword**
The **`self`** keyword represents the **current instance** of the class. It allows access to class attributes and methods.  

✅ `self` must be the **first parameter** in all instance methods.  
❗ Although conventionally named `self`, you can technically name it anything.  

---

### 🔹 **5. Methods (Instance Methods)**
Methods are **functions defined inside a class** that operate on the object’s attributes.  

**📋 Example: Instance Methods**

```python
class Calculator:
    def __init__(self, value):
        self.value = value

    def add(self, num):
        return self.value + num
    
    def multiply(self, num):
        return self.value * num

# Creating Object
calc = Calculator(10)
print(calc.add(5))       # Output: 15
print(calc.multiply(3))  # Output: 30
```



---

### 🔹 **6. Class Variables and Instance Variables**
- **Instance Variables:** Defined inside the `__init__()` method and specific to each object.  
- **Class Variables:** Shared across **all instances** of the class.  

**📋 Example: Class vs. Instance Variables**

```python
class Employee:
    company = "TechCorp"  # Class Variable

    def __init__(self, name, salary):
        self.name = name         # Instance Variable
        self.salary = salary     # Instance Variable

# Creating Objects
emp1 = Employee("Alice", 50000)
emp2 = Employee("Bob", 60000)

print(emp1.company)   # Output: TechCorp
print(emp2.company)   # Output: TechCorp

print("\n********\n")
# Changing Class Variable
Employee.company = "InnoTech"
print(emp1.company)   # Output: InnoTech
print(emp2.company)   # Output: InnoTech
```


---

### 🔹 **7. Encapsulation**
📌 **Encapsulation** is the process of wrapping data (attributes) and methods into a single unit (class).  

✅ In Python, encapsulation is achieved using **access modifiers**:  
- **`_protected`** → Can be accessed within the class and subclasses.  
- **`__private`** → Accessible **only within the class**.  

**📋 Example: Encapsulation**

```python
class BankAccount:
    def __init__(self, account_number, balance):
        self.__account_number = account_number  # Private Attribute
        self.__balance = balance                # Private Attribute

    # Getter Method (Access Private Attribute)
    def get_balance(self):
        return self.__balance

    # Setter Method (Modify Private Attribute)
    def deposit(self, amount):
        self.__balance += amount

# Creating Object
account = BankAccount("123456", 1000)
print(account.get_balance())  # Output: 1000

account.deposit(500)
print(account.get_balance())  # Output: 1500
```


---

### 🔹 **8. Inheritance**
📌 **Inheritance** allows a class (**child class**) to inherit properties and methods from another class (**parent class**).  

✅ Promotes **code reusability**.  
✅ Python supports multiple inheritance.  

**Types of Inheritance:**  
1. **Single Inheritance**  
2. **Multiple Inheritance**  
3. **Multilevel Inheritance**  
4. **Hierarchical Inheritance**  

**📋 Example: Single Inheritance**

```python
# Parent Class
class Animal:
    def speak(self):
        print("Animal makes a sound")

# Child Class
class Dog(Animal):
    def speak(self):
        print("Dog barks")

# Creating Objects
dog = Dog()
dog.speak()  # Output: Dog barks
```


---

### 🔹 **9. Polymorphism**
📌 **Polymorphism** allows methods to have the **same name** but behave **differently** depending on the object calling them.  

✅ Achieved via **method overriding** and **method overloading**.

**📋 Example: Method Overriding**

```python
class Bird:
    def sound(self):
        print("Bird chirps")

class Dog:
    def sound(self):
        print("Dog barks")

# Polymorphism in Action
def make_sound(animal):
    animal.sound()

# Creating Objects
bird = Bird()
dog = Dog()

make_sound(bird)  # Output: Bird chirps
make_sound(dog)   # Output: Dog barks
```


---

### 🔹 **10. Abstraction**
📌 **Abstraction** is the process of hiding implementation details and showing only essential features.  
✅ Achieved using **abstract classes** and **abstract methods** via the `abc` module.

**📋 Example: Abstraction with `abc`**

```python
from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

# Child Class
class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

# Creating Object
square = Square(4)
print(square.area())  # Output: 16
```


---

### 🔹 **11. `super()` Keyword**
📌 The `super()` function is used to call methods from the **parent class** inside the **child class**.

**📋 Example: Using `super()`**

```python
class Parent:
    def show(self):
        print("This is the Parent class")

class Child(Parent):
    def show(self):
        super().show()  # Calls the parent class method
        print("This is the Child class")

# Creating Object
child = Child()
child.show()
```


---

### 🔹 **12. `isinstance()` and `issubclass()`**
- **`isinstance()`** → Checks if an object belongs to a particular class.  
- **`issubclass()`** → Checks if a class is derived from another class.  

**📋 Example:**

```python
print(isinstance(5, int))             # Output: True
print(issubclass(Dog, Animal))        # Output: True
```

## 🚀 **13. Best Practices in OOP (Python)**  

✅ Use **meaningful class names** following **PascalCase** naming convention.  
✅ Follow the **Single Responsibility Principle (SRP)** — each class should focus on a single task.  
✅ Use **getter** and **setter** methods for controlled attribute access.  
✅ Follow **DRY (Don't Repeat Yourself)** by reusing code via inheritance.  
✅ Keep your class methods **modular** and **concise**.

```python
# simple inheritance and method overloading

# base class
class Animal:
    def __init__(self,name):
        self.name = name
    def speak(self):
        print(f"{self.name} makes a sound.") 

# derived class
class Dog(Animal):
    def speak(self):
        print(f"{self.name} barks.") 

dog = Dog("Tommy")
dog.speak()
```

```python
# simple inheritance and constructer overloading

# base class
class Animal:
    def __init__(self,name):
        self.name = name
    def speak(self):
        print(f"{self.name} makes a sound.") 

# derived class
class Dog(Animal):

    def __init__(self, name, behaviour='Friendly'):
        super().__init__(name)
        self.behaviour = behaviour

    def speak(self):
        super().speak()
        print(f"{self.name} barks in {self.behaviour} way.") 

dog = Dog("Dogesh")
dog.speak()
```

```python
# multilevel inheritance

# base class
class Grandparent:
    def __init__(self,name):
        self.name = name
    
    def tell_story(self):
        print(f"{self.name} tells a story.")

# intermediate class
class Parent(Grandparent):
    def work(self):
        print(f"{self.name} is curently working.")

# derived class
class Child(Parent):
    def play(self):
        print(f"{self.name} is Playing.")

child = Child("Jhonny")
child.tell_story()
child.work()
child.play()
```

```python
# Hierarchical Inheritance

class Parent:
    def __init__(self,name):
        self.name = name
    
    def greet(self):
        print(f"Hello my name is {self.name}.")

class Child1(Parent):
    def play(self):
        print(f"{self.name} is Playing.")

class Child2(Parent):
    def play(self):
        print(f"{self.name} is Playing.")

child1 = Child1("Dusty")
child2 = Child2("Ronda")

child1.greet()
child1.play()
print("*"*25)
child2.greet()
child2.play()
```

```python
# multiple inheritance <diamond problem>

# common base class
class A:
    def __init__(self,name):
        self.name = name

    def greet(self):
        print(f"Hello from A, {self.name}")

# intermediate class 1 
class B(A):
    def greet(self):
        print(f"Hello from B, {self.name}")
        super().greet()

# intermediate class 2
class C(A):
    def greet(self):
        print(f"Hello from C, {self.name}")
        super().greet()

# DERIVED class
class D(B,C):
    def greet(self):
        print(f"Hello from D, {self.name}")
        super().greet()

obj = D("Batista")
obj.greet() 
```

```python
# Hybrid Inheritance

# base class
class Animal:
    def __init__(self,name):
        self.name = name

    def sound(self):
        print(f"{self.name} makes a sound.")

# # Intermediate class 1 (Hierarchical)
class Mammal(Animal):
    def feed(self):
        print(f"{self.name} is Feeding Milk.")

# Intermediate class 2 (Multiple)
class Bird(Animal):
    def fly(self):
        print(f"{self.name} is Flying.")

# Derived class (Multiple Inheritance)
class Bat(Mammal,Bird):
    def __init__(self,name):
        Mammal.__init__(self,name) # Explicitly calling the constructor

    def nocturnal(self):
        print(f"{self.name} is nocturnal.")

bat = Bat("Bruce")
bat.sound()
bat.feed()
bat.fly()
bat.nocturnal()
```
