1. Download git according to your system configuration. For windows 

    ```bash
    https://git-scm.com/downloads/win
    ```
2. Create a folder and in that folder right click and select `Open Git Bash Here`

### **Basic Git Setup**

3. Now we need to set up our git.
   
   1. First define username
        ```bash
        git config --global user.name <your name>
        ```         
   2. Then define your email
        ```bash
        git config --global user.email <youremail@gmail.com>
        ```  
    3. Check set up username
        ```bash
        git config --global user.name
        ```
    4. Check setup email
        ```bash
        git config --global user.email
        ```         
### **Basic Git Commands**

4. Now in order to initialize git in your specific folder use this:
        
    ```bash
    git init
    ```

    - Now in order to check if git is initialized use this command:
        
        ```bash
        git status
        ``` 
    
    - Now in order to add files to staging area use this command:
            
        - If you want to add a single file
            
            ```bash
            git add <file name>
            ```

        - Now if you want to add all the files:

            ```bash
            git add . 
            ``` 
    - Next thing would be to add a message about the commited files:
        
        ```bash
        git commit -m "your message"
        ```
### **Basic Commits Commands**

5. Now if we want to details of all commits use this:
    ```bash
    git log
    ``` 
    
    - But if we want to details of all commits in 1 line use this:
        ```bash
        git log --oneline
        ``` 
    - If we want to details and stats of all commits this:
        ```bash
        git log --stat
        ``` 
    
    - If we want to see details of all changes made in staging area before commits use this:
        ```bash
        git diff
        ``` 
    
    - Now if you want to go to specific version of code we just need it's shaw id and we can simply go to that version of code (remember code should be commited)
     
        ```bash
        git checkout 64e9b35
        ```  
        - If you want to get back to orignal version just type `git log --oneline` and copy the latest commit shaw id and use that as 
            ```bash
            git ckeckout <id>
            ```
### **Branching and Merging**

6. If you want to create a branch:
    ```bash
    git branch <name>
    ```
    
    - If you want to create a branch on a specific **shaw id** or commit:
    
        ```bash
        git branch <name> <shaw id>
        ```
    
    - If you want to check all of branches of your repo:
        ```bash
        git branch
        ``` 
    
    - If you want to move to specific branch:
        ```bash
        git checkout <branch name>
        ```

    - Now if you check commit of all branches:
        ```bash
        git log --oneline --all
        ```
        For more further clarity in form of graph
        ```bash
        git log --oneline --all --graph
        ```

    - If you want to merge branch (Fast-forward Merge)
        ```bash
        git merge <other branch name on main>
        ``` 
    - In order to deal with the conflicts:
      

    - If you want to delete a specific branch:
        ```bash
        git branch -D <branch name>
        ``` 


### **Working with Remote Repository**

7. In order to work with remote repo

    - Declaring a remote repo
        
        ```bash
        git remote add origin <github url>
        ``` 
    - Send local changes to remote repository
        
        ```bash
        git push origin main
        ``` 
    - Pull remote changes to local repository
        
        ```bash
        git pull origin main
        ``` 
    - Now if you want to import or download any other person repository use this
        
        ```bash
        git clone <github url>
        ```


### **Understanding Software Versions**

8. If you want to add a specific version to specific commit

    - Here is how to do it:
        ```bash
        git tag -a v1.0.19 -m "Release version 1.0.19" 8184a26
        ```
        Vim editor will open
        Fn + Insert - write some message
        :wq - save and exit

    - Check if tagging successfull
        ```bash
        git log --oneline
        ```
    - Here is how to delete mentioned tag:
        ```bash
        git tag -d v1.0.19
        ```
