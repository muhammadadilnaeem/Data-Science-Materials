## **Types of Data Analytics**


Data analytics can be categorized into several types, each serving different purposes and employing various techniques. Here are the main types of data analytics:

### **1. Descriptive Analytics**

- **Definition**: Focuses on summarizing historical data to understand what has happened in the past.
- **Techniques**: Data aggregation, data mining, statistical analysis.
- **Examples**: Monthly sales reports, customer demographics analysis, website traffic analysis.

### **2. Diagnostic Analytics**

- **Definition**: Aims to determine why something happened by analyzing past data and identifying patterns or anomalies.
- **Techniques**: Drill-down analysis, data discovery, correlation analysis.
- **Examples**: Analyzing a drop in sales to identify causes, customer feedback analysis to understand dissatisfaction.

### **3. Predictive Analytics**

- **Definition**: Uses statistical models and machine learning techniques to forecast future outcomes based on historical data.
- **Techniques**: Regression analysis, time series analysis, machine learning algorithms.
- **Examples**: Sales forecasting, risk assessment, predicting customer behavior.

### **4. Prescriptive Analytics**

- **Definition**: Provides recommendations on actions to take based on data analysis, aiming to optimize outcomes.
- **Techniques**: Optimization algorithms, simulation, decision analysis.
- **Examples**: Supply chain optimization, pricing strategies, resource allocation.

### **5. Exploratory Data Analysis (EDA)**

- **Definition**: An approach to analyzing data sets to summarize their main characteristics, often using visual methods.
- **Techniques**: Visualization tools (e.g., histograms, scatter plots), descriptive statistics.
- **Examples**: Identifying patterns or trends in data before applying more formal analysis.

### **6. Inferential Analytics**

- **Definition**: Involves making inferences about a population based on a sample of data, often using probability theory.
- **Techniques**: Hypothesis testing, confidence intervals, regression analysis.
- **Examples**: Polling data analysis to predict election outcomes, A/B testing results.

### **Summary**

- **Descriptive**: What happened?
- **Diagnostic**: Why did it happen?
- **Predictive**: What is likely to happen?
- **Prescriptive**: What should be done?
- **Exploratory**: What patterns exist?
- **Inferential**: What can be inferred about a larger population?

Each type of analytics serves a unique purpose and is used in different contexts to derive insights from data.
