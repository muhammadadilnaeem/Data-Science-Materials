# **Automatic EDA With `pygwalker`**

**Pygwalker** is a Python library designed for data visualization and exploration. It provides an interactive interface to create visualizations directly from pandas DataFrames, making it easier to analyze and understand data quickly.

### **Benefits of Using Pygwalker**

1. **Interactive Visualizations**: Pygwalker allows users to create dynamic and interactive visualizations that can be explored in real-time.
2. **Ease of Use**: It simplifies the process of generating visualizations without requiring extensive coding knowledge.
3. **Integration with Pandas**: Directly works with pandas DataFrames, allowing for seamless data manipulation and visualization.
4. **Rich Visualization Options**: Offers a variety of chart types and customization options to suit different analytical needs.

### **How It Can Help with Automatic EDA**

Pygwalker automates the visualization aspect of exploratory data analysis (EDA). By quickly generating interactive charts and graphs, it helps users identify patterns, trends, and outliers in their data, making EDA more efficient and insightful.

### **Code Examples**

#### **Simple Code Example**

Here's a basic example of using Pygwalker to visualize a simple DataFrame:

```python
import pandas as pd
import pygwalker as pyg

# Create a sample DataFrame
data = {
    'age': [23, 45, 31, 29, 22],
    'salary': [50000, 60000, 55000, 70000, 48000],
    'department': ['HR', 'IT', 'IT', 'HR', 'Finance']
}
df = pd.DataFrame(data)

# Use Pygwalker to create an interactive visualization
pyg.walk(df)
```

#### **Complex Code Example**

In this example, we will load a custom dataset and use Pygwalker for more advanced visualizations:

```python
import pandas as pd
import pygwalker as pyg

# Load a custom dataset (e.g., Titanic dataset)
df = pd.read_csv('titanic.csv')

# Use Pygwalker to create interactive visualizations
pyg.walk(df)

```

### **Summary**

Pygwalker offers a powerful and intuitive way to visualize data, making exploratory data analysis more accessible. The examples provided demonstrate how to easily create visualizations for both simple and complex datasets, enhancing the overall understanding of your data.
