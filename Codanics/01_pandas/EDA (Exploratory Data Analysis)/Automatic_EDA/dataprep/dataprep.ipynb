# **DataPrep for Automatic EDA**

**DataPrep** is a Python library designed to facilitate data preparation and exploration. It provides tools for cleaning, transforming, and visualizing data, making it easier to perform exploratory data analysis (EDA) quickly and effectively.

### **Benefits of Using DataPrep**

1. **Automated Data Cleaning**: DataPrep automates many data cleaning tasks, reducing the time and effort required to prepare datasets for analysis.
2. **Visualization Tools**: It includes various visualization functions that help to understand data distributions and relationships.
3. **Integration with Pandas**: Works seamlessly with Pandas dataframes, making it accessible for users familiar with Pandas.
4. **User-Friendly**: Offers an intuitive API that simplifies the process of data exploration.

### **How DataPrep Helps with Automatic EDA**

DataPrep streamlines the EDA process by:
- Automatically identifying and handling missing values.
- Providing summary statistics and visualizations.
- Offering insights into data types and distributions.
- Allowing quick comparisons across different segments of the data.

### **Code Examples**

#### **Simple Code Example**

This example demonstrates how to use DataPrep to analyze a simple dataset like the Iris dataset.

```python
import pandas as pd
from dataprep.eda import create_report
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target

# Generate the EDA report
report = create_report(df)
report.show_browser()
```

#### **Complex Code Example**

This example shows how to use DataPrep with both training and testing datasets for a comprehensive analysis.

```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from dataprep.eda import create_report

# Load Wine dataset
wine = load_wine()
df = pd.DataFrame(data=wine.data, columns=wine.feature_names)
df['target'] = wine.target

# Split the dataset into training and testing sets
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

# Generate separate reports for training and testing datasets
train_report = create_report(train_df)
test_report = create_report(test_df)

# Show reports in the browser
train_report.show_browser(title="Training Data Report")
test_report.show_browser(title="Testing Data Report")
```

### **Conclusion**

DataPrep is a powerful tool that simplifies the data preparation and exploration process, making it a valuable asset for data scientists and analysts who want to conduct EDA efficiently.
