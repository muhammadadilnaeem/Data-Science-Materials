## **`About Dataset`**

### [Pakistan Population Census 2017 Dataset](https://www.kaggle.com/datasets/mabdullahsajid/population-of-pakistan-dataset) 

#### **Overview**  

This details presents a comprehensive dataset derived from the Pakistan Population Census conducted in 2017. The dataset encompasses demographic and administrative data, providing insights into population distribution and growth across various levels within Pakistan. It includes data for provinces, divisions, districts, and sub-divisions, along with urban and rural classifications, gender distribution, population dynamics, and historical comparisons with the 1998 census.  

---

#### **Key Features**  

1. **Province:**  
   Represents the major administrative regions of Pakistan where census data was collected.  

2. **Division:**  
   Sub-regional administrative units within provinces, forming the second level of governance.  

3. **District:**  
   Administrative areas within each division, capturing broader population statistics.  

4. **Sub-Division:**  
   Localized administrative zones (e.g., tehsils), offering detailed population insights.  

5. **Area (sq. km):**  
   The land area of each sub-division in square kilometers, enabling spatial analysis.  

6. **Urban Population (2017):**  
   The population residing in urbanized areas for each sub-division in 2017.  

7. **Rural Population (2017):**  
   The population inhabiting rural areas for each sub-division in 2017.  

8. **Male Population (2017):**  
   The total male population in each sub-division for the year 2017.  

9. **Female Population (2017):**  
   The total female population in each sub-division for the year 2017.  

10. **Transgender Population (2017):**  
    The population of transgender individuals in each sub-division for the year 2017.  

11. **Sex Ratio (2017):**  
    A demographic indicator showing the number of females per 1,000 males in 2017.  

12. **Population in 1998:**  
    Historical population data for each sub-division as per the 1998 census.  

13. **Annual Growth Rate:**  
    The yearly percentage increase in population from 1998 to 2017.  

---

#### **Data Source** 

This dataset is sourced from official Pakistan Population Census reports and related government records. Attribution to the original sources is required when utilizing this data in research or analysis.

```python
print(f"\nTotal Population of Pakistan in Rural Areas '1998' : {df['POPULATION 1998 (RURAL)'].sum()}")
print(f"\nTotal Population of Pakistan in URBAN Areas '1998' : {df['POPULATION 1998 (URBAN)'].sum()}")
```

```python
# Increase in Total population of rural areas of pakistan from 1998-2017

pop_df = df['ALL SEXES (RURAL)'].sum() - df['POPULATION 1998 (RURAL)'].sum()

print(f'Total Increase in population of Rural Areas of Pakistan From 1998-2017 : {pop_df}')
```

```python
# Increase in Total population of rural of pakistan from 1998-2017

pop_df = df['ALL SEXES (RURAL)'].sum() - df['POPULATION 1998 (RURAL)'].sum()

t_pop_df = df['ALL SEXES (RURAL)'].sum() + df['POPULATION 1998 (RURAL)'].sum()

per_df = pop_df / t_pop_df * 100

print(f'Total Percent Increase in population of Rural Areas of Pakistan From 1998-2017 : {per_df:.2f} %.')
```

```python
# Increase in Total population of URBAN areas of pakistan from 1998-2017

pop_df = df['ALL SEXES (URBAN)'].sum() - df['POPULATION 1998 (URBAN)'].sum()

t_pop_df = df['ALL SEXES (URBAN)'].sum() + df['POPULATION 1998 (URBAN)'].sum()

per_df = pop_df / t_pop_df * 100

print(f'Total Percent Increase in population of Urban Areas of Pakistan From 1998-2017 : {per_df:.2f} %.')
```
