## **Different Names of `Rows` and `Columns`**

In data science, particularly when working with data tables (like those found in databases or dataframes), rows and columns can be referred to by various names depending on the context. Here are some common terms:

### **Rows**
- **Records**: Each row typically represents a single entry or observation.
- **Instances**: Often used in machine learning contexts to denote a single example.
- **Observations**: Refers to the data collected from a single entity.
- **Samples**: Commonly used in statistical analysis and experiments.
- **Tuples**: In the context of relational databases, a row is often referred to as a tuple.

### **Columns**
- **Features**: In machine learning, columns are often called features or attributes, especially when they represent different characteristics of the data.
- **Variables**: Used in statistics to represent different types of data collected.
- **Fields**: A term often used in databases to describe a single piece of data within a record.
- **Attributes**: Similar to features, often used in data analysis to describe properties of data.
- **Dimensions**: In multidimensional data contexts, columns may be referred to as dimensions.

These terms may vary slightly based on the specific field or application within data science.
