## **Qualitative Data** and **Quantitative Data**

Data can be broadly categorized into two main types: **Qualitative Data** and **Quantitative Data**. Each of these types has further subcategories.

### **1. Qualitative Data**

Qualitative data, also known as categorical data, describes qualities or characteristics and is usually non-numeric.

#### **Subtypes:**

- **Nominal Data**: 
 
  - **Definition**: Categories without any intrinsic order.
  - **Examples**: Gender, hair color, types of cuisine.

- **Ordinal Data**: 
 
  - **Definition**: Categories with a meaningful order but without consistent intervals.
  - **Examples**: Satisfaction ratings (e.g., poor, fair, good), education levels (e.g., high school, bachelor's, master's).

### **2. Quantitative Data**

Quantitative data represents measurable quantities and is numeric in nature. This type of data can be used for mathematical calculations.

#### **Subtypes:**

- **Discrete Data**: 

  - **Definition**: Countable values that can only take certain distinct values.
  - **Examples**: Number of students in a class, number of cars in a parking lot.

- **Continuous Data**: 
 
  - **Definition**: Data that can take any value within a given range and can be measured.
  - **Examples**: Height, weight, temperature, time.

### **Summary of Differences**

- **Qualitative Data**: Non-numeric, describes characteristics (Nominal and Ordinal).
- **Quantitative Data**: Numeric, describes quantities (Discrete and Continuous).

Understanding these types and their subtypes is essential for selecting appropriate analytical techniques and for interpreting data accurately.
