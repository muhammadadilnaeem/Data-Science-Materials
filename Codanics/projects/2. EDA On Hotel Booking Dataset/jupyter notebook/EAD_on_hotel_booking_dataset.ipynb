# **<font color='purple'>About Dataset</font>**



### ***Context***

- This dataset is taken from [kaggle](https://www.kaggle.com/datasets/mojtaba142/hotel-booking). 
- This dataset contains 119390 observations for a City Hotel and a Resort Hotel. 
- Each observation represents a hotel booking between the 1st of July 2015 and 31st of August 2017, including booking that effectively arrived and booking that were canceled.

Here's a feature description table for the **Hotel Booking** dataset, covering all 36 columns:

| **Column Name**             | **Description**                                                                                   |
|-----------------------------|-----------------------------------------------------------------------------------------------|
| `hotel`                     | Type of hotel (Resort Hotel or City Hotel).                                                    |
| `is_canceled`               | Booking status (1 = Canceled, 0 = Not Canceled).                                              |
| `lead_time`                 | Number of days between booking date and arrival date.                                          |
| `arrival_date_year`         | Year of arrival date.                                                                          |
| `arrival_date_month`        | Month of arrival date.                                                                         |
| `arrival_date_week_number`  | Week number of the year for arrival.                                                           |
| `arrival_date_day_of_month` | Day of the month for arrival.                                                                  |
| `stays_in_weekend_nights`   | Number of weekend nights (Saturday/Sunday) the guest stayed or booked to stay.                 |
| `stays_in_week_nights`      | Number of weeknights (Monday–Friday) the guest stayed or booked to stay.                       |
| `adults`                    | Number of adults.                                                                              |
| `children`                  | Number of children.                                                                            |
| `babies`                    | Number of babies.                                                                              |
| `meal`                      | Type of meal booked (e.g., BB = Bed & Breakfast, HB = Half board, FB = Full board).             |
| `country`                   | Country of origin of the guest (ISO 3166-1 code).                                              |
| `market_segment`            | Market segment designation (e.g., Direct, Corporate, Online TA).                               |
| `distribution_channel`      | Booking distribution channel (e.g., Direct, TA/TO, GDS).                                       |
| `is_repeated_guest`         | Indicates if the guest is a repeated guest (1 = Yes, 0 = No).                                 |
| `previous_cancellations`    | Number of previous bookings canceled by the customer.                                          |
| `previous_bookings_not_canceled` | Number of previous bookings not canceled by the customer.                                 |
| `reserved_room_type`        | Code of room type reserved by the customer.                                                    |
| `assigned_room_type`        | Code of room type assigned to the customer.                                                    |
| `booking_changes`           | Number of changes made to the booking.                                                         |
| `deposit_type`              | Type of deposit (No Deposit, Non Refund, Refundable).                                          |
| `agent`                     | ID of the travel agent who made the booking.                                                   |
| `company`                   | ID of the company that made the booking.                                                       |
| `days_in_waiting_list`      | Number of days the booking was on the waiting list.                                            |
| `customer_type`             | Type of customer (e.g., Contract, Group, Transient).                                           |
| `adr`                       | Average Daily Rate (price per room per night).                                                 |
| `required_car_parking_spaces` | Number of car parking spaces required by the customer.                                      |
| `total_of_special_requests` | Number of special requests made by the customer.                                               |
| `reservation_status`        | Final status of the booking (e.g., Canceled, Check-Out, No-Show).                              |
| `reservation_status_date`   | Date when the reservation status was updated.                                                  |
| `arrival_date`              | Complete arrival date (derived from year, month, and day).                                     |
| `total_stay`                | Total number of nights stayed (sum of weekend and week nights).                                |
| `total_guests`              | Total number of guests (sum of adults, children, and babies).                                  |
| `family_or_not`             | Indicates if the booking includes family (1 = Yes, if children/babies > 0, else 0).            |
| `stay_type`                 | Classifies booking as Short Stay, Long Stay, or Extended Stay based on the number of nights.   |

# **<font color='purple'>Now Perform Exploratory Data Analysis on Cleaned Data</font>**


#### **Calculate Percentage of Reservations Cancelled**

```python
cancelled_pecent = df["is_canceled"].value_counts(normalize=True)
cancelled_pecent
```

##### ***`Observations`***

- `62%` of reservations were not cancelled.
- Only `37%` reservations were cancelled.

```python

# plot the cancelation 
plt.figure(figsize=(8,7))
sns.countplot(x='is_canceled', data=df, palette=["green", "red"])
plt.xticks(ticks=[0, 1], labels=['Not canceled', 'Canceled'])
plt.title("Reservation Status Count")
plt.show()
```

```python

# make a count plot to visualize the canceled reservation in hotels

plt.figure(figsize=(8,4))
ax1 = sns.countplot(data=df, x="hotel", hue="is_canceled", palette="Blues")
legend_labels, _ = ax1.get_legend_handles_labels()
ax1.legend()
plt.title("Reservation Status in Different Hotels")
plt.xlabel("Hotel")
plt.ylabel("Number of Reservations")
plt.legend(["Not Canceled","Canceled"])
plt.show()
```

- `Resort Hotels` has Less reservations and less `Cancelation` rate.
- On the other hand, `City Hotels` has more reservations and more `Cancelation` rate.

```python
# filter data to check the reservaions canceled in resort hotels
resort_hotels = df[df['hotel'] == "Resort Hotel"]
resort_hotels['is_canceled'].value_counts(normalize=True)
```


- `72%` Reservations were not calcelled in ***Resort Hotels***.
- Only `28%` Reservations were calcelled in ***Resort Hotels***.

```python
# filter data to check the reservaions canceled in city hotels
city_hotels = df[df['hotel'] == "City Hotel"]
city_hotels['is_canceled'].value_counts(normalize=True)
```


- `58%` Reservations were not calcelled in ***City Hotels***.
- Only `41%` Reservations were calcelled in ***City Hotels***.

```python
# take average daily rate for resort hotels
resort_hotels = resort_hotels.groupby("reservation_status_date")[['adr']].mean()  

# take average daily rate for city hotels
city_hotels = city_hotels.groupby("reservation_status_date")[['adr']].mean()  
```

```python

# visualize average daily rate in both city and resort hotels
plt.figure(figsize=(20,8))
plt.title("Average Daily Rate in City and Resort Hotels", fontsize=30)
plt.plot(resort_hotels.index,resort_hotels['adr'],label="Resort Hotels")
plt.plot(city_hotels.index,city_hotels['adr'],label="City Hotels")
plt.legend(fontsize=22)
plt.show()
```

```python

# now let's visualize reservations in different months
df['month'] = df['reservation_status_date'].dt.month # get month from datetime 
plt.figure(figsize=(16,8))
ax1 = sns.countplot(x="month",hue="is_canceled",data=df,palette="bright")
legend_labels,_ = ax1.get_legend_handles_labels()
ax1.legend(bbox_to_anchor=(1,1))
plt.title("Reservation Status Per Month",fontsize=30)
plt.xlabel("Month")
plt.ylabel("Number Of Reservations")
plt.legend(["Not Canceled", "Canceled"],fontsize=20)
plt.show()
```

- Most of the Reservations were cancelled in Month of *`January`*.
- Least of the Reservations were cancelled in Month of *`September`*.

```python
# plot average daily rate per month for cancelled reservations
plt.figure(figsize=(15,8))
plt.title("Average Daily Rate Per Month", fontsize=30)

# Create a colorful palette with the number of unique months
unique_months = df['month'].nunique()
colors = sns.color_palette("hsv", unique_months)

# Plot with the colorful palette
sns.barplot(
    x="month", 
    y="adr", 
    data=df[df['is_canceled']==1].groupby('month')[['adr']].sum().reset_index(), 
    palette=colors
)

plt.show()
```


***`Now it is Clear That`***

- Most of the Reservations were cancelled in Month of *`January`*.
- Least of the Reservations were cancelled in Month of *`September`*.
- Factor may be Due to `average daily rate for reservation`

```python
# now let's calculate reservation cancelation cancelation rate on the base of different countries according to our data
cancelled_data = df[df['is_canceled']== 1]
top_10_countries = cancelled_data['country'].value_counts()[:10]
```

```python
# now let's visualize the top_ten_countries using pie chart
plt.figure(figsize=(9,9))
plt.title("Top 10 Countries with High Reservation Cncelation Rate",fontsize=30)
plt.pie(top_10_countries,autopct="%.2f",labels=top_10_countries.index)
plt.show()
```


`According to this visualization` :

- `Portugal` has highest reservation cancelation rate with `70.07%`.

```python
# calculate price for the canceled and non-canceld reservations

# Create a DataFrame for canceled reservations and compute mean ADR by date
cancelled_df_adr = cancelled_data.groupby('reservation_status_date')['adr'].mean().reset_index()
cancelled_df_adr.sort_values('reservation_status_date', inplace=True)

# calculate non-canceled data
not_cancelled_data = df[df['is_canceled']== 0]

# Create a DataFrame for not canceled reservations and compute mean ADR by date
not_cancelled_df_adr = not_cancelled_data.groupby('reservation_status_date')['adr'].mean().reset_index()
not_cancelled_df_adr.sort_values('reservation_status_date', inplace=True)

# Plotting both lines on the same figure
plt.figure(figsize=(12, 6))
plt.title('Average Daily Rate')

plt.plot(
    not_cancelled_df_adr['reservation_status_date'], 
    not_cancelled_df_adr['adr'], 
    label='Not Cancelled'
)
plt.plot(
    cancelled_df_adr['reservation_status_date'], 
    cancelled_df_adr['adr'], 
    label='Cancelled'
)

plt.legend()
plt.show()
```


- Since all the data is litle hard to understand, so we will use only from `2016 to 2017` data.

```python
cancelled_df_adr = cancelled_df_adr[(cancelled_df_adr["reservation_status_date"]>"2016") & (cancelled_df_adr['reservation_status_date']<"2017-09")]
not_cancelled_df_adr = not_cancelled_df_adr[(not_cancelled_df_adr["reservation_status_date"]>"2016") & (not_cancelled_df_adr['reservation_status_date']<"2017-09")]
```

```python
# again calculate price for the canceled and non-canceld reservations

# Plotting both lines on the same figure
plt.figure(figsize=(20, 6))
plt.title('Average Daily Rate',fontsize=30)

plt.plot(
    not_cancelled_df_adr['reservation_status_date'], 
    not_cancelled_df_adr['adr'], 
    label='Not Cancelled'
)
plt.plot(
    cancelled_df_adr['reservation_status_date'], 
    cancelled_df_adr['adr'], 
    label='Cancelled'
)

plt.legend(fontsize=20)
plt.show()
```

- So `average daily rate` is the factor that is influencing the cancelation rate.
