## ✅ **1. Data Manipulation & Analysis** 🧹

These libraries help in **cleaning**, **processing**, and **exploring** data before feeding it into ML models.

### 📁 **1.1 Pandas**  
- **Purpose:** Data manipulation, analysis, and structured data handling.  
- **Key Features:**  
  - DataFrames for tabular data 📊  
  - Easy CSV/Excel/SQL data import/export  
  - Data wrangling, filtering, grouping, and aggregation  

```python
import pandas as pd

# Load CSV file
df = pd.read_csv('data.csv')

# Basic Operations
print(df.head())
print(df.describe())

# Filtering
filtered_df = df[df['Age'] > 30]
```

### 📊 **1.2 NumPy**  
- **Purpose:** High-performance numerical computing.  
- **Key Features:**  
  - Multi-dimensional arrays (**ndarray**)  
  - Mathematical functions (e.g., linear algebra, Fourier transforms)  
  - Element-wise operations  

```python
import numpy as np

# Create a NumPy array
arr = np.array([1, 2, 3, 4, 5])

# Basic Operations
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Reshaped Array:\n", arr.reshape(5, 1))
```

---

### 📊 **1.3 Dask**  
- **Purpose:** Parallel computing for large datasets.  
- **Key Features:**  
  - Scales Pandas and NumPy for **big data**  
  - Out-of-core computation for datasets that don’t fit in memory  

```python
import dask.dataframe as dd

# Load large CSV
df = dd.read_csv('large_data.csv')

# Lazy computation
result = df.groupby('Category').mean().compute()
print(result)
```

## 📊 **2. Data Visualization** 🎨

Visualization helps in **exploratory data analysis (EDA)**, identifying patterns, and communicating insights.

### 📈 **2.1 Matplotlib**  
- **Purpose:** Basic data visualization.  
- **Key Features:**  
  - Line plots, bar charts, scatter plots, histograms  
  - Full control over plot aesthetics  

```python
import matplotlib.pyplot as plt

# Sample Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Plot
plt.plot(x, y, color='blue', marker='o')
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## 🤖 **3. Machine Learning Frameworks** 🚀

These libraries offer tools and algorithms for building and training Machine Learning models.

### 📊 **3.1 Scikit-learn** *(The Go-To Library for ML)*  
- **Purpose:** Classical Machine Learning algorithms.  
- **Key Features:**  
  - Classification, regression, clustering  
  - Model selection, cross-validation, hyperparameter tuning  
  - Preprocessing tools  

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Load Dataset
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2)

# Train Random Forest
model = RandomForestClassifier()
model.fit(X_train, y_train)

# Predict & Evaluate
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

---

### ⚡ **3.2 XGBoost** *(Extreme Gradient Boosting)*  
- **Purpose:** High-performance gradient boosting.  
- **Key Features:**  
  - Regularized boosting to avoid overfitting  
  - Handles missing values and imbalanced data  

```python
import xgboost as xgb
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load Data
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2)

# Train XGBoost
model = xgb.XGBClassifier()
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print("XGBoost Accuracy:", accuracy_score(y_test, y_pred))
```

---

### ⚡ **3.3 LightGBM** *(Light Gradient Boosting Machine)*  
- **Purpose:** Fast and scalable gradient boosting framework.  
- **Key Features:**  
  - Efficient with large datasets  
  - Supports categorical features natively  

```python
import lightgbm as lgb
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load Data
digits = load_digits()
X_train, X_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.2)

# Train LightGBM
model = lgb.LGBMClassifier()
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print("LightGBM Accuracy:", accuracy_score(y_test, y_pred))
```

## 🧠 **4. Deep Learning Libraries** 🧮

For complex models like **Neural Networks**, these libraries offer powerful tools.

### 🤖 **4.1 TensorFlow** *(by Google)*  
- **Purpose:** Building deep learning models.  
- **Key Features:**  
  - Supports Neural Networks, CNNs, RNNs  
  - Scalable for both CPUs and GPUs  
  - TensorBoard for visualization  

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Build Neural Network
model = Sequential([
    Dense(32, activation='relu', input_shape=(10,)),
    Dense(1, activation='sigmoid')
])

# Compile
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train (using random data)
import numpy as np
X = np.random.rand(100, 10)
y = np.random.randint(0, 2, 100)
model.fit(X, y, epochs=5)
```

---

### 🤖 **4.2 PyTorch** *(by Facebook)*  
- **Purpose:** Deep learning and dynamic computation graphs.  
- **Key Features:**  
  - Dynamic graphing for flexibility  
  - Strong community and wide adoption in research  
  - Native GPU support  

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Simple Neural Network
class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc = nn.Linear(10, 1)

    def forward(self, x):
        return torch.sigmoid(self.fc(x))

# Initialize Model
model = Net()
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# Random Data
X = torch.rand(100, 10)
y = torch.randint(0, 2, (100, 1), dtype=torch.float)

# Training Loop
for epoch in range(5):
    optimizer.zero_grad()
    output = model(X)
    loss = criterion(output, y)
    loss.backward()
    optimizer.step()
    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```

## 📈 **5. Model Evaluation & Hyperparameter Tuning** 🎛️

### ⚖️ **5.1 Scikit-learn’s GridSearchCV & RandomizedSearchCV**  
- **Purpose:** Hyperparameter tuning using grid or random search.  
- **Example:** Find the best `max_depth` and `n_estimators` for a Random Forest.  

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

# Define Parameter Grid
param_grid = {'n_estimators': [50, 100, 150], 'max_depth': [5, 10, 20]}

# Apply Grid Search
grid_search = GridSearchCV(RandomForestClassifier(), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# Best Parameters
print("Best Params:", grid_search.best_params_)
```

---

### 🎛️ **5.2 Optuna** *(Automated Hyperparameter Tuning)*  
- **Purpose:** Efficient hyperparameter optimization using Bayesian methods.  

```python
import optuna

def objective(trial):
    x = trial.suggest_float('x', -10, 10)
    return (x - 2) ** 2

study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=100)

print("Best Parameter:", study.best_params)
```
