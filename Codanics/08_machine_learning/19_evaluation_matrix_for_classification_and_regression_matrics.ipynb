# 📊 ***`Evaluation Metrics for Regression & Classification Models in Machine Learning`*** 🚀  

### ✅ **Why Are Evaluation Metrics Important?**  
📌 **Evaluation metrics** help assess the performance of Machine Learning (ML) models by measuring how well they make predictions.  

✔ **Regression Metrics** → Used for **continuous output** (e.g., house prices, stock prices).  
✔ **Classification Metrics** → Used for **categorical output** (e.g., spam/not spam, disease/no disease).  

---

# 📈 **1. Evaluation Metrics for Regression Models**  

### 📌 **1.1. Mean Absolute Error (MAE)**  
- Measures the **average absolute difference** between actual and predicted values.  
- **Formula:**  
  $
  MAE = \frac{1}{n} \sum_{i=1}^{n} |Y_i - \hat{Y}_i|
  $  

✔ **Best when** small errors matter more than large ones.  
✔ **Less sensitive to outliers** than MSE.  

```python
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(y_test, y_pred)
print("MAE:", mae)
```


---

### 📌 **1.2. Mean Squared Error (MSE)**  
- Measures the **average squared difference** between actual and predicted values.  
- **Formula:**  
  $
  MSE = \frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2
  $  

✔ **Gives higher weight to larger errors (penalizes large errors more).**  
✔ **More sensitive to outliers** than MAE.  

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test, y_pred)
print("MSE:", mse)
```


---

### 📌 **1.3. Root Mean Squared Error (RMSE)**  
- Square root of **MSE**, making it interpretable in the same unit as the target variable.  
- **Formula:**  
  $
  RMSE = \sqrt{MSE}
  $  

✔ **Easier to interpret than MSE** because it has the same unit as $ Y $.  
✔ **Sensitive to outliers** (due to squaring).  

```python
rmse = mean_squared_error(y_test, y_pred, squared=False)
print("RMSE:", rmse)
```


---

### 📌 **1.4. R² Score (Coefficient of Determination)**  
- Measures the **proportion of variance** explained by the model.  
- **Formula:**  
  $
  R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
  $  
  Where:  
  - **$ SS_{res} $** = Sum of squared residuals.  
  - **$ SS_{tot} $** = Total sum of squares.  

✔ **R² = 1** → Perfect model.  
✔ **R² = 0** → Model is no better than the mean.  
✔ **R² < 0** → Model performs worse than simply using the mean.  


```python
from sklearn.metrics import r2_score
r2 = r2_score(y_test, y_pred)
print("R² Score:", r2)
```



---

### 📌 **1.5. Adjusted R² Score**  
- Adjusts **R² Score** for the number of predictors (useful for multiple regression).  
- **Formula:**  
  $
  Adjusted\ R^2 = 1 - \left( \frac{(1 - R^2)(n - 1)}{n - p - 1} \right)
  $  

✔ **Used when comparing models with different numbers of predictors**.  

```python
def adjusted_r2(r2, n, p):
    return 1 - ((1 - r2) * (n - 1) / (n - p - 1))

adj_r2 = adjusted_r2(r2, len(y_test), X_test.shape[1])
print("Adjusted R² Score:", adj_r2)  
```

# 📉 **2. Evaluation Metrics for Classification Models**  

### 📌 **2.1. Accuracy**  
- Measures the **percentage of correctly predicted instances**.  
- **Formula:**  
  $
  Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
  $  

✔ **Works well for balanced datasets** but **not for imbalanced datasets**.

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy) 
```


---

### 📌 **2.2. Precision (Positive Predictive Value - PPV)**  
- Measures how many **predicted positives are actually positive**.  
- **Formula:**  
  $  
  Precision = \frac{TP}{TP + FP}
  $

✔ **Use when False Positives (FP) matter** (e.g., spam detection). 

```python
from sklearn.metrics import precision_score
precision = precision_score(y_test, y_pred)
print("Precision:", precision) 
```


---

### 📌 **2.3. Recall (Sensitivity, True Positive Rate - TPR)**  
- Measures how many **actual positives were correctly predicted**.  
- **Formula:**  
  $
  Recall = \frac{TP}{TP + FN}
  $  

✔ **Use when False Negatives (FN) matter** (e.g., medical diagnosis). 

```python
from sklearn.metrics import recall_score
recall = recall_score(y_test, y_pred)
print("Recall:", recall)
```


---

### 📌 **2.4. F1 Score (Harmonic Mean of Precision & Recall)**  
- **Balances Precision & Recall**.  
- **Formula:**  
  $
  F1 = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}
  $  

✔ **Useful for imbalanced datasets**.  

```python
from sklearn.metrics import f1_score
f1 = f1_score(y_test, y_pred)
print("F1 Score:", f1)
```


---

### 📌 **2.5. Confusion Matrix**  
- Shows **True Positives, False Positives, True Negatives, False Negatives** in a table format.  

```python
from sklearn.metrics import confusion_matrix
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", conf_matrix)
```


---

### 📌 **2.6. ROC Curve & AUC Score**  
- **ROC Curve (Receiver Operating Characteristic)** → Plots **True Positive Rate (TPR) vs. False Positive Rate (FPR)**.  
- **AUC Score (Area Under Curve)** → Measures **overall classification performance** (higher is better).  

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Get Probabilities
y_prob = log_reg.predict_proba(X_test_scaled)[:, 1]

# Compute ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_prob)

# Plot ROC Curve
plt.plot(fpr, tpr, label="ROC Curve (AUC = {:.2f})".format(roc_auc_score(y_test, y_prob)))
plt.plot([0, 1], [0, 1], linestyle="--", color="gray")  # Random guess line
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()

# Compute AUC Score
auc_score = roc_auc_score(y_test, y_prob)
print("AUC Score:", auc_score)
```

# 🎯 **Key Takeaways**  

✔ **Regression Metrics**: MAE, MSE, RMSE, R² Score, Adjusted R².  
✔ **Classification Metrics**: Accuracy, Precision, Recall, F1 Score, Confusion Matrix, ROC-AUC.  
✔ **Use F1 Score for imbalanced classification problems**.  
✔ **Use Adjusted R² for multiple regression**.  

💡 **Pro Tip:** Always choose the metric that best aligns with your problem **(e.g., F1 Score for imbalanced datasets, RMSE for regression models).** 🚀
