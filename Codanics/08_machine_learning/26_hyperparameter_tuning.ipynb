# 🎯 ***`Hyperparameter Tuning in Machine Learning for Data Science`***  

---

## ✅ **What is Hyperparameter Tuning?**  
📌 **Hyperparameter tuning** is the process of finding the **best combination** of hyperparameters to improve a machine learning model's performance. Unlike model parameters (learned during training), **hyperparameters** are set **before training** and directly influence the learning process.  

✅ Crucial for boosting model performance.  
✅ Helps control **overfitting** and **underfitting**.  
✅ Key to achieving optimal accuracy, precision, or recall.  

---

## 🔎 **1. Hyperparameters vs Parameters**  

| Aspect                | **Hyperparameters** | **Parameters** |
|-----------------------|---------------------|-----------------|
| **Definition**            | Settings configured **before training**. | Values learned by the model **during training**. |
| **Examples**              | Learning rate, max_depth, n_estimators. | Coefficients in linear regression, weights in neural networks. |
| **Tuning Process**         | Found via **GridSearchCV**, **RandomizedSearchCV**, etc. | Automatically adjusted through model training. |
| **Direct Impact on**        | **Model complexity**, **speed**, and **performance**. | **Final model predictions**. |


## 📊 **2. Common Hyperparameters in ML Models**  

### 🔹 **Decision Trees / Random Forest / XGBoost**
- `max_depth` → Limits tree depth to prevent overfitting.  
- `min_samples_split` → Minimum samples required to split a node.  
- `n_estimators` → Number of trees in the ensemble.  
- `learning_rate` → Controls the step size during learning (Boosting models).  

---

### 🔹 **Support Vector Machine (SVM)**
- `C` → Controls the trade-off between **margin width** and **misclassification**.  
- `kernel` → Defines the type of decision boundary (e.g., linear, RBF).  
- `gamma` → Controls the influence of individual data points.  

---

### 🔹 **k-Nearest Neighbors (KNN)**
- `n_neighbors` → Number of neighbors to consider.  
- `metric` → Defines the distance measure (e.g., Euclidean, Manhattan).  

---

### 🔹 **Neural Networks**
- `learning_rate` → Step size during gradient updates.  
- `batch_size` → Number of samples processed in each iteration.  
- `epochs` → Number of times the model sees the entire dataset.  

---

## 🧠 **3. Techniques for Hyperparameter Tuning**  

### 🔹 **1. Manual Search**
✅ Manually adjusting hyperparameters based on experience or intuition.  
❌ Can be inefficient and time-consuming.  

---

### 🔹 **2. GridSearchCV (Exhaustive Search)**
✅ Trains the model on all possible combinations of hyperparameter values.  
✅ Ensures the **best combination** is found.  
❌ Can be computationally expensive for large search spaces.

### 🔹 **3. RandomizedSearchCV (Random Search)**
✅ Randomly samples hyperparameter combinations from a defined range.  
✅ Faster than GridSearchCV for **large search spaces**.  
❌ May miss the **best combination** if the search space is too large.  

**📋 Example Code: RandomizedSearchCV**

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint

# Hyperparameter Distribution
param_dist = {
    'n_estimators': randint(50, 300),
    'max_depth': randint(3, 15),
    'min_samples_split': randint(2, 10)
}

# Initialize Model
rf_model = RandomForestClassifier(random_state=42)

# RandomizedSearchCV Implementation
random_search = RandomizedSearchCV(rf_model, param_dist, n_iter=20, cv=5, scoring='accuracy', random_state=42)
random_search.fit(X_train, y_train)

print("Best Parameters:", random_search.best_params_)
print("Best Accuracy:", random_search.best_score_)
```


---

### 🔹 **4. Bayesian Optimization (BayesSearchCV)**
✅ Uses **Bayesian inference** to select the most promising hyperparameter combinations.  
✅ Efficiently finds optimal values by focusing on the best-performing areas.  
✅ Ideal for **expensive models** with **large search spaces**.  

**📋 Example Code: Bayesian Optimization**

```python
from skopt import BayesSearchCV
from sklearn.ensemble import GradientBoostingClassifier

# Define Parameter Search Space
param_space = {
    'learning_rate': (0.01, 0.3),
    'n_estimators': (100, 300),
    'max_depth': (3, 10),
}

# Initialize Model
gb_model = GradientBoostingClassifier(random_state=42)

# BayesSearchCV Implementation
bayes_search = BayesSearchCV(gb_model, param_space, n_iter=30, cv=5, scoring='accuracy')
bayes_search.fit(X_train, y_train)

print("Best Parameters:", bayes_search.best_params_)
print("Best Accuracy:", bayes_search.best_score_)
```

### 🔹 **5. Automated Machine Learning (AutoML)**
✅ Tools like **TPOT**, **H2O AutoML**, and **AutoKeras** automate model selection and hyperparameter tuning.  
✅ Efficient for complex data and large feature spaces.  
❌ May require extensive computational resources.

## 🔥 **4. Best Practices for Hyperparameter Tuning**  

✅ Start with **RandomizedSearchCV** for initial tuning, then refine with **GridSearchCV**.  
✅ Use **cross-validation** (e.g., `cv=5`) to improve model reliability.  
✅ Tune **one hyperparameter at a time** if computational resources are limited.  
✅ For ensemble models like **XGBoost** or **Random Forest**, focus on:  
   - **`n_estimators`** (number of trees)  
   - **`max_depth`** (tree depth)  
   - **`learning_rate`** (step size)  
✅ For neural networks, optimize:  
   - **`batch_size`**, **`epochs`**, and **`learning_rate`**  
✅ Apply **feature scaling** (e.g., `StandardScaler`) when tuning distance-based models like **KNN** or **SVM**.
