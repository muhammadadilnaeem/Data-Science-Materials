# 🚀 ***`Support Vector Machine (SVM) in Machine Learning for Data Science`***  

---

## ✅ **What is an SVM (Support Vector Machine)?**  

📌 **Support Vector Machine (SVM)** is a powerful **supervised learning algorithm** used for both **classification** and **regression** tasks.  
📌 SVM excels in solving **linear** and **non-linear** classification problems by finding the **optimal hyperplane** that best separates the data points.  

✅ Best for **high-dimensional data**.  
✅ Effective for **binary** and **multi-class classification**.  
✅ Robust against **overfitting**, especially with the **kernel trick**.  

---

## 🎯 **1. Key Concepts in SVM**  

### 🔹 **1.1. Hyperplane**  
📌 A **hyperplane** is a decision boundary that separates different classes in the feature space.  
- For **2D data**, the hyperplane is a **line**.  
- For **3D data**, the hyperplane is a **plane**.  
- For **N-dimensional data**, the hyperplane is an **(N-1)-dimensional boundary**.  

---

### 🔹 **1.2. Support Vectors**  
📌 **Support Vectors** are the data points that are closest to the hyperplane.  
- These points **define the optimal boundary**.  
- The model aims to maximize the **margin** between the hyperplane and these support vectors.  

---

### 🔹 **1.3. Margin**  
📌 The **margin** is the distance between the hyperplane and the closest support vectors.  
- **Wider margins** → Better generalization (lower risk of overfitting).  
- **Narrower margins** → Higher risk of misclassification.

### 🔹 **1.4. Maximum Margin Classifier**  
📌 The optimal hyperplane is the one that maximizes the **margin** between the support vectors.  

---

### 🔹 **1.5. Soft Margin vs. Hard Margin**  
- **Hard Margin** → No misclassifications allowed (used for **linearly separable** data).  
- **Soft Margin** → Allows some misclassifications to improve **generalization** (used for **non-linearly separable** data).  

---

## 🔎 **2. Mathematical Foundation of SVM**  

### 🔹 **Objective: Maximize the Margin**  
For a hyperplane defined as:

$
w \cdot x + b = 0
$

Where:  
- $ w $ → Weight vector (normal to the hyperplane)  
- $ x $ → Input features  
- $ b $ → Bias term  

### 🔹 **Optimization Problem:**  
The SVM aims to:

$
\min_{w, b} \frac{1}{2} ||w||^2
$

**Subject to:**  
$
y_i (w \cdot x_i + b) \geq 1
$

Where:  
- $y_i \in \{ -1, 1 \}$ → Class labels  
- **Maximizing margin** is equivalent to **minimizing $ ||w|| $**.  

---

### 🔹 **Soft Margin with Slack Variable (ξ):**  
For non-linearly separable data, SVM introduces a **slack variable $ \xi $** to allow some misclassifications:

$
\min_{w, b} \frac{1}{2} ||w||^2 + C \sum \xi_i
$

Where:  
- $ C $ → **Regularization parameter** (controls margin size and misclassification penalty)  
- Smaller $ C $ → Wider margin, higher tolerance for misclassification (better generalization).  
- Larger $ C $ → Narrower margin, lower tolerance for misclassification (higher accuracy on training data but risk of overfitting).  

---

## 🌐 **3. Kernel Trick in SVM**  

📌 The **kernel trick** allows SVM to efficiently solve **non-linear** classification problems by mapping data into a **higher-dimensional space**.  

### 🔹 **Popular Kernel Functions in SVM**  

| **Kernel** | **Description** | **Equation** | **Best Use Case** |
|--------------|-----------------|---------------|-------------------|
| **Linear Kernel** | Best for **linearly separable data** | $ K(x, y) = x \cdot y $ | Text classification |
| **Polynomial Kernel** | Captures **complex boundaries** | $ K(x, y) = (x \cdot y + 1)^d $ | Image classification |
| **RBF (Radial Basis Function)** | Maps data into **infinite dimensions** | $ K(x, y) = e^{-\gamma \| x - y \|^2} $ | Non-linear patterns |
| **Sigmoid Kernel** | Mimics behavior of **neural networks** | $ K(x, y) = \tanh(x \cdot y + r) $ | Neural network problems |

## 🏹 **4. Implementing SVM in Python**  

### 📁 **Dataset: Iris Flower Classification (3 Classes - Setosa, Versicolor, Virginica)**  

```python
import numpy as np
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Load Dataset
iris = datasets.load_iris()
X = iris.data
y = iris.target

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train SVM Model with RBF Kernel
svm_model = SVC(kernel='rbf', C=1.0, gamma='scale')
svm_model.fit(X_train, y_train)

# Predictions
y_pred = svm_model.predict(X_test)

# Model Evaluation
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

---

## 🔥 **6. Hyperparameter Tuning in SVM**  

### 🔹 **Important Parameters to Tune:**  
✅ **`C` (Regularization Parameter):**  
- Small $ C $ → Wider margin (better generalization).  
- Large $ C $ → Narrower margin (risk of overfitting).  

✅ **`gamma` (RBF Kernel Parameter):**  
- Small $ \gamma $ → Larger influence for each training point.  
- Large $ \gamma $ → Smaller influence for each point (may lead to overfitting).  

✅ **`kernel` (Kernel Function):**  
- Choose between **linear**, **poly**, **rbf**, or **sigmoid**.

### 🔹 **Tuning Example Using GridSearchCV**

```python
from sklearn.model_selection import GridSearchCV

# Hyperparameter Grid
param_grid = {
    'C': [0.1, 1, 10],
    'gamma': ['scale', 0.1, 1],
    'kernel': ['linear', 'rbf', 'poly']
}

# Grid Search with Cross-Validation
grid_search = GridSearchCV(SVC(), param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best Accuracy:", grid_search.best_score_)
```

```python
# import libraries
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    StandardScaler,
    LabelEncoder,
)
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
)
```

```python
# import the data
df = sns.load_dataset("iris")
df.head()
```

### **SVM on Non-Linear Data**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.axes._axes import _log as matplotlib_axes_logger
from mpl_toolkits import mplot3d
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from matplotlib.colors import ListedColormap

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import make_circles
X, y = make_circles(100, factor=.1, noise=.1)

plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='bwr')
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20)
```

- **Normal SVM**

```python
classifier = SVC(kernel="linear")
classifier.fit(X_train, y_train.ravel())
y_pred = classifier.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python

zero_one_colourmap = ListedColormap(('blue', 'red'))
def plot_decision_boundary(X, y, clf):
    X_set, y_set = X, y
    X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, 
                                 stop = X_set[:, 0].max() + 1, 
                                 step = 0.01),
                       np.arange(start = X_set[:, 1].min() - 1, 
                                 stop = X_set[:, 1].max() + 1, 
                                 step = 0.01))
  
    plt.contourf(X1, X2, clf.predict(np.array([X1.ravel(), 
                                             X2.ravel()]).T).reshape(X1.shape),
               alpha = 0.75, 
               cmap = zero_one_colourmap)
    plt.xlim(X1.min(), X1.max())
    plt.ylim(X2.min(), X2.max())
    for i, j in enumerate(np.unique(y_set)):
        plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = (zero_one_colourmap)(i), label = j)
    plt.title('SVM Decision Boundary')
    plt.xlabel('X1')
    plt.ylabel('X2')
    plt.legend()
    return plt.show()
```

```python
plot_decision_boundary(X, y, classifier)
```

```python
def plot_3d_plot(X, y):
    r = np.exp(-(X ** 2).sum(1))
    ax = plt.subplot(projection='3d')
    ax.scatter3D(X[:, 0], X[:, 1], r, c=y, s=100, cmap='bwr')
    ax.set_xlabel('X1')
    ax.set_ylabel('X2')
    ax.set_zlabel('y')
    return ax
```

```python
plot_3d_plot(X,y)
plt.show()
```

- **kernel="rbf" SVM**

```python
rbf_classifier = SVC(kernel="rbf")
rbf_classifier.fit(X_train, y_train)
y_pred = rbf_classifier.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
plot_decision_boundary(X, y, rbf_classifier)
```

- **kernel="poly" SVM**

```python
poly_classifier = SVC(kernel="poly",degree=2)
poly_classifier.fit(X_train, y_train)
y_pred = poly_classifier.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
plot_decision_boundary(X, y, poly_classifier)
```
