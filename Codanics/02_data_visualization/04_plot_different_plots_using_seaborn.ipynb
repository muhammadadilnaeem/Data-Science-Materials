### **Load data**

```python
df = sns.load_dataset("titanic")
```

```python
df.info()
```

### **Scatter Plot**

```python
# set figure size
plt.figure(figsize=(12,7))
sns.scatterplot(df,x='age',y='fare',hue='sex')
plt.title('Scatter plot of age vs fare')
plt.savefig('plots/scatter_plot_of_age_vs_fare.png',dpi=300)
plt.show()
```

```python
# create a boxplot
sns.set_theme(style="darkgrid")
plt.figure(figsize=(12,7))
sns.boxplot(df,x='sex',y='age',palette='Set3',hue='class')
plt.title('Box plot of Gender vs Age')
plt.savefig('plots/box_plot_of_sex_vs_age.png',dpi=300)
plt.show()
```

```python
# create a boxenplot
sns.set_theme(style="darkgrid")
plt.figure(figsize=(12,7))
sns.boxenplot(df,x='sex',y='age',palette='Set3',hue='class')
plt.title('Box plot of Gender vs Age')
plt.savefig('plots/boxen_plot_of_sex_vs_age.png',dpi=300)
plt.show()
```

### **Set Style of Plots**

```python
# set figure size
sns.set_theme(style='darkgrid')
plt.figure(figsize=(12,7))
sns.scatterplot(df,x='age',y='fare',hue='sex')
plt.title('Scatter plot of age vs fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.savefig('plots/scatter_plot_of_age_vs_fare_darkgrid.png',dpi=300)
plt.show()
```

### **Interpret the Results**

- We have `Age` of Passagers on `x-axis` and `Fare` on `y-axis`.
- The `Scatter Plot` shows relationship between age and fare of **titanic** passangers.
- We further grou the passangers based on the colors. `orange` for males and `blue` for females.
- 2 Male passangers payed highest fare accordint to this data,they might travel in `first class` or `urgent` in ship.
- The oldest person is `male` nearly `80` years.
