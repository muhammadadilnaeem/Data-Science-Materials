### **Import Libraries**

```python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

### **Load Dataset**

```python
# load titanic data from seaborn 

titanic = sns.load_dataset('titanic')
titanic.head()
```

```python

titanic.plot(kind='scatter', x='age', y='fare', color='c')
```

```python
# plot using hex code #0302FC


titanic.plot(kind='scatter', x='age', y='fare', color='#0302FC')
```

```python
titanic.groupby('sex')['age'].mean().plot(kind='bar', colormap='Dark2')
plt.ylabel("Average Fare")
plt.xlabel("Sex")
plt.legend()
plt.show()
```

```python
# bar plot using seabron

sns.barplot(data=titanic,x="sex",y='fare',color='r',hue='class')
```

```python
# box plot using seabron

sns.boxplot(data=titanic,x="sex",y='age',color='r',hue='class')
```

```python
# scatter plot using seabron

sns.scatterplot(data=titanic,x="fare",y='age',color='r',hue='sex')
```

```python
# make scatter plot using plotly

fig = px.scatter(titanic,x='age',y='fare',color='class')
fig.show()
```

```python
# animated plots using plotly

import plotly.express as px
df = px.data.gapminder()
px.scatter(df, x="gdpPercap", y="lifeExp", animation_frame="year", animation_group="country",
           size="pop", color="continent", hover_name="country",
           log_x=True, size_max=55, range_x=[100,100000], range_y=[25,90])
```
