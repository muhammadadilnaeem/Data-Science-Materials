## **Data Visualization Plots Using Plotly**

---------



### **Import Libraris**

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

### **Line Plot**

```python
fig= px.line(x=["Adil","Rauf","Zulqarnain","Zain"], y=[20,42,33,91], title="Friends Plot", labels={"x":"Name","y":"Age"})
fig.show()
```

### **Bar Plot**

```python
fig= px.bar(x=["Adil","Rauf","Zulqarnain","Zain"], y=[20,42,33,91], title="Friends Plot", labels={"x":"Name","y":"Age"},color=["Adil","Rauf","Zulqarnain","Zain"])
fig.write_image("plots/Friends.png")
fig.show()
```

### **Plots on Titanic Dataset**

```python
df = sns.load_dataset("titanic")
```

#### **Scatter plot**

```python
fig = px.scatter(df, x="age", y="fare", color='sex', title="Titanic Plot with Age and Fare using plotly")
fig.show()
```

```python
#df.dropna(inplace=True)
fig = px.scatter(df, x="age", y="fare",color='sex',size='fare', title="Titanic Plot with Age and Fare using plotly")
fig.write_image("plots/Titanic bubble.png")
fig.show()
```

### **Heatmap using Plotly**

```python
# heatmap using plotly go object
data = np.random.rand(10,10)
fig = px.imshow(data, title="Heatmap using plotly go object")
fig.show()
```

### **Heatmap on Titanic Data**

```python
# Select only numeric columns
numeric_df = df.select_dtypes(include=[np.number])

# Compute the correlation matrix
corr_matrix = numeric_df.corr()

# Create the heatmap
fig = px.imshow(corr_matrix, title="Heatmap of Titanic Data using Plotly")
fig.write_image("plots/titanic_heatmap.png")
fig.show()
```

### **Iris Data Heatmap**

```python
df = sns.load_dataset("iris")

# Select only numeric columns
numeric_df = df.select_dtypes(include=[np.number])

# Compute the correlation matrix
corr_matrix = numeric_df.corr()

# Create the heatmap
fig = px.imshow(corr_matrix, title="Heatmap of iris Data using Plotly")
fig.write_image("plots/iris_heatmap.png")
fig.show()
```
