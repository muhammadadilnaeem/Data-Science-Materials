## **15 Plot's Using Plotly**

----------



### **Import Libraries**

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

### **Load Dataset**

```python
df = sns.load_dataset('iris')
```

### **1. Scatter Plot**

```python
fig = px.scatter(df, x='sepal_length', y='sepal_width', color='species',title='Iris Dataset - Sepal Length vs Sepal Width')
fig.write_image('plots/iris_sepal_length_vs_sepal_width.png')
fig.show()
```

### **2. Line Plot**

```python
fig = px.line(df, x='sepal_length', y='sepal_width',color='species', title='Iris Dataset - Sepal Length')
fig.write_image('plots/iris_sepal_length_line_plot.png')
fig.show()
```

### **3. Bar Plot**

```python
df = sns.load_dataset('iris')

df_bar = df.groupby('species').mean().reset_index()
fig = px.bar(df_bar, x='species', y=['sepal_length','sepal_width','petal_length','petal_width'], title='Iris Dataset - Mean Values')
fig.write_image('plots/iris_mean_values_bar.png')
fig.show()
```

### **4. Box Plot**

```python
df = sns.load_dataset('iris')

df_bar = df.groupby('species').mean().reset_index()
fig = px.box(df_bar, x='species', y=['sepal_length','sepal_width','petal_length','petal_width'], color='species',title='Iris Dataset - Mean Values')
fig.write_image('plots/iris_mean_values_box.png')
fig.show()
```

### **5. Voilin Plot**

```python
df = sns.load_dataset('iris')

fig = px.violin(df_bar, x='species', y=['sepal_length','sepal_width','petal_length','petal_width'], color='species',title='Iris Dataset - Voilin Plot',box=True, points="all")
fig.write_image('plots/iris_mean_values_voilin.png')
fig.show()
```

### **6. Histogram**

```python
df = sns.load_dataset('iris')

fig = px.histogram(df_bar, x='species', y=['sepal_length','sepal_width','petal_length','petal_width'], color='species',title='Iris Dataset - Histogram')
fig.write_image('plots/iris_histogram.png')
fig.show()
```

### **7. Piechart**

```python
df = sns.load_dataset('iris')

fig = px.pie(df,values='sepal_width',names='species',title='Iris Dataset - Piechart')
fig.write_image('plots/iris_pie.png')
fig.show()
```

### **8. 3D Scatter Plot**

```python
df = sns.load_dataset('iris')

fig = px.scatter_3d(df, x='sepal_length', y='sepal_width', z='petal_length', color='species',title='Iris Dataset - 3D Scatter Plot')
fig.write_image('plots/iris_scatter_3d.png')
fig.show()
```

### **9. Area Plot**

```python
# area plot on iris dataset using plotly

df = sns.load_dataset('iris')

df_area = df.groupby('species').mean().reset_index()    

fig = px.area(df_area, x='species', y=['sepal_length','sepal_width','petal_length','petal_width'],title='Iris Dataset - Area Plot')
fig.write_image('plots/iris_area_plot.png')
fig.show()
```

### **10. Bubble Chart**

```python
# bubble plot on iris dataset using plotly

df = sns.load_dataset('iris')

fig = px.scatter(df, x='sepal_length', y='sepal_width', size='petal_length', color='species',title='Iris Dataset - Bubble Plot')
fig.write_image('plots/iris_bubble_plot.png')
fig.show()
```

### **11. Sun Bunburst Chart**

```python
# subrbruet plot on iris dataset using plotly

df = sns.load_dataset('iris')

fig = px.sunburst(df, path=['species','sepal_length','sepal_width','petal_length','petal_width'],title='Iris Dataset - Sunburst Plot')

fig.write_image('plots/iris_sunburst_plot.png')

fig.show()
```

### **12. Parallel Coordicate**

```python
# Parallel Coordinate Plot on iris dataset using plotly

df = sns.load_dataset('iris')

# Map species to numeric values
species_mapping = {'setosa': 1, 'versicolor': 2, 'virginica': 3}
df['species_num'] = df['species'].map(species_mapping)

fig = px.parallel_coordinates(df, color='species_num', title='Iris Dataset - Parallel Coordinate Plot',
							  color_continuous_scale=px.colors.diverging.Tealrose)

fig.write_image('plots/iris_parallel_coordinate_plot.png')

fig.show()
```

### **12. Polar Coordicate**

```python
# radar plot on iris dataset using plotly

df = sns.load_dataset('iris')

df_radar = df.groupby('species').mean().reset_index()

# Reshape the dataframe for radar plot
df_radar_melted = df_radar.melt(id_vars=['species'], var_name='measurement', value_name='value')

fig = px.line_polar(df_radar_melted, r='value', theta='measurement', color='species', line_close=True, title='Iris Dataset - Radar Plot')

fig.write_image('plots/iris_radar_plot.png')

fig.show()
```
