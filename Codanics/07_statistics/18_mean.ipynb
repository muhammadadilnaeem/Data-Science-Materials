# 📊 **`Mean in Statistics for Data Science`** 🔍  

## 🔹 **What is Mean?**  

✅ **Definition**: The **mean** (or **arithmetic average**) is the **sum of all values divided by the number of values** in a dataset.  

✅ **Formula**:  
$$
\text{Mean} (\bar{X}) = \frac{\sum X_i}{n}
$$
where:  
- $X_i$ = Each data point  
- $n$ = Number of data points  

✅ **Why is Mean Important?**  
✔ **Summarizes large datasets** in a single value.  
✔ **Used in machine learning** for feature scaling.  
✔ **Essential in statistical modeling** for trend analysis.  



---

## 🔹 **1. Types of Mean in Statistics**  

| **Type** | **Formula** | **Usage** |
|---------|------------|-------------|
| **Arithmetic Mean** ➗ | $$\bar{X} = \frac{\sum X_i}{n}$$ | Basic average (default mean). |
| **Weighted Mean** ⚖️ | $$\bar{X} = \frac{\sum (X_i \cdot w_i)}{\sum w_i}$$ | Used when some values are more important than others. |
| **Geometric Mean** 📏 | $$\bar{X}_g = \sqrt[n]{X_1 \times X_2 \times ... \times X_n}$$ | Used in **growth rates** (e.g., stock prices). |
| **Harmonic Mean** 🎵 | $$\bar{X}_h = \frac{n}{\sum \frac{1}{X_i}}$$ | Used in **ratios & rates** (e.g., speed calculations). |



---

## 🔹 **2. Arithmetic Mean (Simple Average) ➗**  

✅ **Definition**: The sum of all values divided by the number of values.  
✅ **Example Calculation**:  
📌 Data: **[10, 20, 30, 40, 50]**  
$$
\text{Mean} = \frac{10 + 20 + 30 + 40 + 50}{5} = 30
$$

📌 **Python Code for Mean Calculation**  

```python

import numpy as np

data = [10, 20, 30, 40, 50]
mean_value = np.mean(data)
print("Mean:", mean_value)
```


📊 **Output**: `Mean: 30`  

✅ **When to Use?**  
✔ Best for **normal distributions**.  
❌ **Not good for skewed data** (affected by outliers).  

📌 **Example Where Mean is Misleading**:  
Data: **[10, 20, 30, 40, 1000]** (Outlier: **1000**)  
$$
\text{Mean} = \frac{10 + 20 + 30 + 40 + 1000}{5} = 220
$$
🔥 **The mean is pulled up by the outlier!**  


---

## 🔹 **3. Weighted Mean (Handling Different Weights) ⚖️**  

✅ **Definition**: The mean where **some values are more important than others**.  
✅ **Formula**:  
$$
\bar{X} = \frac{\sum (X_i \cdot w_i)}{\sum w_i}
$$
where:  
- $X_i$ = Data points  
- $w_i$ = Weights  

✅ **Example Calculation**:  
📌 Exam scores: **Assignments (30%), Midterm (30%), Final (40%)**  

$$
\text{Weighted Mean} = \frac{(80 \times 0.3) + (85 \times 0.3) + (90 \times 0.4)}{0.3 + 0.3 + 0.4} = 85.5
$$

📌 **Python Code for Weighted Mean**  


```python

import numpy as np

grades = [80, 85, 90]
weights = [0.3, 0.3, 0.4]
weighted_mean = np.average(grades, weights=weights)
print("Weighted Mean:", weighted_mean)
```


📊 **Output**: `Weighted Mean: 85.5`  

✅ **When to Use?**  
✔ When **some values contribute more** than others (e.g., **weighted grades, stock portfolios**).  





---

## 🔹 **4. Geometric Mean (Used for Growth Rates) 📏**  

✅ **Definition**: The **n-th root of the product** of values.  
✅ **Formula**:  
$$
\bar{X}_g = \sqrt[n]{X_1 \times X_2 \times ... \times X_n}
$$

✅ **Example Calculation**:  
📌 Growth rates: **[1.1, 1.2, 1.3, 1.4, 1.5]**  
$$
\text{Geometric Mean} = \sqrt[5]{1.1 \times 1.2 \times 1.3 \times 1.4 \times 1.5} = 1.278
$$

📌 **Python Code for Geometric Mean**  


```python

from scipy.stats.mstats import gmean

growth_rates = [1.1, 1.2, 1.3, 1.4, 1.5]
geo_mean = gmean(growth_rates)
print("Geometric Mean:", geo_mean)
```


📊 **Output**: `Geometric Mean: 1.278`  

✅ **When to Use?**  
✔ **Stock market returns** & **growth rate calculations**.  
✔ **More accurate for percent changes over time**.  



---

## 🔹 **5. Harmonic Mean (Used for Rates & Ratios) 🎵**  

✅ **Definition**: The reciprocal of the average of reciprocals.  
✅ **Formula**:  

$$
\bar{X}_h = \frac{n}{\sum \frac{1}{X_i}}
$$

✅ **Example Calculation**:  
📌 Average speed of two trips: **60 km/h & 40 km/h**  

$$
\text{Harmonic Mean} = \frac{2}{\left(\frac{1}{60} + \frac{1}{40}\right)} = 48 \text{ km/h}
$$

📌 **Python Code for Harmonic Mean**  


```python

from scipy.stats import hmean

speeds = [60, 40]
harmonic_mean = hmean(speeds)
print("Harmonic Mean:", harmonic_mean)
```


📊 **Output**: `Harmonic Mean: 48`  

✅ **When to Use?**  
✔ **Averaging rates, speeds, and ratios**.  
✔ Used in **finance (PE Ratios) and physics (resistance in parallel circuits)**

---

## 🔹 **6. Comparison of Mean Types**  

| **Type** | **Formula** | **Best For** | **Example** |
|---------|------------|-------------|-------------|
| **Arithmetic Mean** ➗ | $$\frac{\sum X_i}{n}$$ | General averaging | **Exam scores, salaries** |
| **Weighted Mean** ⚖️ | $$\frac{\sum (X_i \cdot w_i)}{\sum w_i}$$ | Different weights | **Stock portfolio returns** |
| **Geometric Mean** 📏 | $$\sqrt[n]{X_1 \times X_2 \times ... \times X_n}$$ | Growth rates | **Stock market, Population growth** |
| **Harmonic Mean** 🎵 | $$\frac{n}{\sum \frac{1}{X_i}}$$ | Averages of rates | **Speed calculations, PE Ratios** |
