# 📊 **`Levene's Test in Statistics for Data Science`** 🔍  

## 🔹 **What is Levene's Test?**  

✅ **Definition**: **Levene’s Test** is a statistical test used to **check if two or more groups have equal variances** (**homogeneity of variance**).  
✅ **Why is Levene’s Test Important?**  
✔ Many statistical tests (t-tests, ANOVA) **assume equal variances**.  
✔ Helps **decide whether to use a standard t-test or Welch’s t-test**.  
✔ **Essential in A/B testing, clinical trials, and finance.**  

📌 **Example Questions**:  
- "Do male and female employees have similar salary variability?"  
- "Do different ad campaigns have the same variance in click-through rates?"  
- "Is the variance in test scores similar across different schools?"  

---

# 🔥 **1. Key Concepts in Levene’s Test**  

📌 **Levene’s Test Hypotheses**  

| **Hypothesis** | **Interpretation** |
|--------------|------------------|
| **Null Hypothesis (H₀)** ❌ | "The variances in the groups are equal (homogeneous)." |
| **Alternative Hypothesis (H₁)** ✅ | "The variances in the groups are NOT equal (heterogeneous)." |

✅ **If p > 0.05** → **Fail to reject H₀** → Variances are equal ✅.  
✅ **If p < 0.05** → **Reject H₀** → Variances are NOT equal ❌.  

---

# 📊 **2. Levene’s Test in Python**  

### **Example: Checking Salary Variance Between Male & Female Employees**  
📌 **Scenario**: We have **salary data** for 30 male and 30 female employees. We check if their **variance is equal** before performing a t-test.  

📌 **Step 1: Generate Sample Data & Perform Levene’s Test**  

```python

import numpy as np
import scipy.stats as stats

# Simulated salary data
male_salaries = np.random.normal(loc=55000, scale=5000, size=30)  # Mean=55,000, Std Dev=5,000
female_salaries = np.random.normal(loc=53000, scale=7000, size=30)  # Mean=53,000, Std Dev=7,000

# Perform Levene’s Test
stat, p_value = stats.levene(male_salaries, female_salaries)

print(f"\nLevene’s Test Statistic: {stat:.3f}, p-value: {p_value:.5f}")

# Decision
if p_value > 0.05:
    print("\nFail to reject H₀: Variances are equal.")
else:
    print("\nReject H₀: Variances are NOT equal.")
```


📊 **Output**:  
✔ **p > 0.05** → Variances are **equal** ✅ (Use regular t-test).  
✔ **p < 0.05** → Variances are **not equal** ❌ (Use Welch’s t-test).  

✅ **When to Use?**  
✔ **Before running a t-test or ANOVA** to check for equal variance.  
✔ **In A/B testing, checking if groups have different spread of values.**

# 🔍 **3. Why is Levene’s Test Important?**  

📌 **Many statistical tests assume equal variance:**  
- **Standard t-Test (Student's t-Test)** assumes **equal variances**.  
- **Welch’s t-Test** is used when **variances are unequal**.  

✅ **Levene’s Test Helps Decide:**  
- ✔ If **p > 0.05**, use **regular t-test** (equal variance).  
- ✔ If **p < 0.05**, use **Welch’s t-test** (unequal variance).  

📌 **Example: Running Welch’s t-Test if Variances Are Unequal**

```python

# Perform Welch’s t-test (does NOT assume equal variances)
t_stat, p_value = stats.ttest_ind(male_salaries, female_salaries, equal_var=False)

print(f"\nWelch’s t-Test Statistic: {t_stat:.3f}, \n\np-value: {p_value:.5f}")
```


📊 **Output**:  
- If **p < 0.05** → Salaries are **significantly different**.  
- If **p > 0.05** → No strong evidence of a salary difference.

# 📉 **4. Levene’s Test vs. Bartlett’s Test**  

| **Feature** | **Levene’s Test** ✅ | **Bartlett’s Test** ❌ |
|------------|-----------------|-----------------|
| **Best for** | Normally & non-normally distributed data | Normally distributed data only |
| **More robust?** | ✅ Yes | ❌ No |
| **Recommended for Data Science?** | ✅ Yes | ❌ No |

✅ **Levene’s Test is better because it works for both normal & non-normal data.**

# 🎯 **6. Summary: Key Takeaways on Levene’s Test**  

| **Feature** | **Levene’s Test** |
|------------|-----------------|
| **Purpose** | Checks if groups have equal variances |
| **H₀ (Null Hypothesis)** | Variances are equal |
| **H₁ (Alternative Hypothesis)** | Variances are NOT equal |
| **Decision Rule** | **p > 0.05** → Equal variances ✅<br> **p < 0.05** → Unequal variances ❌ |
| **Best Used Before?** | **t-Test, ANOVA, Regression** |
| **Alternative Test?** | Bartlett’s Test (only for normal data) |

- ✅ **Always check variance equality before running t-tests & ANOVA!**
