# 📊 **`Data Analysis & Types of Data Analysis in Data Science`** 🔍  

## 🔹 **What is Data Analysis?**  

✅ **Definition**: **Data analysis** is the process of **examining, cleaning, transforming, and interpreting data** to extract useful insights, support decision-making, and improve predictions.  

📌 **Why is Data Analysis Important?**  
✔ **Identifies patterns & trends** 📈.  
✔ **Improves decision-making** 🤖.  
✔ **Optimizes business & scientific research** 💡.  

---

## 🔹 **1. Types of Data Analysis in Data Science**  

| **Type of Analysis** | **Purpose** | **Example** |
|--------------------|------------|-------------|
| **Descriptive Analysis** 📊 | Summarizes historical data | "What happened?" (Sales revenue trend over time) |
| **Diagnostic Analysis** 🕵️‍♂️ | Finds causes of patterns | "Why did it happen?" (Drop in customer engagement) |
| **Predictive Analysis** 🔮 | Forecasts future trends | "What will happen?" (Stock price prediction) |
| **Prescriptive Analysis** 🎯 | Suggests actions for outcomes | "What should we do?" (Recommending ad strategies) |
| **Exploratory Data Analysis (EDA)** 🔍 | Finds hidden patterns & relationships | "What insights can we extract?" (Customer segmentation) |
| **Inferential Analysis** 📉 | Draws conclusions about populations | "How does this apply generally?" (A/B testing results) |

---

## 🟢 **1. Descriptive Data Analysis 📊**  

✅ **Definition**: Describes and summarizes historical data using **statistics & visualizations**.  
✅ **Key Metrics**:  
✔ **Measures of Central Tendency** (Mean, Median, Mode).  
✔ **Measures of Dispersion** (Variance, Standard Deviation, Range).  
✔ **Data Visualization** (Bar Charts, Pie Charts, Histograms).  



📌 **Example: Analyzing Sales Data in Python**  


```python

import pandas as pd

# Simulated Sales Data
data = pd.DataFrame({'Sales': [200, 250, 300, 150, 400, 350]})

# Summary Statistics
data.describe()  # Mean, Min, Max, Std Dev
```


📊 **Output**: Provides **mean sales, min/max, standard deviation, etc.**  

✅ **Best Use Cases**:  
✔ Summarizing customer demographics.  
✔ Understanding historical product sales.  
✔ Analyzing website traffic trends.  


---

## 🔵 **2. Diagnostic Data Analysis 🕵️‍♂️**  

✅ **Definition**: Identifies **causes & relationships** in data.  
✅ **Key Techniques**:  
✔ **Correlation Analysis** (How variables relate).  
✔ **Regression Analysis** (Cause-effect relationships).  
✔ **Hypothesis Testing** (A/B testing).  


📌 **Example: Correlation Between Ad Spend & Sales**  


```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Simulated data
ad_spend = [100, 200, 300, 400, 500]
sales = [150, 220, 300, 380, 470]

# Scatter plot to visualize relationship with modified colors and additional details
sns.regplot(x=ad_spend, y=sales, scatter_kws={'color': 'red'}, line_kws={'color': 'green'})

# Adding titles and labels
plt.title("Ad Spend vs Sales")
plt.xlabel("Ad Spend ($)")
plt.ylabel("Sales ($)")
plt.legend(['Data Points','Regression Line'], loc='upper left')

# Show plot
plt.show()
```



📊 **Output**: A **positive correlation** between ad spend and sales.  

✅ **Best Use Cases**:  
✔ Identifying why customer churn increased.  
✔ Diagnosing reasons for website traffic drops.  
✔ Understanding factors affecting employee productivity.  



---

## 🔴 **3. Predictive Data Analysis 🔮**  

✅ **Definition**: Uses **historical data & machine learning models** to **predict future trends**.  
✅ **Key Techniques**:  
✔ **Time Series Forecasting** (Predict future sales).  
✔ **Regression Models** (Predict housing prices).  
✔ **Classification Models** (Fraud detection).  



📌 **Example: Predicting Sales Using Linear Regression**  


```python

import numpy as np
from sklearn.linear_model import LinearRegression

# Simulated Data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Months
y = np.array([200, 250, 300, 350, 400])  # Sales

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict next month's sales
predicted_sales = model.predict([[6]])
print(f"\nPredicted Sales for Month 6: {predicted_sales[0]}")
```


📊 **Output**: Predicts sales for **Month 6** based on past data.  

✅ **Best Use Cases**:  
✔ Stock market & sales forecasting.  
✔ Predicting customer churn.  
✔ Demand planning in supply chain management. 


---

## 🟠 **4. Prescriptive Data Analysis 🎯**  

✅ **Definition**: Provides **actionable recommendations** based on predictive insights.  
✅ **Key Techniques**:  
✔ **Optimization Algorithms** (Choosing best marketing strategy).  
✔ **Decision Trees & AI Models** (Automated decision-making).  


📌 **Example: Recommending a Marketing Strategy Using Decision Trees**  


```python

import numpy as np
from sklearn.tree import DecisionTreeClassifier

# Simulated Data
features = np.array([[100, 1], [200, 1], [300, 0], [400, 1], [500, 0]])  # Ad Spend, Discount Offered
outcomes = np.array([1, 1, 0, 1, 0])  # 1 = High Sales, 0 = Low Sales

# Train model
model = DecisionTreeClassifier()
model.fit(features, outcomes)

# Predict best strategy
new_ad_spend = [[350, 1]]  # New campaign with discount
recommendation = model.predict(new_ad_spend)
print("\nRecommended Strategy:", "High Sales Expected 😎" if recommendation[0] == 1 else "Low Sales Expected 😥")
  
```


📊 **Output**: Suggests if the **new campaign will boost sales**.  

✅ **Best Use Cases**:  
✔ Personalized customer recommendations (Netflix, Amazon).  
✔ Optimizing inventory management.  
✔ Fraud prevention in banking.

## 🟡 **5. Exploratory Data Analysis (EDA) 🔍**  

✅ **Definition**: **Explores & visualizes** hidden patterns in data before applying models.  
✅ **Key Techniques**:  
✔ **Pair Plots & Heatmaps** (Feature relationships).  
✔ **Box Plots** (Detecting outliers).  
✔ **Clustering & Dimensionality Reduction** (Identifying groups in data).  


📌 **Example: Heatmap for Feature Correlations**  


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Simulated dataset
data = pd.DataFrame({'Sales': np.random.randint(100, 500, 10),
                     'Ad Spend': np.random.randint(50, 300, 10),
                     'Customer Visits': np.random.randint(200, 1000, 10)})

# Heatmap of correlations
sns.heatmap(data.corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.show()
```

📊 **Output**: Shows **relationships between features**.  

✅ **Best Use Cases**:  
✔ Understanding **customer behavior**.  
✔ Detecting **data anomalies & outliers**.  
✔ Feature selection for **machine learning models**.  



---

## 🟣 **6. Inferential Analysis (Statistical Testing) 📉**  

✅ **Definition**: Uses **samples** to make **generalizations** about a **population**.  
✅ **Key Techniques**:  
✔ **Hypothesis Testing** (A/B Testing).  
✔ **t-Tests & ANOVA** (Comparing means).  
✔ **Chi-Square Test** (Categorical relationships).  


📌 **Example: A/B Testing for Website Performance**  


```python

import scipy.stats as stats

# Simulated Data: Conversion Rates for A/B Testing
group_A = [20, 22, 19, 24, 23]
group_B = [25, 28, 27, 26, 29]

# t-Test
t_stat, p_value = stats.ttest_ind(group_A, group_B)
print(f"\nT-Statistic: {t_stat}") 
print(f"\nP-Value: {p_value}")
```


📊 **Output**: If **p-value < 0.05**, Version B **significantly outperforms** Version A.  

✅ **Best Use Cases**:  
✔ Comparing **product effectiveness**.  
✔ Evaluating **campaign performance**.  
✔ Validating **scientific hypotheses**.
