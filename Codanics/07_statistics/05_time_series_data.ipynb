# ⏳ **Time Series Data in Statistics for Data Science** 📊  

Time series data plays a crucial role in **forecasting, anomaly detection, financial modeling, and predictive analytics**. It represents data points collected **over time at regular intervals**. Let’s dive deep into **time series concepts, analysis methods, and real-world applications**! 🚀  

---

## 🔹 **1. What is Time Series Data?**  
✅ **Definition**: Time series data is a sequence of observations recorded **chronologically over time**. Each data point is **timestamped** and arranged in **time order**.  

✅ **Key Characteristics**:  
✔ **Time-ordered observations** ⏳.  
✔ **Regular intervals** (hourly, daily, weekly, etc.) 📅.  
✔ **Can be continuous or discrete**.  
✔ **Has temporal dependencies** (previous values influence future values).  

✅ **Examples of Time Series Data**:  
📈 **Stock Market Prices** (daily closing price of Apple stock).  
🌡 **Weather Data** (hourly temperature readings).  
💰 **Sales Revenue** (monthly sales of a retail store).  
🚗 **Traffic Flow** (number of vehicles passing per minute).  
💓 **Heart Rate Monitoring** (beats per second in a medical device).  

📌 **Example: Sample Time Series Data (Pandas DataFrame)**  
```python
import pandas as pd

data = {
    'Date': pd.date_range(start='2024-01-01', periods=5, freq='D'),
    'Sales': [100, 120, 150, 170, 160]
}
df = pd.DataFrame(data)
print(df)
```
📅 **Output**:  
| Date       | Sales |
|------------|------|
| 2024-01-01 | 100  |
| 2024-01-02 | 120  |
| 2024-01-03 | 150  |
| 2024-01-04 | 170  |
| 2024-01-05 | 160  |

## 🔹 **2. Components of Time Series Data 📊**  

Time series data can be broken down into **four main components**:  

### 📌 **1. Trend (T)**  
📈 **Definition**: Long-term increase or decrease in the data over time.  
📉 **Example**: The **global average temperature** is rising over decades.  

### 📌 **2. Seasonality (S) 🎯**  
📆 **Definition**: Repeating patterns at fixed time intervals (daily, weekly, yearly).  
🔁 **Example**: **E-commerce sales spike during holidays** (Black Friday, Christmas).  

### 📌 **3. Cyclic Variations (C) 🔄**  
🔄 **Definition**: Recurring **long-term fluctuations** that are not fixed like seasonality.  
💰 **Example**: **Economic cycles (recessions, booms)** that last several years.  

### 📌 **4. Irregular (Random) Component (R) 🎲**  
🎲 **Definition**: Unexpected, unpredictable fluctuations due to noise or external factors.  
💥 **Example**: **Stock market crash due to a financial crisis**.  

📌 **Time Series Decomposition Equation**:  
- **Additive Model**: $$ Y(t) = T(t) + S(t) + C(t) + R(t) $$  
- **Multiplicative Model**: $$ Y(t) = T(t) \times S(t) \times C(t) \times R(t) $$  

📌 **Example (Python Code to Decompose Time Series)**  
```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from statsmodels.tsa.seasonal import seasonal_decompose

# Simulating time series data
np.random.seed(42)
date_rng = pd.date_range(start='2023-01-01', periods=100, freq='D')
trend = np.linspace(50, 150, 100)  # Increasing trend
seasonality = 10 * np.sin(np.linspace(0, 20, 100))  # Seasonal pattern
noise = np.random.normal(scale=5, size=100)  # Random fluctuations
sales_data = trend + seasonality + noise

# Creating DataFrame
df = pd.DataFrame({'Date': date_rng, 'Sales': sales_data})
df.set_index('Date', inplace=True)

# Time Series Decomposition
decomposition = seasonal_decompose(df['Sales'], model='additive', period=30)
decomposition.plot()
plt.show()
```

## 🔹 **3. Stationarity in Time Series Data 📉**  

📌 **Definition**: A time series is **stationary** if its **statistical properties** (mean, variance, autocorrelation) remain **constant over time**.  

✅ **Why Stationarity is Important?**  
✔ Many statistical models (ARIMA, SARIMA) require **stationary data**.  
✔ Non-stationary data can lead to **unreliable forecasts**.  

📌 **How to Check for Stationarity?**  
1️⃣ **Plot the Time Series** – If the mean or variance changes over time, it is **non-stationary**.  
2️⃣ **Augmented Dickey-Fuller (ADF) Test** – A **hypothesis test** to check stationarity.  

📌 **Example (Python Code to Check Stationarity with ADF Test)**  
```python
from statsmodels.tsa.stattools import adfuller

# Apply ADF Test
adf_result = adfuller(df['Sales'])
print(f'ADF Statistic: {adf_result[0]}')
print(f'p-value: {adf_result[1]}')

if adf_result[1] < 0.05:
    print("Time Series is Stationary")
else:
    print("Time Series is Non-Stationary")
```

📌 **How to Make Data Stationary?**  
✔ **Differencing** – Subtracting previous observations $$ Y_t - Y_{t-1} $$  
✔ **Log Transformation** – $$ Y' = \log(Y) $$ to reduce trend.  
✔ **Seasonal Differencing** – $$ Y_t - Y_{t-12} $$ to remove seasonal effects.

## 🔹 **4. Time Series Forecasting Methods 🔮**  

### 📌 **1. Moving Averages (SMA & EMA) 📉**  
✅ **Simple Moving Average (SMA)**:  
$$ SMA_t = \frac{Y_{t-1} + Y_{t-2} + ... + Y_{t-n}}{n} $$  

✅ **Exponential Moving Average (EMA)**:  
$$ EMA_t = \alpha Y_t + (1 - \alpha) EMA_{t-1} $$  
📌 Used in **trend smoothing & financial analysis**.  

### 📌 **2. ARIMA (AutoRegressive Integrated Moving Average) 🔍**  
📈 Used for **stationary** time series forecasting.  
📌 **ARIMA(p, d, q) Model**:  
- **p** → Number of autoregressive (AR) terms.  
- **d** → Differencing order to make data stationary.  
- **q** → Number of moving average (MA) terms.  

📌 **Example (Python Code for ARIMA Forecasting)**  
```python
from statsmodels.tsa.arima.model import ARIMA

# Fit ARIMA model
model = ARIMA(df['Sales'], order=(2, 1, 2))
model_fit = model.fit()
print(model_fit.summary())

# Forecast
forecast = model_fit.forecast(steps=10)
print(forecast)
```

### 📌 **3. Seasonal ARIMA (SARIMA) for Seasonal Data 🎯**  
Extends ARIMA by adding **seasonal components**:  
$$ SARIMA(p, d, q) \times (P, D, Q, s) $$  
where **s** is the seasonal period.  

### 📌 **4. LSTMs & Deep Learning for Time Series 🤖**  
📌 **Recurrent Neural Networks (RNNs) & Long Short-Term Memory (LSTM) Networks** handle **complex patterns in sequential data**.
