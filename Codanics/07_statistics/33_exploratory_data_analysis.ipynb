# 📊 **`Exploratory Data Analysis (EDA) in Statistics for Data Science`** 🔍  

## 🔹 **What is Exploratory Data Analysis (EDA)?**  

✅ **Definition**: **Exploratory Data Analysis (EDA)** is the **process of analyzing datasets** to **summarize key characteristics, detect patterns, and identify outliers** before applying statistical models or machine learning algorithms.  

✅ **Why is EDA Important?**  
- ✔ **Detects data quality issues (missing values, duplicates, outliers)** 🛑.  
- ✔ **Identifies patterns & trends** 📈.  
- ✔ **Provides insights using visualizations** 📊.  
- ✔ **Ensures the right data is used for model training** 🤖.  

📌 **Key Steps in EDA**  
- 1️⃣ **Understand Dataset Structure** → Columns, data types, missing values.  
- 2️⃣ **Summary Statistics** → Mean, median, standard deviation, skewness, etc.  
- 3️⃣ **Visualizations** → Histograms, box plots, scatter plots, correlation heatmaps.  
- 4️⃣ **Outlier & Missing Value Detection** → Handling anomalies before modeling.

# 📂 **1. Loading & Understanding the Dataset**  

📌 **Python Code for Dataset Overview**  
```python
import pandas as pd

# Load dataset
df = pd.read_csv("data.csv")  # Replace with actual dataset file

# Display first 5 rows
print(df.head())

# Check basic info
print(df.info())

# Summary statistics
print(df.describe())
```
📊 **Output**:  
- ✔ **Shows dataset structure, column types, missing values, and statistical summary.**  

✅ **What to Look For?**  
- ✔ **Data types** → Are they numerical, categorical, or mixed?  
- ✔ **Missing values** → Need imputation or removal?  
- ✔ **Duplicates** → Need deduplication?

# 📊 **2. Summary Statistics: Understanding Data Distribution**  

📌 **Python Code for Summary Statistics**  
```python
# Summary statistics for numerical columns
print(df.describe())

# Checking for missing values
print(df.isnull().sum())

# Checking for duplicate rows
print("Duplicate Rows:", df.duplicated().sum())
```
📊 **Output**:  
- ✔ **Mean, Median, Standard Deviation, Min, Max values for each numerical feature.**  
- ✔ **Number of missing values & duplicates.**  

✅ **Key Insights from Summary Statistics**  
- ✔ **If Mean ≠ Median** → Data might be skewed.  
- ✔ **High Standard Deviation** → Large variability in data.  
- ✔ **Min & Max values** → Check for extreme values (possible outliers).

# 📉 **3. Data Visualization for Insights**  

## **📌 3.1 Histogram: Understanding Data Distribution**  
```python
import matplotlib.pyplot as plt

# Histogram of a numerical column
df['column_name'].hist(bins=30, color='blue', edgecolor='black')
plt.title("Histogram of column_name")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
📊 **Output**: Shows how values are **distributed (normal, skewed, or multimodal).**  

✅ **Interpretation**  
- ✔ **Bell-shaped curve** → Normal distribution ✅.  
- ✔ **Right-skewed (Positive Skew)** → High values are more frequent 📈.  
- ✔ **Left-skewed (Negative Skew)** → Low values are more frequent 📉.

## **📌 3.2 Box Plot: Detecting Outliers**  
```python
# Boxplot for outlier detection
df.boxplot(column=['column_name'])
plt.title("Box Plot of column_name")
plt.show()
```
📊 **Output**: Identifies **outliers as dots beyond whiskers.**  

✅ **How to Handle Outliers?**

## **📌 3.3 Correlation Heatmap: Identifying Relationships**  
```python
import seaborn as sns

# Compute correlation matrix
corr_matrix = df.corr()

# Create heatmap
plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
📊 **Output**: Shows **relationships between numerical variables**.  

✅ **Interpretation**  
- ✔ **Correlation close to +1 or -1** → Strong positive/negative correlation.  
- ✔ **Correlation near 0** → No significant relationship.  
- ✔ **High correlation between two features** → Consider removing one to avoid redundancy.

# 🚨 **4. Handling Missing Data**  

📌 **Checking for Missing Values**  
```python
print(df.isnull().sum())
```
📊 **Output**: Shows count of missing values per column.  

✅ **Strategies to Handle Missing Data**  
- ✔ **Drop missing values** (if missing % is small).  
- ✔ **Fill with Mean/Median (for numerical data).**  
- ✔ **Fill with Mode (for categorical data).**  

📌 **Python Code for Imputation**  
```python
# Fill numerical columns with median
df['num_column'].fillna(df['num_column'].median(), inplace=True)

# Fill categorical column with mode
df['cat_column'].fillna(df['cat_column'].mode()[0], inplace=True)
```
📊 **Output**: Missing values replaced with **median/mode**.  

---

# ⚡ **5. Detecting & Handling Outliers**  

📌 **Finding Outliers using IQR (Interquartile Range)**  
```python
Q1 = df['column_name'].quantile(0.25)
Q3 = df['column_name'].quantile(0.75)
IQR = Q3 - Q1

# Define outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Detect outliers
outliers = df[(df['column_name'] < lower_bound) | (df['column_name'] > upper_bound)]
print("Outliers found:", outliers.shape[0])
```
📊 **Output**: Number of **outliers detected**.  

✅ **Ways to Handle Outliers**  
- ✔ **Remove extreme values if they are data errors.**  
- ✔ **Apply log transformation to reduce effect of large values.**  
- ✔ **Winsorization: Replace extreme values with a threshold.**
