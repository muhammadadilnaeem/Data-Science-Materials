# 📊 **`Interquartile Range (IQR) in Statistics for Data Science`** 📏  

## 🔹 **What is Interquartile Range (IQR)?**  

✅ **Definition**: The **Interquartile Range (IQR)** is the range of the **middle 50% of data**, which helps measure **variability** while ignoring outliers.  
✅ **Why is IQR Important?**  
✔ **Resistant to outliers** (unlike range or standard deviation).  
✔ **Best for skewed distributions**.  
✔ **Used in box plots & outlier detection**.  

📌 **Formula for IQR**  
$$
IQR = Q_3 - Q_1
$$
where:  
- **Q1 (First Quartile, 25th percentile)** = Median of lower half of data.  
- **Q3 (Third Quartile, 75th percentile)** = Median of upper half of data.  

✅ **IQR tells us how spread out the middle 50% of values are!**  




---

## 🔹 **1. How to Calculate IQR?**  

📌 **Example Calculation**  
📌 **Dataset (Sorted):** **[10, 20, 30, 40, 50, 60, 70, 80, 90]**  
✔ **Q1 (25th percentile) = 30**  
✔ **Q3 (75th percentile) = 70**  
✔ **IQR = 70 - 30 = 40**  

📌 **Python Code to Calculate IQR**  

```python

import numpy as np

data = [10, 20, 30, 40, 50, 60, 70, 80, 90]  # Sorted dataset
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
print("\nInterquartile Range (IQR):", IQR)
```


📊 **Output**: `IQR: 40` ✅  

✅ **When to Use?**  
✔ **Best for skewed data & non-normal distributions**.  
✔ **Great for outlier detection in box plots**.  



---

## 🔹 **2. How to Use IQR for Outlier Detection?**  

📌 **Outliers are values that fall outside this range**:  
$$
\text{Lower Bound} = Q1 - 1.5 \times IQR
$$
$$
\text{Upper Bound} = Q3 + 1.5 \times IQR
$$

📌 **Python Code for Outlier Detection Using IQR**  


```python

# Finding outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Detecting outliers
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print("\nOutliers:", outliers)
```


📊 **Output**: Identifies outliers based on IQR!  

✅ **Why is this useful?**  
✔ **Filters extreme values** that may skew data analysis.  
✔ **Improves machine learning model performance** by reducing noise.

## 🔹 **4. Comparison: IQR vs. Other Measures of Spread**  

| **Measure** | **Formula** | **Best For** | **Affected by Outliers?** |
|------------|------------|-------------|-----------------|
| **Range** 📏 | $X_{\text{max}} - X_{\text{min}}$ | Quick estimate | ✅ Yes |
| **IQR** 🏗️ | $Q3 - Q1$ | Skewed data, outlier detection | ❌ No |
| **Variance** 🔄 | $\frac{\sum (X_i - \bar{X})^2}{N}$ | Statistical models | ✅ Yes |
| **Standard Deviation** 📉 | $\sqrt{\text{Variance}}$ | Feature scaling, ML models | ✅ Yes |
