# 📊 **`Hypothesis Testing in Statistics for Data Science`** 🔍  

## 🔹 **What is Hypothesis Testing?**  

✅ **Definition**: **Hypothesis testing** is a statistical method used to **make decisions or inferences** about a population based on sample data.  
✅ **Why is Hypothesis Testing Important in Data Science?**  
- ✔ **Validates assumptions** in A/B testing, experiments, and machine learning models.  
- ✔ **Reduces uncertainty** by providing statistical evidence.  
- ✔ **Used in business, healthcare, and finance for decision-making.**  

📌 **Key Concept**: **We never "prove" a hypothesis; we either reject or fail to reject it.**  

---

# 🔥 **1. Key Components of Hypothesis Testing**  

📌 **The Two Competing Hypotheses**  

| **Term** | **Definition** | **Example** |
|---------|--------------|------------|
| **Null Hypothesis (H₀)** ❌ | The default assumption (no effect, no difference) | "A new drug has the same effect as the old drug." |
| **Alternative Hypothesis (H₁ or Hₐ)** ✅ | The claim we want to test (effect exists) | "The new drug is more effective than the old drug." |

📌 **Decision Criteria**  

| **Term** | **Definition** | **Typical Values** |
|---------|--------------|-----------------|
| **Significance Level (α)** 🎯 | Probability of rejecting H₀ when it's true | 0.05 (5%) or 0.01 (1%) |
| **p-value** 📉 | Probability of obtaining observed results if H₀ is true | If **p < α**, reject H₀ |

📌 **Errors in Hypothesis Testing**  

| **Error Type** | **Definition** | **Example** |
|--------------|--------------|------------|
| **Type I Error (False Positive)** ❌ | Rejecting H₀ when it’s actually true | Saying a drug works when it doesn’t |
| **Type II Error (False Negative)** 🚨 | Failing to reject H₀ when it’s false | Saying a drug doesn’t work when it actually does |

✅ **Example**: Testing if an ad campaign increases sales  
- ✔ **H₀**: "The ad campaign has no effect on sales."  
- ✔ **H₁**: "The ad campaign increases sales."  
- ✔ If **p < 0.05**, we reject H₀ and conclude the campaign is effective.  

---

# 📊 **2. Types of Hypothesis Tests**  

| **Test Type** | **Use Case** | **Example** |
|--------------|------------|-------------|
| **Z-Test** | Large sample (n > 30), known population variance | Testing average customer spending vs. industry average |
| **T-Test** | Small sample (n < 30), unknown variance | Comparing sales before & after an ad campaign |
| **Chi-Square Test** | Categorical data, frequency counts | Analyzing customer preferences across regions |
| **ANOVA (Analysis of Variance)** | Comparing 3+ group means | Examining test scores across different schools |
| **Correlation Test** | Checking relationships between variables | Studying the link between ad spend & revenue |

✅ **How to Choose the Right Test?**  
- ✔ **Numerical Data?** → Use **T-Test, Z-Test, ANOVA**.  
- ✔ **Categorical Data?** → Use **Chi-Square Test**.  
- ✔ **Testing Relationships?** → Use **Correlation or Regression**.

# 📉 **3. Step-by-Step Hypothesis Testing (Example: A/B Testing)**  

### **Scenario: Does a new website design increase user engagement?**  
- ✔ **H₀**: "The new design has no impact on engagement."  
- ✔ **H₁**: "The new design increases engagement."  

📌 **Step 1: Collect Sample Data**  


```python

import numpy as np

# Simulated engagement data (Old vs. New design)
old_design = np.random.normal(50, 10, 30)  # Old website engagement scores
new_design = np.random.normal(55, 10, 30)  # New website engagement scores
```

📌 **Step 2: Perform T-Test (Comparing Two Means)**  

```python

from scipy import stats

# Independent t-test
t_stat, p_value = stats.ttest_ind(old_design, new_design)

# Print results
print(f"\nT-statistic: {t_stat:.3f}, P-value: {p_value:.5f}")

# Decision
if p_value < 0.05:
    print("\nReject H₀: The new design significantly increases engagement.")
else:
    print("\nFail to reject H₀: No significant impact on engagement.")
    
```


📊 **Output**:  
- If **p < 0.05**, reject H₀ → **New design increases engagement** 🎯.  
- If **p > 0.05**, fail to reject H₀ → **No strong evidence for improvement**.  

✅ **When to Use?**  
- ✔ **A/B Testing (website changes, marketing campaigns, UX studies).**  
- ✔ **Comparing customer behavior before & after promotions.**  




---

# 📊 **4. One-Tailed vs. Two-Tailed Tests**  

📌 **One-Tailed Test**  
- ✔ Tests if one mean is **greater OR smaller** than another.  
- ✔ Example: "Does the new drug work **better** than the old one?"  

📌 **Two-Tailed Test**  
- ✔ Tests if means are **different in any direction**.  
- ✔ Example: "Is there any difference in effectiveness between two drugs?"  

📌 **Python Code for One-Tailed Test**  

```python


# One-tailed T-test (Checking if new design engagement > old)
t_stat, p_value = stats.ttest_ind(old_design, new_design, alternative='greater')

if p_value < 0.05:
    print("\nNew design has significantly higher engagement.")
else:
    print("\nNo strong evidence for improvement.")
    
```


📊 **Output**: Conclusion depends on **p-value**.  

✅ **When to Use?**  
- ✔ **One-Tailed** → If the goal is **to check for improvement**.  
- ✔ **Two-Tailed** → If the goal is **to check for any difference**.

# 📌 **6. Summary: Key Steps in Hypothesis Testing**  

| **Step** | **Action** | **Example** |
|---------|-----------|------------|
| **1. Define Hypotheses** | Set H₀ & H₁ | "Does ad spend affect sales?" |
| **2. Choose Significance Level (α)** | Common values: **0.05, 0.01** | 5% risk of Type I error |
| **3. Collect Sample Data** | Use random sampling | 100 customer purchase records |
| **4. Select Statistical Test** | T-test, Chi-square, ANOVA | Compare means, proportions, or relationships |
| **5. Compute p-value** | Use statistical formulae | If **p < 0.05**, reject H₀ |
| **6. Draw Conclusion** | Reject or fail to reject H₀ | "The ad campaign increased sales." |
