# 📊 **`ANOVA (Analysis of Variance) in Statistics for Data Science`** 🔍  

## 🔹 **What is ANOVA?**  

✅ **ANOVA (Analysis of Variance)** is a statistical method used to **compare the means of two or more groups** to determine if there are any **statistically significant differences** between them.  

✅ **Why Use ANOVA in Data Science?**  
✔ **Tests the impact of one or more factors on a continuous variable.**  
✔ **More efficient than multiple t-tests** → avoids increased Type I error.  
✔ Widely used in **A/B testing, experimental design, market research, healthcare, and machine learning.**  

📌 **Example Questions:**  
- "Do **three marketing strategies** result in different sales?" (*One-Way ANOVA*)  
- "Do **gender** and **education level** impact salary?" (*Two-Way ANOVA*)  
- "How do **age**, **income**, and **location** influence customer spending?" (*N-Way ANOVA*)  

---

# ⚖️ **1. Types of ANOVA**  

| **ANOVA Type** | **Purpose** | **Example Use Case** |
|----------------|-------------|----------------------|
| **One-Way ANOVA** 🎯 | Tests the effect of **one independent variable** on a dependent variable | Comparing sales across 3 ad campaigns |
| **Two-Way ANOVA** 🆚 | Tests the effects of **two independent variables** and their interaction | Evaluating the effect of **gender** and **education level** on income |
| **N-Way ANOVA** 🔄 | Tests the effects of **multiple independent variables** and their interactions | Analyzing how **age**, **income**, and **location** affect spending habits |

✅ **Key Differences:**  
- **One-Way ANOVA** → One factor (e.g., **Product Type**)  
- **Two-Way ANOVA** → Two factors (e.g., **Gender** & **Ad Type**)  
- **N-Way ANOVA** → More than two factors (e.g., **Age**, **Region**, & **Income**)  
 



---

# 📊 **2. One-Way ANOVA (Single Factor ANOVA)**  

✅ **Purpose**: Compares means across **three or more groups** for a **single independent variable**.  

📌 **Example**:  
"Do **three marketing strategies** result in different sales?"  

### **Hypotheses:**  
- **H₀ (Null Hypothesis)**: All group means are equal.  
- **H₁ (Alternative Hypothesis)**: At least one group mean is different.  

### **Python Example: One-Way ANOVA** 

```python

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Simulated sales data for 3 strategies
np.random.seed(42)
strategy_A = np.random.normal(200, 15, 30)
strategy_B = np.random.normal(220, 15, 30)
strategy_C = np.random.normal(210, 15, 30)

# One-Way ANOVA
f_stat, p_value = stats.f_oneway(strategy_A, strategy_B, strategy_C)

print(f"\nF-Statistic: {f_stat:.3f}, \n\nP-Value: {p_value:.5f}")

if p_value < 0.05:
    print("\nReject H₀: At least one strategy leads to different sales.")
else:
    print("\nFail to reject H₀: No significant difference.")  
```

📊 **Interpretation:**  
- **p < 0.05** → At least one group is different ✅  
- **p > 0.05** → No significant difference ❌

# 🆚 **3. Two-Way ANOVA (Two Factors)**  

✅ **Purpose**: Examines the effect of **two independent variables** and their **interaction** on a dependent variable.  


📌 **Example**:  
"Does **gender** and **education level** impact employee salary?"  

### **Hypotheses:**  
1. **H₀₁**: No effect of **Gender** on Salary.  
2. **H₀₂**: No effect of **Education** on Salary.  
3. **H₀₃**: No **interaction** between Gender & Education.  

### **Python Example: Two-Way ANOVA**  

```python

import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Sample Data: Salary by Gender & Education Level
data = {
    'Salary': [55000, 58000, 60000, 52000, 50000, 57000, 61000, 54000, 56000, 59000],
    'Gender': ['Male', 'Male', 'Male', 'Female', 'Female', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Education': ['Bachelors', 'Masters', 'PhD', 'Bachelors', 'Masters', 'PhD', 'Masters', 'Bachelors', 'PhD', 'Masters']
}

df = pd.DataFrame(data)

# Build ANOVA model
model = ols('Salary ~ C(Gender) + C(Education) + C(Gender):C(Education)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```



### ✅ **Interpretation:**  
- **Gender (p = 0.045)** → Significant → Gender impacts salary.  
- **Education (p = 0.025)** → Significant → Education impacts salary.  
- **Interaction (p = 0.189)** → Not significant → Gender & Education don't interact.

# 🔄 **4. N-Way ANOVA (Multiple Factors)**  

✅ **Purpose**: Examines the impact of **more than two factors** and their **interactions** on a continuous dependent variable.  

📌 **Example**:  
"How do **Age**, **Region**, and **Income** impact Customer Spending?"  

### **Hypotheses:**  
1. Each factor has **no effect** on the dependent variable (H₀).  
2. No interaction effects between factors (H₀).  

### **Python Example: N-Way ANOVA**

```python

import warnings
warnings.filterwarnings("ignore")

# Simulated data with 3 factors
data = {
    'Spending': np.random.randint(100, 500, 30),
    'Age_Group': ['Young', 'Middle', 'Senior'] * 10,
    'Region': ['North', 'South', 'West'] * 10,
    'Income_Level': ['Low', 'Medium', 'High'] * 10
}

df = pd.DataFrame(data)

# Build N-Way ANOVA model
model = ols('Spending ~ C(Age_Group) + C(Region) + C(Income_Level) + C(Age_Group):C(Region) + C(Age_Group):C(Income_Level) + C(Region):C(Income_Level)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```


📊 **Interpretation:**  
- Check **p-values** for each factor and their **interactions**.  
- **p < 0.05** → Factor or interaction has a significant effect.  
- **p > 0.05** → No significant effect.

# ⚖️ **5. Assumptions of ANOVA**  

| **Assumption** | **Description** | **How to Check?** |
|----------------|-----------------|------------------|
| **Independence** 📊 | Observations should be independent | Check data collection method |
| **Normality** 📈 | Residuals should be normally distributed | Use **Shapiro-Wilk** or **Q-Q Plot** |
| **Homoscedasticity** 📏 | Equal variances across groups | Use **Levene’s Test** |

### ✅ **Levene’s Test Example (Equal Variances Check):** 

```python

from scipy.stats import levene

stat, p_value = levene(strategy_A, strategy_B, strategy_C)
print(f"Levene’s Test P-Value: {p_value:.5f}")

if p_value > 0.05:
    print("Variances are equal. Proceed with ANOVA.")
else:
    print("Variances are NOT equal. Use Welch’s ANOVA.")
```


---

# 📉 **6. Post-Hoc Analysis (After Significant ANOVA Results)**  

If ANOVA shows significant results (**p < 0.05**), use **Post-Hoc Tests** to determine **which groups differ**.

### ✅ **Tukey’s HSD (Honest Significant Difference):**  

```python

from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Combine data
spending = np.concatenate([strategy_A, strategy_B, strategy_C])
labels = ['A']*30 + ['B']*30 + ['C']*30

# Tukey's Test
tukey = pairwise_tukeyhsd(endog=spending, groups=labels, alpha=0.05)
print(tukey)
```

|----------------|----------------------------|----------------------------|------------------------------------|
| **One-Way ANOVA** 🎯 | 1                          | Mean differences across groups | Sales across 3 marketing strategies |
| **Two-Way ANOVA** 🆚 | 2                          | Main effects & interaction   | Impact of gender & education on income |
| **N-Way ANOVA** 🔄 | 3+                         | Multiple effects & interactions | How age, income, and region impact spending |


---

# ✅ **Final Takeaways:**  
- **ANOVA** helps detect if **group means differ** significantly.  
- **One-Way ANOVA** → Single factor.  
- **Two-Way ANOVA** → Two factors + interaction.  
- **N-Way ANOVA** → Multiple factors and complex interactions.  
- Always **check assumptions** (normality, equal variances) before running ANOVA.  
- Use **Post-Hoc tests (like Tukey’s HSD)** if results are significant.
