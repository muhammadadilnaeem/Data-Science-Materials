# 📊 **`Data Distributions & Types of Data Distributions in Statistics for Data Science`** 📈  

## 🔹 **What is Data Distribution?**  

✅ **Definition**: A **data distribution** shows how data points are **spread out** over a range of values. It helps in **understanding patterns, trends, and probabilities** in datasets.  

✅ **Why is Data Distribution Important?**  
✔ Helps in **data visualization & interpretation**.  
✔ Essential for **statistical modeling & machine learning**.  
✔ Used for **hypothesis testing & probability analysis**.  

---

## 🔹 **1. Types of Data Distributions in Statistics**  

| **Type of Distribution** | **Shape** | **Example** |
|-----------------|------------------|-------------------|
| **Normal Distribution** 📊 | Bell-shaped, symmetric | Heights, IQ scores, stock returns |
| **Uniform Distribution** 📏 | Equal probability for all values | Rolling a fair die 🎲, random number generators |
| **Binomial Distribution** 🎯 | Discrete, two outcomes | Coin flips, success/failure experiments |
| **Poisson Distribution** 🔢 | Counts rare events in fixed intervals | Number of customer arrivals, accidents per day 🚑 |
| **Exponential Distribution** ⏳ | Skewed, models waiting times | Time between system failures, call center wait times |
| **Log-Normal Distribution** 📈 | Right-skewed, positive values only | Stock prices, income distribution |
| **Gamma Distribution** 📉 | Generalized Poisson/Exponential | Modeling insurance claims, rainfall amounts |
| **Chi-Square Distribution** 🔥 | Right-skewed, sum of squared values | Hypothesis testing, variance analysis |
| **Beta Distribution** 🎰 | Values between 0 & 1 | A/B testing, Bayesian statistics |




---

## 🔹 **2. Normal Distribution (Gaussian) 📊**  

✅ **Definition**: A symmetrical, **bell-shaped** curve where most values are **clustered around the mean (μ)**.  
✅ **Key Properties**:  
✔ **Mean (μ), Median, and Mode are the same**.  
✔ **Defined by Mean (μ) & Standard Deviation (σ)**.  
✔ **Follows the Empirical Rule (68-95-99.7 Rule)**.  

📌 **Python Code to Generate Normal Distribution**  


```python

import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

mu, sigma = 50, 10  # Mean=50, Std Dev=10
data = np.random.normal(mu, sigma, 1000)  # Generate 1000 samples

plt.hist(data, bins=30, density=True, alpha=0.6, color='b', edgecolor='black')
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()  
```



📊 **Output**: A **bell-shaped curve** showing a normal distribution.  

✅ **Applications**  
✔ **Stock price fluctuations**.  
✔ **Machine learning models** (assumptions in linear regression).  
✔ **IQ scores, blood pressure readings**


---

## 🔹 **3. Uniform Distribution 📏**  

✅ **Definition**: All values occur **equally often** → **flat distribution**.  
✅ **Example**: Rolling a **fair die (1-6 all equally likely)**.  

📌 **Python Code for Uniform Distribution**  


```python

import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 10, 1000)  # Uniform between 0 and 10
plt.hist(data, bins=30, density=True, color='g', edgecolor='black')
plt.title("Uniform Distribution")
plt.show()
```


📊 **Output**: A **flat histogram** where all values occur equally.  

✅ **Applications**  
✔ **Random number generation**.  
✔ **Simulations & Monte Carlo methods**.  
✔ **Equal probability experiments**.  



----


## 🔹 **4. Binomial Distribution 🎯**  

✅ **Definition**: Models the number of **successes in n independent trials** with **probability p** of success.  
✅ **Formula**:  

$$
P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k}
$$

where:  
- **n** = Number of trials  
- **k** = Number of successes  
- **p** = Probability of success  

📌 **Example: Probability of Getting 3 Heads in 5 Coin Flips**  


```python

import numpy as np
import matplotlib.pyplot as plt

n, p = 5, 0.5  # 5 trials, 50% probability of heads
binomial_data = np.random.binomial(n, p, 1000)

plt.hist(binomial_data, bins=np.arange(0, 7) - 0.5, density=True, rwidth=0.8)
plt.xticks(range(0, 6))
plt.title("Binomial Distribution (n=5, p=0.5)")
plt.show()  
```



📊 **Output**: Histogram showing the probability of **0-5 heads in 5 flips**.  

✅ **Applications**  
✔ **A/B Testing & marketing experiments**.  
✔ **Quality control (defective vs. non-defective products)**.  
✔ **Sports win/loss probabilities**.



---

## 🔹 **5. Poisson Distribution 🔢**  

✅ **Definition**: Models the probability of **k events occurring in a fixed time interval** when events happen **randomly & independently**.  
✅ **Formula**:  

$$
P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}
$$

where:  
- **λ (lambda)** = Average event rate  
- **k** = Number of events  

📌 **Example: Modeling Website Visits per Hour (λ = 10)**  


```python

import numpy as np
import matplotlib.pyplot as plt

lambda_ = 10  # Average 10 visits per hour
poisson_data = np.random.poisson(lambda_, 1000)

plt.hist(poisson_data, bins=30, density=True, color='orange', edgecolor='black')
plt.title("Poisson Distribution (λ=10)")
plt.show()
```


📊 **Output**: A **right-skewed histogram** showing event occurrences.  

✅ **Applications**  
✔ **Number of customer arrivals at a store**.  
✔ **Network traffic & server requests**.  
✔ **Defect counts in manufacturing**.  



----



## 🔹 **6. Exponential Distribution ⏳**  

✅ **Definition**: Models **waiting times** between Poisson-distributed events.  
✅ **Formula**:  


$$
f(x; \lambda) = \lambda e^{-\lambda x}
$$

where **λ = rate parameter**.  


📌 **Example: Time Between Calls in a Call Center**  

```python

import numpy as np
import matplotlib.pyplot as plt

lambda_ = 1/5  # Average wait time of 5 minutes
exp_data = np.random.exponential(5, 1000)

plt.hist(exp_data, bins=30, density=True, color='purple', edgecolor='black')
plt.title("Exponential Distribution (λ=1/5)")
plt.show()
```


📊 **Output**: A **right-skewed histogram** showing wait times.  

✅ **Applications**  
✔ **Time until system failure** (reliability testing).  
✔ **Customer service wait times**.

### **More Distributions**

```python

import numpy as np
import matplotlib.pyplot as plt

# Generating example datasets with different distributions
np.random.seed(0)  # For reproducibility

# Normal distribution data
normal_data = np.random.normal(loc=0, scale=1, size=1000)

# Uniform distribution data
uniform_data = np.random.uniform(low=-2, high=2, size=1000)

# Exponential distribution data
exponential_data = np.random.exponential(scale=1, size=1000)

# Plotting the distributions
plt.figure(figsize=(12, 8))

# Normal Distribution Plot
plt.subplot(3, 1, 1)
plt.hist(normal_data, bins=30, color='skyblue', alpha=0.7)
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Uniform Distribution Plot
plt.subplot(3, 1, 2)
plt.hist(uniform_data, bins=30, color='lightgreen', alpha=0.7)
plt.title('Uniform Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Exponential Distribution Plot
plt.subplot(3, 1, 3)
plt.hist(exponential_data, bins=30, color='salmon', alpha=0.7)
plt.title('Exponential Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
