# 📊 **`Normal Distribution & Standard Deviation in Statistics for Data Science`** 📈  

## 🔹 **What is Normal Distribution?**  

✅ **Definition**: The **normal distribution (Gaussian distribution)** is a **symmetrical, bell-shaped curve** where most of the data points are concentrated around the mean.  

✅ **Key Properties**:  
✔ **Mean (μ), Median, and Mode are equal**.  
✔ **Symmetrical shape** (left and right halves are mirror images).  
✔ **Defined by Mean (μ) & Standard Deviation (σ)**.  
✔ **Follows the Empirical Rule (68-95-99.7 Rule)**.  

📌 **Probability Density Function (PDF) of Normal Distribution**  
$$
f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x - \mu)^2}{2\sigma^2}}
$$
where:  
- **μ** = Mean (center of distribution).  
- **σ** = Standard deviation (spread of data).  
- **π** and **e** are mathematical constants.  

---

## 🔹 **1. Standard Deviation & Normal Distribution**  

✅ **Definition**: **Standard Deviation ($\sigma$)** measures how spread out the data is from the mean.  
✅ **Relationship with Normal Distribution**:  
✔ **Small $\sigma$** → Data is tightly clustered around the mean.  
✔ **Large $\sigma$** → Data is more spread out.  

📌 **Empirical Rule (68-95-99.7 Rule)**  
$$
\begin{aligned}
\text{68\%} & \text{ of data falls within } 1\sigma \text{ of the mean} \\
\text{95\%} & \text{ of data falls within } 2\sigma \text{ of the mean} \\
\text{99.7\%} & \text{ of data falls within } 3\sigma \text{ of the mean}
\end{aligned}
$$

📌 **Example with Mean = 50 & Standard Deviation = 10**  
✔ **68\% of values** fall between **40 and 60**.  
✔ **95\% of values** fall between **30 and 70**.  
✔ **99.7\% of values** fall between **20 and 80**.  

📌 **Python Code to Visualize Normal Distribution**

```python

import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Define mean & standard deviation
mu, sigma = 50, 10

# Generate data
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, mu, sigma)

# Plot Normal Distribution
plt.plot(x, y, label="Normal Distribution", color="blue")
plt.axvline(mu, color='black', linestyle='dashed', label="Mean (μ)")
plt.axvline(mu - sigma, color='red', linestyle='dashed', label="1σ")
plt.axvline(mu + sigma, color='red', linestyle='dashed')
plt.axvline(mu - 2*sigma, color='orange', linestyle='dashed', label="2σ")
plt.axvline(mu + 2*sigma, color='orange', linestyle='dashed')
plt.axvline(mu - 3*sigma, color='green', linestyle='dashed', label="3σ")
plt.axvline(mu + 3*sigma, color='green', linestyle='dashed')
plt.title("Normal Distribution & Standard Deviation")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.legend()
plt.show() 
```


📊 **Output**: A **bell-shaped curve** with standard deviation markers.  

✅ **Interpretation**  
✔ **Most values fall close to the mean**.  
✔ **Only a few values are in the tails (beyond 3σ).**

## 🔹 **2. Z-Score & Standard Deviation in Normal Distribution**  

✅ **Definition**: A **Z-score (Standard Score)** tells us **how many standard deviations a value is from the mean**.  

✅ **Formula**:  
$$
Z = \frac{X - \mu}{\sigma}
$$
where:  
- $X$ = Data point  
- $\mu$ = Mean  
- $\sigma$ = Standard Deviation  

📌 **Example**  
📌 If **X = 70**, **μ = 50**, **σ = 10**  
$$
Z = \frac{70 - 50}{10} = 2
$$
🔥 **A Z-score of 2 means the value is 2 standard deviations above the mean!**  

📌 **Python Code to Calculate Z-Score**  

```python

# Example values
X = 70
Z = (X - mu) / sigma
print("Z-Score:", Z)
```


📊 **Output**: `Z-Score: 2.0`  

✅ **When to Use Z-Scores?**  
✔ **Finding probabilities in a normal distribution**.  
✔ **Standardizing data for machine learning models**.  
✔ **Comparing values from different distributions**.

## 🔹 **3. Standard Normal Distribution (Z-Distribution) 📉**  

✅ **Definition**: A **normal distribution with μ = 0 and σ = 1**.  
✅ **Z-table**: Used to find probabilities for any Z-score.  

📌 **Example: Probability of Z ≤ 1.5**  
Using Z-table: **P(Z ≤ 1.5) = 0.9332 (93.32%)**  

📌 **Python Code to Find Probability from Z-Score**  

```python

import scipy.stats as stats

# Probability of Z ≤ 1.5
prob = stats.norm.cdf(1.5)
print("\nProbability (Z ≤ 1.5):", prob)
```

📊 **Output**: `Probability: 0.9332`  

✅ **Why is this Useful?**  
✔ Helps **find probabilities** without recalculating for different μ & σ.  
✔ Used in **hypothesis testing & confidence intervals**.
