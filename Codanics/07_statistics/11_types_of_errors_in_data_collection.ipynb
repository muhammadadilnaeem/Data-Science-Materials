# 📊 **`Types of Errors in Data Collection in Statistics for Data Science`** 🔍  

**Data collection errors** occur when **recorded data differs from the actual values** due to mistakes, bias, or limitations in the data-gathering process. Identifying and reducing these errors is crucial for **accurate analysis, reliable models, and valid conclusions** in data science.  

---

## 🔹 **1. Classification of Data Collection Errors**  
Data collection errors can be categorized as:  
1️⃣ **Random Errors** 🎲 (Unpredictable variations).  
2️⃣ **Systematic Errors** 🎯 (Consistent biases).  
3️⃣ **Sampling Errors** 🔍 (Non-representative data).  
4️⃣ **Measurement Errors** ⚙️ (Faulty instruments or methods).  
5️⃣ **Processing Errors** 🖥️ (Data entry or transformation mistakes).  
6️⃣ **Nonresponse Errors** 📭 (Missing data due to non-participation).  

---

# 🟥 **1. Random Errors (Noise) 🎲**  
✅ **Definition**: **Unpredictable errors** caused by chance variations in measurement, human behavior, or external conditions.  

✅ **Causes**:  
✔ **Inconsistent human responses** (e.g., fatigue during surveys).  
✔ **Environmental factors** (e.g., weather affecting temperature sensors).  
✔ **Temporary instrument malfunctions**.  

✅ **Impact**:  
❌ Reduces **precision** of data.  
❌ Increases **variance** but does not create bias.  

✅ **Example**:  
📌 A person measuring their **blood pressure multiple times** gets slightly different readings each time.  

📌 **How to Reduce Random Errors?**  
✔ Take **multiple measurements** and average them.  
✔ Increase **sample size**.  






📌 **Python Example: Simulating Random Errors**  


```python
import numpy as np

true_value = 100
random_errors = np.random.normal(0, 5, 10)  # Random noise with mean=0, std=5

# Simulated observed values
observed_values = true_value + random_errors
print("Observed Data with Random Errors:", observed_values)
```

📊 **Output**: Values fluctuate around **100** due to random noise.  



---

# 🟧 **2. Systematic Errors (Bias) 🎯**  
✅ **Definition**: **Consistent, repeatable errors** that shift data in **one direction** due to bias in measurement or data collection.  

✅ **Causes**:  
✔ **Faulty instruments** (e.g., miscalibrated weighing scale).  
✔ **Interviewer bias** (e.g., leading questions in surveys).  
✔ **Selection bias** (e.g., only collecting data from urban areas).  

✅ **Impact**:  
❌ Leads to **inaccurate conclusions**.  
❌ Causes **skewed predictions** in machine learning.  

✅ **Example**:  
📌 A scale always adds **+2 kg** to every weight measurement.  

📌 **How to Reduce Systematic Errors?**  
✔ **Calibrate measuring instruments** regularly.  
✔ Use **double-blind experiments** in surveys.  

📌 **Python Example: Simulating Systematic Bias**  


```python
import numpy as np

bias = 3  # Constant overestimation of values
true_values = np.array([50, 60, 70, 80, 90])
biased_values = true_values + bias
print("True Values:", true_values)
print("Biased Measurements:", biased_values)
```

📊 **Output**: Every value is **shifted by +3** → clear systematic bias.  



---

# 🟨 **3. Sampling Errors (Selection Bias) 🔍**  
✅ **Definition**: **Errors caused by selecting a non-representative sample**, leading to biased results.  

✅ **Causes**:  
✔ **Undercoverage** – Certain groups are **excluded** from the sample.  
✔ **Self-selection bias** – Only certain people **choose to respond**.  
✔ **Convenience sampling** – Only easy-to-reach participants are included.  

✅ **Impact**:  
❌ Reduces **generalizability** of the study.  
❌ Creates **false correlations** and **wrong conclusions**.  

✅ **Example**:  
📌 A survey on **political opinions** conducted only on **social media users** excludes older populations.  

📌 **How to Reduce Sampling Errors?**  
✔ Use **random sampling** methods.  
✔ Increase **sample diversity**.  


📌 **Python Example: Checking Sampling Bias**  


```python
import numpy as np

# Full population
population = np.arange(1, 101)  # Numbers 1 to 100

# Biased sample (first 10 values)
biased_sample = population[:10]  # Non-random selection

# Unbiased sample (random selection)
random_sample = np.random.choice(population, size=10, replace=False)

print("Biased Sample:", biased_sample)
print("Unbiased Sample:", random_sample)
```

📊 **Output**: The **first 10 values are biased**, while **random selection provides a fair representation**.  



---

# 🟩 **4. Measurement Errors (Instrument Errors) ⚙️**  
✅ **Definition**: Errors caused by **faulty measurement tools or incorrect recording methods**.  

✅ **Causes**:  
✔ **Misaligned sensors** (e.g., a thermometer consistently measuring **2°C higher**).  
✔ **Human misinterpretation** (e.g., doctor misreading X-ray scans).  
✔ **Improper data formatting** (e.g., incorrectly rounding values).  

✅ **Impact**:  
❌ Reduces **accuracy & reliability** of measurements.  
❌ Can create **systematic biases** if errors occur consistently.  

✅ **Example**:  
📌 A **speedometer in a car** shows speeds **5% higher than actual speeds**.  

📌 **How to Reduce Measurement Errors?**  
✔ Use **high-quality, calibrated equipment**.  
✔ Automate data collection **to avoid human errors**.  


📌 **Python Example: Detecting Measurement Errors**  


```python
import numpy as np

# Simulated true & measured values with instrument errors
true_values = np.array([10, 20, 30, 40, 50])
measured_values = true_values * 1.05  # 5% error due to miscalibration

# Checking deviation
error = measured_values - true_values
print("Measurement Errors:", error)
```

📊 **Output**: Every value has a **5% overestimation error**.  



----


# 🟦 **5. Processing Errors (Data Entry Errors) 🖥️**  
✅ **Definition**: Errors introduced **during data entry, transformation, or cleaning**.  

✅ **Causes**:  
✔ **Typos during manual data entry** (e.g., entering **"23" instead of "32"**).  
✔ **Incorrect data transformations** (e.g., dividing by **100 instead of 10**).  
✔ **Merging errors** (e.g., duplicate customer records).  

✅ **Impact**:  
❌ Produces **false patterns** in data.  
❌ Leads to **incorrect machine learning predictions**.  

✅ **Example**:  
📌 A dataset records **date of birth as 2098 instead of 1998**.  

📌 **How to Reduce Processing Errors?**  
✔ Use **automated data validation checks**.  
✔ Implement **duplicate record detection**.  


📌 **Python Example: Detecting Duplicates in Data**  


```python
import pandas as pd

# Simulated dataset
data = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4],
    'Age': [25, 30, 30, 40, 50]
})

# Finding duplicates
duplicates = data.duplicated()
print("Duplicate Rows:\n", data[duplicates])
```

📊 **Output**: The **duplicate record for ID 2 is flagged**.  





---

# 🟪 **6. Nonresponse Errors (Missing Data) 📭**  
✅ **Definition**: Errors caused when **data is missing** because respondents **fail to participate or provide answers**.  

✅ **Causes**:  
✔ **Survey fatigue** – Participants do not answer all questions.  
✔ **Technical failures** – Sensors fail to record data.  
✔ **Privacy concerns** – Users refuse to share personal details.  


📌 **Python Example: Detecting Missing Data**  


```python
import pandas as pd

data = pd.DataFrame({'Age': [25, 30, None, 40, 50]})  # One missing value
print(data.isnull().sum())  # Counts missing values
```

📊 **Output**: **1 missing value detected in 'Age' column**.  




---

# 🎯 **Final Takeaways**  
✔ **Random Errors** – Unpredictable fluctuations, reduced by **large samples**.  
✔ **Systematic Errors** – Consistent bias, fixed by **calibration & better methods**.  
✔ **Sampling Errors** – Poor sample selection, solved with **random sampling**.  
✔ **Measurement Errors** – Instrumental issues, solved with **automation & validation**.  
✔ **Processing Errors** – Data entry mistakes, solved with **checks & cleaning**.  
✔ **Nonresponse Errors** – Missing data, solved with **imputation or better survey design**.
