### **Import Libraries**

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import scipy.stats as stats
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

### **Use `one-sample t-Test` on titanic data from seaborn library**

```python

# load tips data
df = sns.load_dataset("titanic")
df.head()
```

```python

# plot distribution of data
sns.histplot(df["age"],kde=True,color="b")
plt.show()
```

```python

# shpiro wilk test to see the normality of the data
statistic, p = stats.shapiro(df['age'])

# print the results based on p value and if else conditions
if p < 0.05:
    print("Reject null hypothesis, data is not normally distributed")
else:
    print("Fail to reject null hypothesis, data is normally distributed")
```

```python

# impute missing values in 'age' and 'fare' column
df["age"] = df["age"].fillna(df["age"].median())
df["fare"] = df["fare"].fillna(df["fare"].median())

print("Missind values in 'age' and 'fare' imputed successfully.")
```


>mu = 35

>is there any difference between ages os people and mu(sample)?

```python

# apply 1 sample t-Test
statistic, p = stats.ttest_1samp(df['age'], 35)

# print the results based on p value and if else conditions
if p < 0.05:
    print("\nReject null hypothesis, there is a significant difference between the groups.")
else:
    print("\nFail to reject null hypothesis, there is no significant difference between the groups.")
```

### **Use `two-sample t-Test` on titanic data from seaborn library**

```python

# lets run the two sample t test on ages of male and female from titanic
age_male = df[df['sex'] == 'male']['age']
age_female = df[df['sex'] == 'female']['age']

statistic, p = stats.ttest_ind(age_male, age_female)

# print the results based on p value and if else conditions
if p < 0.05:
    print(f"\nReject null hypothesis, because p vaule: {p} , there is a significant difference between ages of male and female passengers of titanic.")
else:
    print(f"\nFail to reject null hypothesis,because p vaule: {p}, there is not any differences between ages of female and male passengers.")
    
```
