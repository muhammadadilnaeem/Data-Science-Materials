# 📊 **`Descriptive Statistics in Data Science`** 🔍  

## 🔹 **What is Descriptive Statistics?**  

✅ **Definition**: **Descriptive statistics** is the branch of statistics that **summarizes and describes** the key features of a dataset **without making inferences or predictions**.  
✅ **Why is Descriptive Statistics Important?**  
✔ Provides **a quick summary of data** 📊.  
✔ Helps in **data cleaning & preprocessing** for machine learning 🧹.  
✔ Identifies **patterns, trends, and outliers** 🔍.  

📌 **Two Main Types of Descriptive Statistics**  
1️⃣ **Measures of Central Tendency** → What is the "typical" value? (Mean, Median, Mode)  
2️⃣ **Measures of Dispersion (Variability)** → How spread out is the data? (Range, Variance, Standard Deviation, IQR)  

---

# 📈 **1. Measures of Central Tendency (Summary of Data Center) 🎯**  

📌 **Three Key Metrics**  

| **Measure** | **Definition** | **Formula** | **Best Used For** |
|------------|--------------|-------------|----------------|
| **Mean (Average)** ➗ | Sum of values divided by count | $\bar{X} = \frac{\sum X_i}{n}$ | Normally distributed data |
| **Median** 🔀 | Middle value when sorted | Middle of sorted data | Skewed data (resistant to outliers) |
| **Mode** 📊 | Most frequently occurring value | Most common value | Categorical & frequency-based data |

📌 **Example: Dataset = [10, 20, 30, 40, 50, 100]**  
✔ **Mean** = $\frac{(10+20+30+40+50+100)}{6} = 41.67$  
✔ **Median** = **35** (middle value after sorting)  
✔ **Mode** = **No mode** (all unique values)

📌 **Python Code for Central Tendency**  

```python

import numpy as np
from scipy import stats

data = [10, 20, 30, 40, 50, 100]

mean_value = np.mean(data)
median_value = np.median(data)

mode_value = stats.mode(data, keepdims=True)  # Ensure compatibility with new SciPy versions

# Extract mode and count properly
mode_result = mode_value.mode if mode_value.count > 1 else "No Mode"

print("\nMean:", mean_value)
print("\nMedian:", median_value)
print("\nMode:", mode_result)
```


📊 **Output**:  
- **Mean:** 41.67  
- **Median:** 35  
- **Mode:** No Mode  

✅ **Which Measure to Use?**  
✔ **Mean** → Best for normal distributions.  
✔ **Median** → Best for **skewed** data (not affected by outliers).  
✔ **Mode** → Best for **categorical data** (e.g., most common customer preference).  



----

# 📉 **2. Measures of Dispersion (Spread of Data) 📏**  

📌 **Four Key Metrics**  

| **Measure** | **Definition** | **Formula** | **Best Used For** |
|------------|--------------|-------------|----------------|
| **Range** 📏 | Difference between max & min values | $X_{\max} - X_{\min}$ | Quick spread estimate |
| **Variance ($\sigma^2$, $s^2$)** 🔄 | Average squared deviation from the mean | $\frac{\sum (X_i - \bar{X})^2}{n}$ | Statistical modeling |
| **Standard Deviation ($\sigma$, $s$)** 📉 | Square root of variance | $\sigma = \sqrt{\text{Variance}}$ | Understanding spread in same units |
| **Interquartile Range (IQR)** 🏗️ | Spread of middle 50% of data | $Q_3 - Q_1$ | Outlier detection & skewed data |

📌 **Example: Dataset = [10, 20, 30, 40, 50, 100]**  
✔ **Range** = $100 - 10 = 90$  
✔ **Variance** = $1347.22$  
✔ **Standard Deviation** = $36.72$  
✔ **IQR** = $Q_3 (45) - Q_1 (25) = 20$  

📌 **Python Code for Dispersion**

```python


data_range = max(data) - min(data)
variance = np.var(data, ddof=1)  # Sample Variance
std_dev = np.std(data, ddof=1)  # Sample Standard Deviation
Q1, Q3 = np.percentile(data, [25, 75])
IQR = Q3 - Q1

print("\nRange:", data_range)
print("\nVariance:", variance)
print("\nStandard Deviation:", std_dev)
print("\nIQR:", IQR)
```


📊 **Output**:  
- **Range:** 90  
- **Variance:** 1016.66  
- **Standard Deviation:** 31.88  
- **IQR:** 25  

✅ **When to Use Dispersion Metrics?**  
✔ **Variance & Standard Deviation** → Best for **normal distributions**.  
✔ **IQR** → Best for **skewed data** & **outlier detection**.

# 🔄 **3. Skewness & Kurtosis (Data Shape & Tails) 🔍**  

✅ **Skewness** → Measures **asymmetry** in data distribution.  
✔ **Positive Skew** 📈 → Right tail is longer.  
✔ **Negative Skew** 📉 → Left tail is longer.  

✅ **Kurtosis** → Measures **tail heaviness** (outliers).  
✔ **High Kurtosis (Leptokurtic)** 📌 → More extreme values.  
✔ **Low Kurtosis (Platykurtic)** 🏔️ → Fewer outliers.  



📌 **Python Code for Skewness & Kurtosis**  

```python

skewness = stats.skew(data)
kurtosis = stats.kurtosis(data)

print("\nSkewness:", skewness)
print("\nKurtosis:", kurtosis)
```


📊 **Output**: Shows **how asymmetric & heavy-tailed** the data is.  

✅ **Why It Matters?**  
✔ **High Skewness** → Log transformation may help.  
✔ **High Kurtosis** → More outliers, affects ML models.

# 📊 **4. Visualization of Descriptive Statistics**  

✅ **Histogram → Distribution of Data**

```python

import matplotlib.pyplot as plt

plt.hist(data, bins=10, edgecolor='black')
plt.title("Histogram of Data")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

📊 **Output**: A **histogram** showing data spread.  


✅ **Box Plot → Outlier Detection & IQR**  

```python

plt.boxplot(data)
plt.title("Box Plot of Data")
plt.show()
```

📊 **Output**: Identifies **outliers, median, & quartiles**.
