# 📊 **`Median in Statistics for Data Science`** 🔍  

## 🔹 **What is Median?**  

✅ **Definition**: The **median** is the middle value of a dataset **when arranged in ascending order**.
  
✅ **Why is it Important?**  
✔ **Not affected by outliers** (robust measure).  
✔ **Represents the central point of skewed data**.  
✔ **Used in income analysis, real estate, and machine learning**.  

📌 **Formula for Median**  
- If **n is odd**:  
  $$
  \text{Median} = X_{\left(\frac{n+1}{2}\right)}
  $$
- If **n is even**:  
  $$
  \text{Median} = \frac{X_{\left(\frac{n}{2}\right)} + X_{\left(\frac{n}{2} + 1\right)}}{2}
  $$





---

## 🔹 **1. How to Calculate Median?**  

📌 **Example 1: Odd Number of Values**  
📌 Data: **[10, 20, 30, 40, 50]**  
✔ Sorted: **[10, 20, 30, 40, 50]**  
✔ Middle value: **Median = 30**  

📌 **Example 2: Even Number of Values**  
📌 Data: **[10, 20, 30, 40, 50, 60]**  
✔ Sorted: **[10, 20, 30, 40, 50, 60]**  
✔ Middle values: **30 & 40**  
✔ **Median = (30 + 40) / 2 = 35**  



📌 **Python Code to Calculate Median**  


```python

import numpy as np

data = [10, 20, 30, 40, 50, 60]
median_value = np.median(data)
print("Median:", median_value)
```

📊 **Output**: `Median: 35`  



---

## 🔹 **2. When to Use Median Instead of Mean?**  

| **Scenario** | **Use Median?** | **Use Mean?** |
|-------------|----------------|---------------|
| **Normally Distributed Data** | ❌ | ✅ |
| **Skewed Data (Outliers Present)** | ✅ | ❌ |
| **House Prices, Salaries, Income** | ✅ | ❌ |
| **Exam Scores (Normally Distributed)** | ❌ | ✅ |


📌 **Example Where Mean is Misleading**  
📌 Data: **[10, 20, 30, 40, 1000]** (Outlier: **1000**)  
- **Mean** = \( (10+20+30+40+1000)/5 = 220 \) ❌  
- **Median** = **30** ✅ (Best choice!)  



📌 **Python Example Showing Why Median is Better**  


```python

import numpy as np

data_with_outlier = [10, 20, 30, 40, 1000]
print("\nMean:", np.mean(data_with_outlier))
print("\nMedian:", np.median(data_with_outlier))
```


📊 **Output**:  
- `Mean: 220` (Incorrect representation due to outlier)  
- `Median: 30` (Correct representation)

## 🔹 **3. Applications of Median in Data Science**  

📌 **Business & Finance**  
✔ **Median Salary** → Used to avoid CEO outliers in salary reports.  
✔ **Median House Prices** → Better than mean for real estate pricing.  

📌 **Healthcare & Medicine**  
✔ **Median Lifespan** → Used in medical research to compare survival rates.  
✔ **Median Patient Recovery Time** → Avoids extreme outlier cases.  

📌 **Machine Learning & AI**  
✔ **Median Imputation** → Used to fill missing values in skewed data.  
✔ **Feature Engineering** → Helps normalize data distribution.
