# 📊 **`How to Choose the Right Statistical Test in Statistics for Data Science`**  

## 🔍 **1. Choosing the Right Statistical Method**  

Choosing the correct statistical test is crucial for accurate analysis. Follow these key steps to make the right choice:  

### ✅ **Step 1: Identify Your Data Type**  
- **Numerical Data (Continuous)** → Heights, weights, temperatures, etc. 📏  
- **Categorical Data** → Colors, gender, brands, etc. 🎨  
- **Ordinal Data** → Ratings (e.g., 1-5 stars), rankings 🏆  

### ✅ **Step 2: Check the Number of Groups**  
- Comparing **1 group** → Single-sample tests  
- Comparing **2 groups** → Two-sample tests  
- Comparing **3+ groups** → ANOVA or Chi-Square  

### ✅ **Step 3: Determine Data Distribution**  
- **Normally Distributed (Bell Curve) 📈** → Use parametric tests

    - For Parametric Test Assumptions are:
     
      - ***Step:1. Normality Test***\
      Tests to be used:
        - **`1. Shapiro-Wilk Test`** : (*Specific (Reliable)*)
        - **`2. Kolmogorov-Smirrnov Test`** : (*General (Less Reliable)*) 
    
      - ***Step:2. Homogeneity Test***\
      The variance of the variable in data are equal.\
      Tests to be used:
        - **`Levene's Test`**  
    
      - ***Step:3. Purpose***\
      Know Purpose of your research question.\
      Two types of Purpose:

        - **`Comparison`** : (*Difference (Atleast Two Groups Should Be There)*)
          - Male vs Female
          - Control Group vs Treatment Group  
          - Grouping Individulas by Color Preferences
              
        - **`Relationship`** : (*Connection (Find the Connections)*)
          - Can food predict weight of a group of individuals?
          - Do fertilizers application increase crop growth?
          
            - We seek following Here:
              - *Connection*
              - *Correlation*    
              - *Causation*    
              - *Prediction* 
        
      - ***Step:4. Data Type***\
      Know the type of data you are working with.

        - **`Categorical`** (*Qualitative*):
            - No numeric Meaning
            - Represented in Texts
            - (e.g: **character**, **factors**)

        - **`Numerical`** (*Quantitative*):
            - Mostly represented in Numbers
            - (e.g: Numerical variable **int** and **float**) 

      - ***Step:5. Statistical Tests***\
      Choose a Statistical tests from 3 main Families:

        - **`1. Chi-Squared`**:
          - **Purpose** : Comparison
          - **Data** : Categorical Only
          - **Types** : 
            - *Chi-Squared test of homogeneity* 
            - *Chi-Squared test of Independence*
          - **When to use?** : 
            - Can be used with any number of levels or groups.   
            - Nothing affests this.


        - **`2. t-Test / ANOVA`**:
          - **Purpose** : Comparison
          - **Data** : Categorical and Numerical
          - **Types** : 
            - ***1. One-sample t-Test*** 
              - For one sample Group with a Known Mean 
            - ***2. Two-sample t-Test***
              - Un-paired t-test (*Two Different Groups*)  
              - Paired t-test (*Two Same Groups*)
            - ***3. ANOVA (Analysis Of Variance)*** 
              - 3+ Levels of Groups are involved
                - `One-way ANOVA` : 
                  - Even one of the group is significant you will get significant results, but doesn't tell you which one.
                - `Two-way ANOVA` : 
                - `Repeated measures of ANOVA` : 
                  - 3+ Paired groups, scale up of Paired t-Test.       

        - **`3. Correlation`**:
          - **Purpose** : Relationship
          - **Data** : Numerical Only
          - **Types** : 
            - `1. Pearson's Corelation` : 
              - One **Independent** and one **dependent** Variable.
            - `2. Regression` : 
              - One **Independent** and one **dependent** Variable.
          - **Corelation** :
            - Tells us how closely connected two variables are?
            - "Is food a predictor of weight gain?"
          - **Regression** :
            - Tells us about a specific mathematical equation that describes the relationship.
            - This helps us find Data Points not measured yest.
            - *(e.g : missing values can be predicted like this!)*       
         






- **Not Normally Distributed (Skewed) 📊** → Use non-parametric tests\
If data is not Normally Distributed then

    - ***`I. Normalize Your Data`*** : 
      - *Standardrization* 
      - *Min-Max Scaling* 
      - *Log-Transformation*
    - ***`II. Use Alternative Non-Parametric Tests`***
      -  **1. Chi-Squared**
         - Same as for parametric tests
      -  **2. t-Test / ANOVA**
         - `One-sample t-Test` :
           - *One-Sample Wilcoxon Signed rank Test*     
         - `Two-sample t-Test` :
           - *a) Unpaired t-Test* : (Mann Whitney's U-test)     
           - *b) Paired t-Test* : (Wilcoxon)
         -  `ANOVA`
             - *Kruskal-Wallis-Test*
      - **3. Pearson Correlation** : 
          - *Spearman's Corelation*
          - *Kendall's Tau*
        - *Regression*           
 

### ✅ **Step 4: Define Your Goal**  
| Analysis Goal 🏹 | Example | Suggested Test 📊 |
|-----------------|---------|-----------------|
| Compare Mean of One Group to a Fixed Value | Is average height = 170 cm? | One-Sample **t-Test** |
| Compare Means of Two Independent Groups | Are males taller than females? | **Independent t-Test** |
| Compare Means of Paired Groups | Did students improve after training? | **Paired t-Test** |
| Compare More Than Two Groups | Do different diets affect weight loss? | **ANOVA** (if normal) / **Kruskal-Wallis** (if not) |
| Check Association Between Two Categorical Variables | Do gender and product preference relate? | **Chi-Square Test** |
| Check Correlation Between Two Variables | Do temperature and ice cream sales correlate? | **Pearson (linear)** / **Spearman (non-linear)** |
| Predict a Continuous Outcome | How does advertising affect sales? | **Linear Regression** |
| Predict a Categorical Outcome | Will a customer buy a product? | **Logistic Regression** |

## ⚠️ **2. Do’s and Don’ts of Choosing a Statistical Test**  

### ✅ **Do’s**  
✔️ Always check the **data type** before selecting a test 🧐  
✔️ Ensure your sample size is large enough for accurate results 📊  
✔️ Use visualizations to explore data before applying tests 📈  
✔️ Consider **assumptions** (normality, independence, equal variance) before using parametric tests ✔️  
✔️ Double-check if your test is **one-tailed** or **two-tailed**  

### ❌ **Don’ts**  
❌ Don’t apply a parametric test to non-normal data unless transformed  
❌ Don’t assume correlation = causation 🚫  
❌ Don’t ignore outliers—they can mislead results! 🚨  
❌ Don’t perform too many tests without adjustments (e.g., Bonferroni correction)

## ✅ **3. Reliable vs. Non-Reliable Results**  

| **Reliable Results** 🟢 | **Non-Reliable Results** 🔴 |
|-----------------|----------------|
| ✅ Large enough sample size 📊 | ❌ Small sample, leading to randomness |
| ✅ Follows statistical assumptions | ❌ Assumptions violated (e.g., normality ignored) |
| ✅ Properly chosen test | ❌ Wrong test for data type |
| ✅ No cherry-picking results 🍒 | ❌ Selective reporting |
| ✅ Clearly reported p-values and confidence intervals | ❌ Misinterpreted p-values |

### 🎯 **What Makes a Test Result Reliable?**  
- **P-Value (p < 0.05)** → Statistically significant ✅  
- **Effect Size** → Measures strength of the relationship  
- **Confidence Interval (e.g., 95%)** → Tells us the range where true value likely lies

## 📊 **4. Making Data Visualization for Statistical Analysis**  

Visualizing data helps in selecting the right test and interpreting results correctly!  

### 🔥 **Best Plots for Statistical Analysis**  
| **Purpose** | **Recommended Plot** |
|------------|--------------------|
| Checking Normality | Histogram 📊, Q-Q Plot 📈 |
| Comparing Two Groups | Boxplot 📦, Violin Plot 🎻 |
| Correlation Between Two Variables | Scatter Plot 🔵📈 |
| Checking Distribution of Data | Histogram 📊, Density Plot 🌊 |
| Comparing More Than Two Groups | Bar Chart 📊, ANOVA Boxplot 📦 |

### 🔍 **Example: Checking Normality Before Applying a t-Test**  
- ✅ **Histogram** → Check if data looks like a bell curve 🛎  
- ✅ **Q-Q Plot** → Points should align on the diagonal line for normality  

**If data is NOT normal → Use non-parametric alternatives like Mann-Whitney U Test instead of t-Test!**

## **🏆 5. Interpreting Results from a Statistical Test**  

Understanding the output is just as important as running the test!  

### 📌 **Key Terms in Test Results**  
- **P-value < 0.05** → Reject null hypothesis (significant result) ✅  
- **P-value > 0.05** → Fail to reject null (not significant) ❌  
- **Effect Size** → Tells how strong the relationship is  
- **Confidence Interval (CI)** → If CI doesn’t include zero, effect is significant  

### 🚀 **Example: Interpreting a t-Test Result**  
```
t(48) = 2.31, p = 0.025  
```
🔹 **t(48)** → Test statistic with 48 degrees of freedom  
🔹 **p = 0.025** → Since p < 0.05, we reject the null hypothesis → The difference is significant! 🎉
