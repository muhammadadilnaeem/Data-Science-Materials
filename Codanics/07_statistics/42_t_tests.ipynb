# 📊 **`t-Tests in Statistics for Data Science`** 🔍  

## 🔹 **What is a t-Test?**  

✅ **Definition**: A **t-test** is a statistical test used to determine whether **the means of two groups are significantly different** from each other.  

✅ **Why Use a t-Test?**  
- ✔ **Tests if observed differences are due to chance or real effects**.  
- ✔ **Used in A/B testing, business analytics, healthcare, and social sciences**.  
- ✔ **Works well for small sample sizes (n < 30)**.  

📌 **Example Questions**:  
- "Do customers who see **Ad A** spend more than those who see **Ad B**?"  
- "Is the **average test score** of two different schools significantly different?"  
- "Does a **new drug** lower blood pressure more than a placebo?"  

---

# 🔥 **1. Types of t-Tests**  

| **Test Type** | **Purpose** | **Example Use Case** |
|--------------|------------|----------------------|
| **One-Sample t-Test** 🎯 | Compares a sample mean to a known population mean | "Is the average salary in a company $50,000?" |
| **Independent (Two-Sample) t-Test** 🆚 | Compares means of two independent groups | "Do male and female employees earn the same salary?" |
| **Paired t-Test** 🔄 | Compares means of the same group before & after treatment | "Did a new training program improve employee performance?" |

✅ **Choosing the Right t-Test**  
- ✔ **Compare a group vs. a population mean?** → Use **One-Sample t-Test**.  
- ✔ **Compare two different groups?** → Use **Independent t-Test**.  
- ✔ **Compare the same group before & after treatment?** → Use **Paired t-Test**.  

---

# 📊 **2. One-Sample t-Test: Comparing a Sample to a Population**  

📌 **Example**:  
- ✔ A company claims that the **average employee salary is $50,000**.  
- ✔ We test if a **sample of 30 employees** has a mean salary significantly different from $50,000.  

📌 **Step 1: Define Hypotheses**  
- **H₀ (Null Hypothesis)**: "The average salary is **$50,000** (no difference)."  
- **H₁ (Alternative Hypothesis)**: "The average salary is **not $50,000** (significant difference)."  

📌 **Step 2: Perform One-Sample t-Test in Python**  


```python

import numpy as np
import scipy.stats as stats

# Sample data: Employee salaries
salaries = np.random.normal(loc=48000, scale=5000, size=30)  # Mean=48,000, Std Dev=5,000

# Population mean
pop_mean = 50000  

# Perform One-Sample t-Test
t_stat, p_value = stats.ttest_1samp(salaries, pop_mean)

print(f"\nT-statistic: {t_stat:.3f}, \n\np-value: {p_value:.5f}")

# Decision
if p_value < 0.05:
    print("\nReject H₀: The average salary is significantly different from $50,000.")
else:
    print("\nFail to reject H₀: No significant difference from $50,000.")
```


📊 **Output**:  
- ✔ **p < 0.05** → Reject H₀ → **Salary is significantly different from $50,000**.  
- ✔ **p > 0.05** → Fail to reject H₀ → **No strong evidence of a difference**.  

✅ **When to Use?**  
- ✔ Checking if **test scores, sales, or prices deviate from a benchmark**.

# 🆚 **3. Independent (Two-Sample) t-Test: Comparing Two Groups**  

📌 **Example**:  
✔ Do **male and female employees** earn the **same salary**?  

📌 **Step 1: Define Hypotheses**  
- **H₀**: "The average salaries of both genders are the **same**."  
- **H₁**: "The average salaries of both genders are **different**."  

📌 **Step 2: Perform Independent t-Test in Python**  

```python

# Simulated salaries for two groups
male_salaries = np.random.normal(loc=52000, scale=5000, size=30)
female_salaries = np.random.normal(loc=50000, scale=5000, size=30)

# Perform Independent t-Test
t_stat, p_value = stats.ttest_ind(male_salaries, female_salaries)

print(f"\nT-statistic: {t_stat:.3f}, \n\np-value: {p_value:.5f}")

# Decision
if p_value < 0.05:
    print("\nReject H₀: Salaries are significantly different between genders.")
else:
    print("\nFail to reject H₀: No significant salary difference between genders.")
    
```


📊 **Output**:  
- ✔ **p < 0.05** → Reject H₀ → **Salaries are significantly different**.  
- ✔ **p > 0.05** → Fail to reject H₀ → **No strong evidence of a difference**.  

✅ **When to Use?**  
- ✔ **A/B Testing** (Ad A vs. Ad B click rates).  
- ✔ **Comparing customer satisfaction in two regions**.  



---

# 🔄 **4. Paired t-Test: Comparing Before & After Treatment**  

📌 **Example**:  
- ✔ Did **a new training program improve employee productivity**?  
- ✔ Measure **productivity scores** before & after training.  

📌 **Step 1: Define Hypotheses**  
- **H₀**: "Productivity before and after training is the **same**."  
- **H₁**: "Training **improves productivity**."  

📌 **Step 2: Perform Paired t-Test in Python**  


```python


# Simulated productivity scores before & after training
before_training = np.random.normal(loc=60, scale=5, size=30)
after_training = before_training + np.random.normal(loc=5, scale=2, size=30)  # Slight improvement

# Perform Paired t-Test
t_stat, p_value = stats.ttest_rel(before_training, after_training)

print(f"\nT-statistic: {t_stat:.3f}, \n\np-value: {p_value:.5f}")

# Decision
if p_value < 0.05:
    print("\nReject H₀: Training significantly improved productivity.")
else:
    print("\nFail to reject H₀: No strong evidence of improvement.")
```


📊 **Output**:  
- ✔ **p < 0.05** → Reject H₀ → **Training improved productivity**.  
- ✔ **p > 0.05** → Fail to reject H₀ → **No strong evidence of improvement**.  

✅ **When to Use?**  
- ✔ **Before/After Comparisons** (e.g., weight loss before & after a diet).  
- ✔ **Customer satisfaction before & after product changes**.

# 🎯 **5. Summary: Choosing the Right t-Test**  

| **Test Type** | **Comparison** | **Example** |
|--------------|---------------|------------|
| **One-Sample t-Test** 🎯 | Sample mean vs. Population mean | "Is the average test score **50**?" |
| **Independent t-Test** 🆚 | Two **independent** groups | "Do men and women have different salaries?" |
| **Paired t-Test** 🔄 | Same group before & after | "Did training increase productivity?" |

✅ **If comparing means, always check assumptions**:  
- ✔ Data should be **normally distributed**.  
- ✔ Groups should have **equal variance** (use **Levene’s test** if unsure).
