# 📊 **`Inferential Statistics for Data Science`** 🔍  

## 🔹 **What is Inferential Statistics?**  

✅ **Definition**: **Inferential statistics** is a branch of statistics that helps make **predictions, decisions, or generalizations** about a **population** based on a **sample**.  

✅ **Why is Inferential Statistics Important in Data Science?**  
- ✔ Helps **draw conclusions from sample data** 📊.  
- ✔ Essential for **A/B testing, surveys, and market research**.  
- ✔ Used in **machine learning, hypothesis testing, and forecasting** 🤖.  

📌 **Two Key Concepts in Inferential Statistics**  
- 1️⃣ **Estimation** → Estimating population parameters (e.g., mean, proportion) from a sample.  
- 2️⃣ **Hypothesis Testing** → Testing claims or assumptions about a population.  

---

# 📊 **1. Sampling & Estimation in Inferential Statistics**  

✅ **Definition**: Since analyzing the entire population is impractical, inferential statistics relies on **samples** to estimate population parameters.  

📌 **Types of Estimation**  

| **Method** | **Description** | **Example** |
|------------|---------------|------------|
| **Point Estimation** 📍 | A single value estimate for a population parameter | Sample mean as an estimate of the population mean |
| **Confidence Interval** 📏 | A range of values likely to contain the true population parameter | Estimating average salary with 95% confidence |
| **Margin of Error** ⚠️ | The range above/below the point estimate | +/- $5,000 around the mean salary |

📌 **Python Code: Confidence Interval for Sample Mean**  

```python


import numpy as np
import scipy.stats as stats

# Sample data (e.g., average customer spending)
sample_data = np.random.normal(loc=100, scale=15, size=50)  # Mean=100, Std=15, Sample Size=50

# Calculate sample mean and standard error
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)
confidence_level = 0.95
margin_of_error = stats.t.ppf((1 + confidence_level) / 2, len(sample_data) - 1) * (sample_std / np.sqrt(len(sample_data)))

# Confidence Interval
lower_bound = sample_mean - margin_of_error
upper_bound = sample_mean + margin_of_error

print(f"\n95% Confidence Interval: ({lower_bound:.2f}, {upper_bound:.2f})")
```


📊 **Output**: A **confidence interval** that estimates the population mean.  

✅ **When to Use?**  
- ✔ When making **predictions based on sample data**.  
- ✔ **Uncertainty quantification in business & finance**.

# 🔍 **2. Hypothesis Testing: Validating Assumptions**  

✅ **Definition**: Hypothesis testing helps in **making decisions** by testing claims about a population based on sample data.  

📌 **Key Terms in Hypothesis Testing**  

| **Term** | **Definition** | **Example** |
|----------|--------------|------------|
| **Null Hypothesis (H₀)** ❌ | No effect or difference | "Ad spending has no impact on sales" |
| **Alternative Hypothesis (H₁)** ✅ | There is an effect or difference | "Higher ad spending increases sales" |
| **p-value** 📉 | Probability of observing data assuming H₀ is true | p < 0.05 means rejecting H₀ |
| **Significance Level (α)** 🎯 | Threshold for rejecting H₀ | Typically 0.05 (5%) |
| **Type I Error (False Positive)** ⚠️ | Rejecting H₀ when it's actually true | Saying a drug works when it doesn’t |
| **Type II Error (False Negative)** 🚨 | Failing to reject H₀ when it's false | Saying a drug doesn’t work when it actually does |

📌 **Python Code: Hypothesis Testing Using T-Test**  

```python

# Sample Data: Sales before & after ad campaign
before_ads = np.random.normal(5000, 500, 30)
after_ads = np.random.normal(5500, 500, 30)

# Perform independent t-test
t_stat, p_value = stats.ttest_ind(before_ads, after_ads)

# Print results
print(f"\nT-statistic: {t_stat:.3f}, P-value: {p_value:.5f}")

# Conclusion
if p_value < 0.05:
    print("\nReject H₀: Ad spending significantly affects sales")
else:
    print("\nFail to reject H₀: No significant impact of ad spending on sales")
```



📊 **Output**:  
- If **p < 0.05** → Reject H₀ (Significant difference).  
- If **p > 0.05** → Fail to reject H₀ (No strong evidence).  

✅ **When to Use?**  
- ✔ **A/B Testing (Marketing, UX Design, Web Testing)**.  
- ✔ **Medical Trials & Scientific Research**.

# 📉 **3. Correlation & Regression: Understanding Relationships**  

✅ **Definition**: Measures the **strength & direction** of relationships between variables.  

📌 **Correlation vs. Regression**  

| **Method** | **Definition** | **Example** |
|------------|---------------|------------|
| **Correlation** 🔗 | Measures association strength (not causation) | "Is there a link between income and spending?" |
| **Linear Regression** 📈 | Predicts dependent variable (Y) using independent variable (X) | "Does ad spending predict sales revenue?" |

📌 **Python Code: Correlation & Regression**  

```python

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Simulated data
X = np.random.normal(50, 10, 100).reshape(-1, 1)  # Independent Variable
y = (3 * X + np.random.normal(0, 5, 100)).flatten()  # Dependent Variable

# Ensure both arrays have the same length
X = X[:100]
y = y[:100]

# Scatter Plot
sns.scatterplot(x=X.flatten(), y=y)
plt.title("Ad Spend vs. Sales Revenue")
plt.xlabel("Ad Spend")
plt.ylabel("Sales")
plt.show()

# Correlation Coefficient
correlation = np.corrcoef(X.flatten(), y)[0, 1]
print(f"Correlation: {correlation:.2f}")

# Regression Model
model = LinearRegression()
model.fit(X, y)
print(f"\nRegression Coefficient (Slope): {model.coef_[0]:.2f}")
```


📊 **Output**:  
- **Scatter plot showing relationship**.  
- **Correlation value** (-1 to +1) indicating strength & direction.  
- **Regression coefficient** for predicting outcomes.  

✅ **When to Use?**  
✔ **Predicting trends in sales, finance, healthcare**.  
✔ **Understanding relationships in datasets**.

# 🔄 **4. Comparing Inferential vs. Descriptive Statistics**  

| **Feature** | **Descriptive Statistics** 📊 | **Inferential Statistics** 🔍 |
|------------|---------------------|---------------------|
| **Definition** | Summarizes dataset features | Makes predictions from data |
| **Focus** | Central tendency, dispersion | Estimation, hypothesis testing |
| **Use Case** | Understanding dataset properties | Generalizing to a population |
| **Example** | "Average income is $50,000" | "95% of incomes fall between $45,000-$55,000" |

✅ **Inferential statistics makes predictions, while descriptive statistics summarizes data!**
