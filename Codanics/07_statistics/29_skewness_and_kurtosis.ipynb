# 📊 **`Skewness & Kurtosis in Statistics for Data Science`** 📈  

## 🔹 **What are Skewness & Kurtosis?**  

✅ **Skewness** → Measures **asymmetry** of the data distribution.  
✅ **Kurtosis** → Measures **how heavy or light the tails** of a distribution are compared to a normal distribution.  

📌 **Why are Skewness & Kurtosis Important?**  
✔ **Understanding data shape** before applying statistical models.  
✔ **Detecting outliers & anomalies**.  
✔ **Assessing normality** for machine learning algorithms.  

---

# 📊 **1. Skewness: Measuring Asymmetry**  

✅ **Definition**: Skewness measures how **symmetrical** or **asymmetrical** a distribution is around its mean.  

📌 **Types of Skewness**  
| **Type** | **Description** | **Graph Shape** |
|---------|----------------|---------------|
| **Symmetric (Zero Skewness)** | Mean ≈ Median ≈ Mode | 🔔 Bell-shaped (Normal) |
| **Positive Skew (Right Skewed)** | Mean > Median > Mode (Tail on the right) | 📈 Long tail on the right |
| **Negative Skew (Left Skewed)** | Mean < Median < Mode (Tail on the left) | 📉 Long tail on the left |

📌 **Formula for Skewness (Pearson’s Skewness Coefficient)**  
$$
\text{Skewness} = \frac{\sum (X_i - \bar{X})^3}{n \cdot \sigma^3}
$$
where:  
- $X_i$ = Data points  
- $\bar{X}$ = Mean  
- $n$ = Number of observations  
- $\sigma$ = Standard Deviation  

📌 **Python Code to Calculate Skewness**  





```python

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Simulated data
data = np.random.normal(50, 10, 1000)  # Normal distribution
skew_value = stats.skew(data)
print("\nSkewness:", skew_value)

# Plot histogram
plt.hist(data, bins=30, color='blue', edgecolor='black', alpha=0.6)
plt.title("Histogram of Data")
plt.show()
```


📊 **Output**:  
- If **Skewness ≈ 0**, data is **normally distributed**.  
- If **Skewness > 0**, data is **right-skewed**.  
- If **Skewness < 0**, data is **left-skewed**.  

✅ **Interpreting Skewness**  
✔ **|Skewness| < 0.5** → **Fairly symmetric**.  
✔ **0.5 < |Skewness| < 1** → **Moderately skewed**.  
✔ **|Skewness| > 1** → **Highly skewed**.

# 📈 **2. Kurtosis: Measuring Tails & Peakedness**  

✅ **Definition**: Kurtosis measures whether data is **heavy-tailed (many outliers) or light-tailed (few outliers)** compared to a normal distribution.  

📌 **Types of Kurtosis**  

| **Type** | **Description** | **Graph Shape** |
|---------|----------------|---------------|
| **Mesokurtic (Normal Kurtosis = 3)** | Standard normal distribution | 🔔 Medium peak, normal tails |
| **Leptokurtic (High Kurtosis > 3)** | Heavy tails (many outliers) | 📌 Tall peak, fat tails |
| **Platykurtic (Low Kurtosis < 3)** | Light tails (few outliers) | 🏔️ Flat peak, thin tails |

📌 **Formula for Kurtosis**  
$$
\text{Kurtosis} = \frac{\sum (X_i - \bar{X})^4}{n \cdot \sigma^4}
$$

📌 **Python Code to Calculate Kurtosis** 

```python

kurtosis_value = stats.kurtosis(data)  # Default Fisher's definition (Kurtosis - 3)
print("\nKurtosis:", kurtosis_value)
```


📊 **Output**:  
- **Kurtosis ≈ 0** → **Mesokurtic** (Normal).  
- **Kurtosis > 0** → **Leptokurtic** (Heavy tails, more outliers).  
- **Kurtosis < 0** → **Platykurtic** (Light tails, fewer outliers).  

✅ **Interpreting Kurtosis**  

- ✔ **Kurtosis ≈ 0 (Mesokurtic)** → Normal distribution.  
- ✔ **Kurtosis > 0 (Leptokurtic)** → More extreme values (outliers).  
- ✔ **Kurtosis < 0 (Platykurtic)** → Less extreme values (light tails).

# 📉 **3. Skewness & Kurtosis in Machine Learning & Data Science**  

📌 **Business & Finance**  
- ✔ **Stock price returns** → High kurtosis means extreme market movements (financial crashes).  
- ✔ **Income distribution** → Right-skewed, few people earn much more than average.  

📌 **Machine Learning & AI**  
- ✔ **Feature scaling** → If data is highly skewed, **log transformation** or **Box-Cox transformation** is needed.  
- ✔ **Algorithm selection** → Linear models assume **normality (zero skewness, mesokurtic kurtosis)**.  

📌 **Healthcare & Medicine**  
- ✔ **Patient response times** → Can be right-skewed if most recover quickly but a few take much longer.  
- ✔ **Hospital stay durations** → Usually follow a Poisson or exponential distribution (right-skewed).  



----



# 🔄 **4. How to Handle Skewness & Kurtosis?**  

✅ **For Right-Skewed Data (Positive Skewness)**  
- ✔ Apply **log transformation**  
- ✔ Use **square root transformation**  
- ✔ Try **Box-Cox transformation**  

✅ **For Left-Skewed Data (Negative Skewness)**  
- ✔ Use **power transformation (e.g., squaring values)**  


📌 **Python Code to Normalize Skewed Data (Log Transformation)**
  

```python

import numpy as np
import matplotlib.pyplot as plt

data_transformed = np.log(data[data > 0])  # Log transformation for right-skewed data
plt.hist(data_transformed, bins=30, color='green', edgecolor='black', alpha=0.6)
plt.title("Histogram After Log Transformation")
plt.show()
```


📊 **Output**: The skewness **reduces**, making the distribution more normal.  

✅ **For High Kurtosis (Heavy Tails, Many Outliers)**  
- ✔ Use **Winsorization** (limit extreme values).  
- ✔ Apply **Robust Scalers** (e.g., Median Absolute Deviation).  



----



# 📌 **5. Summary: Skewness vs. Kurtosis**  

| **Metric** | **Definition** | **Key Interpretation** | **Fixing Strategies** |
|-----------|--------------|----------------|-----------------|
| **Skewness** 📊 | Measures asymmetry | **Right Skew** → Long right tail, **Left Skew** → Long left tail | Log/square root transformation |
| **Kurtosis** 📈 | Measures tail heaviness | **Leptokurtic (High)** → More outliers, **Platykurtic (Low)** → Fewer outliers | Winsorization, Robust Scaling |

### **Find Skewness and Kurtosis**

```python

# Import libraries
import numpy as np
from scipy.stats import skew, kurtosis

# Generating sample data
np.random.seed(0)
data = np.random.normal(0, 1, 1000)  # Normal distribution

# Calculating skewness and kurtosis
data_skewness = skew(data)
data_kurtosis = kurtosis(data, fisher=True) # "fisher=True" gives exact kurtosis

# Displaying the results
print(f"\nSkewness of Sample Dataset: {data_skewness}")
print(f"\nKurtosis of Sample Dataset: {data_kurtosis}")
```

```python

# create histplot of sample data
sns.histplot(data=data,kde=True,color="g")
plt.show()
 
```
