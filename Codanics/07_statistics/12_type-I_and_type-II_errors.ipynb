# 🚦 **`Type-I and Type-II Errors in Statistics for Data Science`** 📊  

## 🔹 **What Are Type-I and Type-II Errors?**  

In **statistics & data science**, we often conduct **hypothesis testing** to make decisions. However, there’s always a chance of making an **incorrect decision** due to randomness or data limitations.  

✅ **Type-I and Type-II errors** occur when we **wrongly reject or wrongly accept** a hypothesis.  

| **Error Type**  | **Definition** | **Real-World Example** |
|--------------|-------------------|-------------------------|
| **Type-I Error (False Positive) 🚨** | Rejecting a **true** null hypothesis (**H₀**) | A COVID test says **"Positive"** when the person is actually **negative**. |
| **Type-II Error (False Negative) ⚠️** | Failing to reject a **false** null hypothesis (**H₀**) | A cancer test says **"No Cancer"** when the patient actually **has cancer**. |

---

# 🔹 **1. Understanding Hypothesis Testing** 🎯  

Before diving into errors, let's recap **hypothesis testing**:  

- **Null Hypothesis (H₀):** The assumption that **nothing has changed** (e.g., "The new drug has no effect").  
- **Alternative Hypothesis (H₁):** The assumption that **something has changed** (e.g., "The new drug works better than the old one").  

In hypothesis testing, we:  
✔ Accept or reject **H₀** based on **statistical evidence**.  
✔ Use a **p-value** (probability of observing results given that H₀ is true).  

---

# 🔹 **2. Type-I Error (False Positive) 🚨**  

✅ **Definition**: **Rejecting H₀ when it is actually true**.  
🔍 **False Alarm** – We wrongly detect an effect when **none exists**.  

✅ **Mathematical Representation**:  

$$
\alpha = P(\text{Rejecting H₀ | H₀ is true})
$$

where **α** (alpha) is the **significance level** (usually **0.05 or 5%**).  

✅ **Example Scenarios**:  
- 🔬 **Medical Test**: A **pregnancy test** says "positive," but the woman is **not pregnant**.  
- 🚔 **Fraud Detection**: A **bank flags a genuine transaction as fraud**.  
- 🎬 **Spam Filter**: A **legitimate email is marked as spam**.  

✅ **How to Reduce Type-I Error?**  
✔ Lower the **significance level (α)** (e.g., use **0.01 instead of 0.05**).  
✔ Use **Bonferroni correction** if conducting **multiple tests**.  


📌 **Python Example: Reducing Type-I Error with Lower α**  


```python
from scipy.stats import norm

alpha = 0.05  # Default significance level
z_critical = norm.ppf(1 - alpha)  # Z-score for 95% confidence
print(f"Z-Critical Value for α={alpha}: {z_critical}")
```

📊 **Output**:  
`Z-Critical Value for α=0.05: 1.645` (For **95% confidence level**)  



---

# 🔹 **3. Type-II Error (False Negative) ⚠️**  

✅ **Definition**: **Failing to reject H₀ when it is actually false**.  
🔍 **Missed Detection** – We fail to detect an effect when **one actually exists**.  

✅ **Mathematical Representation**:  

$$
beta = P(\text{Failing to reject H₀ | H₀ is false})
$$

where **β (beta)** is the probability of making a Type-II error.  

✅ **Example Scenarios**:  
- 🏥 **Cancer Screening**: A test fails to detect **cancer in a patient who has it**.  
- 🏦 **Loan Fraud Detection**: A fraud detection system **misses an actual fraud case**.  
- 📡 **Cybersecurity**: An **intrusion detection system fails to flag a hacker attack**.  

✅ **How to Reduce Type-II Error?**  
✔ **Increase sample size (n)** – More data means better detection.  
✔ **Increase significance level (α)** – A higher **α (e.g., 0.10 instead of 0.05)** lowers **β**.  
✔ **Improve test power (1 - β)** – Use **power analysis** to find optimal sample size.  


📌 **Python Example: Power Analysis to Reduce Type-II Error**  


```python
from statsmodels.stats.power import TTestIndPower

# Parameters: effect size, alpha level, power, sample size
analysis = TTestIndPower()
sample_size = analysis.solve_power(effect_size=0.5, alpha=0.05, power=0.8)
print(f"Required Sample Size to Reduce Type-II Error: {round(sample_size)}")
```

📊 **Output**:  
`Required Sample Size to Reduce Type-II Error: 64` (For **80% power** at α=0.05)

# 🔹 **4. Type-I vs. Type-II Errors (Comparison Table) 📊**  

| **Feature**  | **Type-I Error (False Positive) 🚨** | **Type-II Error (False Negative) ⚠️** |
|--------------|--------------------------------|--------------------------------|
| **Definition** | Rejecting a **true** null hypothesis (H₀) | Failing to reject a **false** null hypothesis (H₀) |
| **Mistake** | Detecting an effect that **doesn't exist** | Missing an effect that **actually exists** |
| **Symbol** | **α (alpha)** | **β (beta)** |
| **Example (Medical Test)** | Test **incorrectly says "Positive"** (not actually sick) | Test **incorrectly says "Negative"** (actually sick) |
| **Example (Fraud Detection)** | **Flagging a real user as fraud** | **Missing an actual fraud case** |
| **Impact** | **False alarms** lead to unnecessary actions | **Missed detections** lead to dangerous consequences |
| **How to Reduce?** | Lower **α (e.g., 0.01 instead of 0.05)** | Increase **sample size** & test **power (1 - β)** |



---

# 🔹 **5. Balancing Type-I and Type-II Errors (Tradeoff) ⚖️**  

Increasing **α (allowing more Type-I errors)** **reduces** Type-II errors.  
Reducing **α (allowing fewer Type-I errors)** **increases** Type-II errors.  

✅ **Choosing α & β Depends On Context**:  
- **Medical Tests 🏥**: **Reduce Type-II errors** (avoid missing diseases).  
- **Spam Filters 📩**: **Reduce Type-I errors** (avoid filtering real emails).  
- **Fraud Detection 💰**: Balance both, depending on cost of mistakes.  


📌 **Python Example: Visualizing Type-I vs. Type-II Errors**  


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# X-axis values (standard normal distribution)
x = np.linspace(-4, 4, 1000)

# Distributions for Null & Alternative Hypothesis
null_dist = norm.pdf(x, loc=0, scale=1)
alt_dist = norm.pdf(x, loc=1.5, scale=1)  # Shifted right

# Plotting
plt.plot(x, null_dist, label="Null Distribution (H₀)", color="blue")
plt.plot(x, alt_dist, label="Alternative Distribution (H₁)", color="red")

# Marking Type-I & Type-II Errors
plt.fill_between(x, null_dist, where=(x > 1), color='orange', alpha=0.3, label="Type-I Error (α)")
plt.fill_between(x, alt_dist, where=(x < 1), color='green', alpha=0.3, label="Type-II Error (β)")

plt.legend()
plt.title("Type-I vs. Type-II Error")
plt.xlabel("Test Statistic")
plt.ylabel("Probability Density")
plt.show()
```

📊 **Output**: A graph showing **Type-I (False Positives) in orange** and **Type-II (False Negatives) in green**.



---

# 🎯 **Final Takeaways**  
✔ **Type-I Error 🚨 (False Positive)** → Detecting an effect **when none exists** (α).  
✔ **Type-II Error ⚠️ (False Negative)** → Missing an effect **that actually exists** (β).  
✔ **Reducing α increases β**, and vice versa → **tradeoff based on problem context**.  
✔ **Larger sample sizes & statistical power** help **reduce Type-II errors**.
