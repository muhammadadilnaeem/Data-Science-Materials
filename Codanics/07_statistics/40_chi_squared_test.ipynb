# 📊 **`Chi-Square Test in Statistics for Data Science`** 🔍  

## 🔹 **What is the Chi-Square Test?**  

✅ **Definition**: The **Chi-Square (χ²) test** is a statistical method used to determine whether **categorical variables are independent or associated** in a dataset.  

✅ **Why is the Chi-Square Test Important in Data Science?**  
- ✔ **Analyzes categorical data** (e.g., gender, product category, purchase behavior).  
- ✔ **Used in A/B testing, market research, and medical studies**.  
- ✔ **Tests relationships between two categorical variables**.  

📌 **Example Question**:  
- "Is there a relationship between **customer gender and product preference**?"  
- "Are **education level and job type** independent?"  

---

# 🔥 **1. Types of Chi-Square Tests**  

| **Test Type** | **Purpose** | **Example** |
|--------------|------------|-------------|
| **Chi-Square Goodness-of-Fit Test** 📊 | Compares observed vs. expected frequencies in one categorical variable | "Do customer age groups follow a specific distribution?" |
| **Chi-Square Test for Independence** 🔗 | Tests if two categorical variables are related | "Is there an association between education level and job type?" |

✅ **Chi-Square Test for Independence is the most common in data science.**  

---

# 🔍 **2. Chi-Square Test for Independence: Step-by-Step**  

### **Scenario: Does Gender Affect Product Preference?**  
✔ We collect data on **500 customers** (Male/Female) and their **preference for two products (A/B)**.  
✔ We create a **contingency table** (cross-tabulation of categories).  
✔ We perform a **Chi-Square Test** to check if product preference is **independent of gender**.  

📌 **Step 1: Create the Contingency Table**  
| **Gender / Product** | **Product A** | **Product B** | **Total** |
|----------------|-----------|-----------|-------|
| **Male**   | 100  | 150  | 250  |
| **Female** | 120  | 130  | 250  |
| **Total** | 220  | 280  | 500  |

---

📌 **Step 2: Define Hypotheses**  

| **Hypothesis** | **Interpretation** |
|--------------|------------------|
| **Null Hypothesis (H₀)** ❌ | "Gender and product preference are independent (no relationship)." |
| **Alternative Hypothesis (H₁)** ✅ | "Gender and product preference are dependent (there is a relationship)." |

✅ **We check if observed differences are due to random chance or a true association.**  

---

📌 **Step 3: Perform the Chi-Square Test in Python**  

```python

import numpy as np
import scipy.stats as stats

# Create contingency table
observed = np.array([[100, 150], [120, 130]])  # Male vs. Female product choices

# Perform Chi-Square Test
chi2_stat, p_value, dof, expected = stats.chi2_contingency(observed)

# Print results
print(f"\nChi-Square Statistic: {chi2_stat:.3f}")
print(f"\nP-Value: {p_value:.5f}")
print(f"\nDegrees of Freedom: {dof}")
print("\nExpected Frequencies Table:\n", expected)

# Decision
if p_value < 0.05:
    print("\nReject H₀: Gender and product preference are related.")
else:
    print("\nFail to reject H₀: No significant relationship.")
```


📊 **Output**:  
✔ **Chi-Square Statistic** → Measures how far observed data deviates from expected data.  
✔ **p-value** → Probability of observing this data if H₀ is true.  
✔ **Degrees of Freedom (dof)** → \((rows - 1) \times (columns - 1)\).  
✔ **Expected Frequencies** → If gender and product preference were independent.  

✅ **Interpretation**:  
- If **p < 0.05**, reject H₀ → **Gender affects product preference**.  
- If **p > 0.05**, fail to reject H₀ → **No strong evidence of a relationship**.  
  


---

# 📉 **3. Assumptions of the Chi-Square Test**  

✅ **When Can You Use the Chi-Square Test?**  
- ✔ **Categorical data only** (e.g., Male/Female, Product A/B).  
- ✔ **Observations must be independent** (no repeated measures).  
- ✔ **Expected cell count should be at least 5** for accuracy.  

❌ **When NOT to Use the Chi-Square Test?**  
- 🚫 If data is **numerical or continuous** (use t-tests or regression instead).  
- 🚫 If expected frequencies are **too small** (use Fisher’s Exact Test).

# 📊 **4. Chi-Square Goodness-of-Fit Test**  

✅ **Purpose**: Tests if a **sample follows a known distribution**.  
✅ **Example**: Are **employee promotions equally distributed across departments?**  

📌 **Observed vs. Expected Frequencies**  

| **Department** | **Observed** | **Expected** |
|--------------|-------------|-------------|
| Sales       | 50          | 40          |
| IT          | 30          | 40          |
| HR          | 20          | 20          |

📌 **Python Code for Goodness-of-Fit Test** 

```python

# Observed and expected frequencies
observed = np.array([50, 30, 20])
expected = np.array([40, 40, 20])

# Perform Chi-Square Test
chi2_stat, p_value = stats.chisquare(observed, expected)

# Print results
print(f"\nChi-Square Statistic: {chi2_stat:.3f}")
print(f"\nP-Value: {p_value:.5f}")

# Decision
if p_value < 0.05:
    print("\nReject H₀: Promotions are not equally distributed.")
else:
    print("\nFail to reject H₀: No significant difference.")
```



📊 **Output**:  
- If **p < 0.05**, promotions **are NOT equally distributed**.  
- If **p > 0.05**, promotions **follow the expected distribution**.  

✅ **Used for:**  
- ✔ Checking if voting preferences match expected outcomes.  
- ✔ Analyzing customer purchases by brand share.

# 🖼 **`Chi-Squared Test`** 🕵️‍♀️

### **Import Libraries**

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
from scipy import stats
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

### **Load the dataset**

```python
df = sns.load_dataset("titanic")

df.head()
```

---

## **Perform Chi-Square Test**

- Only performed for *categorical variables*.

- For this we need to define a ***Null Hypothesis (H₀)*** and an ***Alternative Hypothesis (H₁)***.

- ***`Null Hypothesis (H₀)`*** : 
    - There is no significant association between gender (`"sex"`) and survival (`"survived"`) on Titanic.
    - This means any observed differences in survival rates between genders in the dataset are due to chance and not due to an underlying relationship. 


- ***`Alternative Hypothesis (H₁)`*** : 
    - There is a significant association between gender (`"sex"`) and survival (`"survived"`) on Titanic.
    - This implies that the differences in survival rates are not just due to chance but are influenced by **passanger's gender**. 

```python
# create a contingency table
contigency_table = pd.crosstab(df["sex"],df["survived"])
contigency_table
```

```python
# perform chi-squared test
chi2_stat, p_value, dof, excepted = stats.chi2_contingency(contigency_table)

# print the results
print("\nChi-Squared Stastistic: ",chi2_stat)
print("\nP-value: ",p_value)
print("\nDegree of Freedom: ",dof)
print(f"\nExpected: \n{excepted}")

# Decision
if p_value < 0.05:
    print("\nReject H₀: Survival Rates are independent of Passangers Gender.")
else:
    print("\nFail to reject H₀: Survival Rates are influenced by Passangers Gender.")
```
