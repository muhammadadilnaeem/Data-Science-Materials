# 📊 **`Standard Deviation vs. Standard Error in Statistics for Data Science`** 📈  

## 🔹 **What Are Standard Deviation and Standard Error?**  

✅ **Standard Deviation (SD, σ)** → Measures **spread of data** in a dataset.  
✅ **Standard Error (SE, SE of Mean)** → Measures **accuracy of a sample mean** as an estimate of the population mean.  

📌 **Key Differences**  

| **Concept**  | **Standard Deviation (σ, s)** 📊 | **Standard Error (SE)** 📉 |
|-------------|--------------------------------|-------------------------|
| **Definition** | Spread of data around the mean | Accuracy of sample mean as an estimate of population mean |
| **Formula** | $\sigma = \sqrt{\frac{\sum (X_i - \bar{X})^2}{N}}$ | $SE = \frac{\sigma}{\sqrt{n}}$ |
| **Symbol** | **$\sigma$ (population), $s$ (sample)** | **SE or SEM** |
| **Depends on** | Variability in the dataset | Sample size ($n$) |
| **Used for** | Understanding dispersion in a dataset | Estimating reliability of sample mean |
| **Effect of Increasing Sample Size** | **No change** | **Decreases (higher accuracy)** |

---

## 🔹 **1. Standard Deviation (σ, s) 📊**  

✅ **Definition**: Measures **how spread out data points** are around the mean.  
✅ **Formula**:  

### **1️⃣ Population Standard Deviation (σ)**
$$
\sigma = \sqrt{\frac{\sum (X_i - \mu)^2}{N}}
$$
where:  
- $X_i$ = Each data point  
- $\mu$ = Population mean  
- $N$ = Population size  

### **2️⃣ Sample Standard Deviation (s)**
$$
s = \sqrt{\frac{\sum (X_i - \bar{X})^2}{n - 1}}
$$
where:  
- $\bar{X}$ = Sample mean  
- $n$ = Sample size  

📌 **Example Calculation**  
📌 Data: **[10, 20, 30, 40, 50]**  
✔ **Step 1**: Mean = **30**  
✔ **Step 2**: Find squared differences:  
$$
(10-30)^2, (20-30)^2, (30-30)^2, (40-30)^2, (50-30)^2
$$
✔ **Step 3**: Compute SD:  
$$
\sigma = \sqrt{\frac{400 + 100 + 0 + 100 + 400}{5}} = 14.14
$$

📌 **Python Code for Standard Deviation**  



```python

import numpy as np

data = [10, 20, 30, 40, 50]

# Population Standard Deviation
pop_std = np.std(data)

# Sample Standard Deviation (ddof=1 for unbiased estimate)
sample_std = np.std(data, ddof=1)

print("\nPopulation Standard Deviation:", pop_std)
print("\nSample Standard Deviation:", sample_std)
```


📊 **Output**:  
- `Population Standard Deviation: 14.14`  
- `Sample Standard Deviation: 15.81`  

✅ **When to Use?**  
✔ **Understanding spread in datasets**.  
✔ **Feature scaling in Machine Learning**.  
✔ **Volatility analysis in finance**.  



----



## 🔹 **2. Standard Error (SE) 📉**  

✅ **Definition**: Measures **how accurately a sample mean estimates the population mean**.  
✅ **Formula**:  
$$
SE = \frac{\sigma}{\sqrt{n}}
$$

where:  
- $\sigma$ = Standard deviation  
- $n$ = Sample size

📌 **Effect of Sample Size (n)**  
✔ **Larger n → Smaller SE** (More accurate estimate).  
✔ **Smaller n → Larger SE** (Less accuracy).  

📌 **Example Calculation**  
📌 Sample Standard Deviation (s) = **15.81**  
📌 Sample Size (n) = **5**  

$$
SE = \frac{15.81}{\sqrt{5}} = \frac{15.81}{2.236} = 7.07
$$

📌 **Python Code for Standard Error Calculation**  


```python

import scipy.stats as stats

sample_size = 5
standard_error = sample_std / np.sqrt(sample_size)
print("\nStandard Error (SE):", standard_error)
```


📊 **Output**: `Standard Error: 7.07` ✅  

✅ **When to Use?**  
✔ **Estimating population mean from a sample**.  
✔ **Building confidence intervals**.  
✔ **Assessing reliability of survey results**.

## 🔹 **4. Visualizing Standard Deviation & Standard Error**  

📌 **Effect of Increasing Sample Size on SE**

```python

import matplotlib.pyplot as plt
import numpy as np

sample_sizes = np.arange(1, 101)  # Sample sizes from 1 to 100
std_dev = 15  # Assumed standard deviation
standard_errors = std_dev / np.sqrt(sample_sizes)

plt.plot(sample_sizes, standard_errors, label="Standard Error")
plt.xlabel("Sample Size (n)")
plt.ylabel("Standard Error")
plt.title("Effect of Sample Size on Standard Error")
plt.legend()
plt.show()
```


📊 **Output**:  
- **As sample size increases, SE decreases** (higher accuracy).
