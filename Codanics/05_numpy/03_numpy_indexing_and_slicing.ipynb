# **Numpy Indexing and Slicing**

--------

```python
# import numpy library
import numpy as np
```

```python
# create an array
a = np.array([1,2,3,4,5])  
```

```python
a[0] # to view first element
```

```python
a[-1] # to view last element
```

```python
b = np.arange(0,12,2)
c = b.reshape(2,3)
c
```

```python
a[(a<5)|(a>7)]
```

```python
# vstack

a = np.array([1,2,3,4,5]) 
b = np.array([6,7,8,9,10])
c = np.vstack((a,b))
c 
```

```python
# hstack

a = np.array([1,2,3,4,5]) 
b = np.array([6,7,8,9,10])
c = np.hstack((a,b))
c 
```
