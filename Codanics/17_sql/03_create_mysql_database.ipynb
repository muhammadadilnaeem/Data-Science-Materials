## **Create a Database using ___Mysql___**

-------

- First go to command prompt, to access ___mysql___ type this command

```bash
mysql -u root -p
```


- Press __Enter__ and then type ___password___



- A screen will look like this



- Now if you want to know how many databases are there in your system use this command and press ___Enter___.
 
```bash
show databases;
```


- Now in order to create a ___database___ use this command:

```bash
create database <name>;
```



- Now in order to access this database use this command:

```bash
use <name>;
```


- In order to delete a specific Database use this command;

```bash
DROP DATABASE prcticedb;
```
