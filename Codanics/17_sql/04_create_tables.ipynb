## **Create Tables in Mysql Database**

- In order to create tables in database first use that database;

```bash
USE practice_db;  
```



- Now in order to see tables in this database use this command;
  
```bash
SHOW tables;  
```


- Now in order to create table;

```bash
CREATE TABLE students (
    stuID int,
    stuNAME varchar(45),
    stuAGE int,
    stuCITY varchar(50)
);
```




- In order to see table structure use this;
 
```bash
DESC students;  
```


- Now in order to add content into table;

```bash
INSERT INTO students (
    stuID,
    stuNAME,
    stuAGE,
    stuCITY
) VALUES (
    1,
    'Adil',
    22,
    'Lahore'
); 
```


- Now in Order to see the entered data use this;
```bash
SELECT * FROM students;
```



- Add few more entries




- Now we can work with different commands;

```bash
SELECT * FROM students where stuAGE >=24;
```  


- Add a column in table;

```bash
 ALTER TABLE students ADD COLUMN stuCOUNTRY TEXT; 
```

- Now in Order to set Country we can do it one by 1 with this command;

```bash
UPDATE students SET stuCOUNTRY = "PAKISTAN" WHERE stuID=1;
```

- Or you Can update all the entries with certain condition;

```bash
UPDATE students SET stuCOUNTRY = "PAKISTAN" WHERE stuAGE>20;
```




- Now if you want to delete some entries with certain condition use this;

```bash
DELETE FROM students WHERE stuAGE>27; 
```




- To check distinct values in certain column use this command;

```bash
SELECT DISTINCT stuAGE FROM students;
``` 



- In order to count how many time a particular age exist in my column use this command;

```bash
SELECT stuAGE, COUNT(*) AS age_count
FROM students
GROUP BY stuAGE
ORDER BY stuAGE;
```
