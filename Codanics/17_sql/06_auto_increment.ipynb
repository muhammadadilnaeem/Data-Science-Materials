## ***Auto Increment in Mysql***

In SQL, **autoincrement** is a feature used to automatically generate a unique value for a column in a database table, typically for a primary key. When a new record is inserted into the table, the autoincrement column automatically increases its value based on a defined increment. This ensures that each record has a unique identifier without requiring manual input.

### **Key Points:**

1. **Primary Key**: Autoincrement is often used for primary key columns to ensure uniqueness.
  
2. **Syntax**: The syntax for defining an autoincrement column varies between database systems. Here are examples:

   - **MySQL**:
     ```sql
     CREATE TABLE Users (
         id INT AUTO_INCREMENT PRIMARY KEY,
         name VARCHAR(100)
     );
     ```

   - **PostgreSQL**:
     ```sql
     CREATE TABLE Users (
         id SERIAL PRIMARY KEY,
         name VARCHAR(100)
     );
     ```

   - **SQLite**:
     ```sql
     CREATE TABLE Users (
         id INTEGER PRIMARY KEY AUTOINCREMENT,
         name TEXT
     );
     ```

3. **Increment Value**: By default, the increment is usually set to 1, but it can often be configured to a different value.

4. **Handling Overflows**: If the maximum value is reached, an error will occur unless the data type is changed to accommodate larger values.

5. **Usage**: It simplifies the process of inserting records since you don’t need to manually specify a value for the autoincrement column.

### **Example Usage:**

```sql
INSERT INTO Users (name) VALUES ('Alice');
INSERT INTO Users (name) VALUES ('Bob');
```

In this example, the `id` column will automatically be assigned values like 1, 2, etc., for each new entry.

### **AUTO_INCREMENT**

```bash
CREATE TABLE students(
    stuID int NOT NULL AUTO_INCREMENT,
    stuNAME varchar(255),
    stuAGE int,
    PRIMARY KEY (stuID)
);
```



- We do not need to add stuID as it will be auto increment





- You can set auto increment properly

- Let's delete a row

```bash
DELETE FROM students WHERE stuID = 3;
```

- Now sql will automatically stuID;

```bash
SET @count = 0;
UPDATE students SET stuID = @count := @count + 1;
ALTER TABLE students AUTO_INCREMENT = 1;
```
