```python
# import required libraries
import mysql.connector
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
# Step 1: Connect to your MySQL database
db_config = {
    'host': 'localhost',          # e.g., 'localhost' or '127.0.0.1'
    'user': 'root',      # your MySQL username
    'password': '******',  # your MySQL password
    'database': '******'   # your MySQL database name
}

conn = mysql.connector.connect(**db_config)
```

```python
# Step 2: Read the 'payments' table into a pandas DataFrame
payments_df = pd.read_sql("SELECT * FROM payments", conn)
```

```python
# Step 3: Save the DataFrame to a CSV file
payments_df.to_csv('../practice_dataset/payments.csv', index=False)
```

```python
# Close the connection
conn.close()

print("Payments table has been extracted to 'payments.csv'")
```

### **Load the dataset**

```python
df = pd.read_csv(r"../practice_dataset/payments.csv")
```

```python
# view first 5 rows of dataset
df.head()
```

```python
# plot a fancy histplot on amount column
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.histplot(df['amount'], kde=True)
plt.title('Distribution of Amount')
plt.show()
```

```python
# now plot a fancy boxen plot
plt.figure(figsize=(10, 6))
sns.boxenplot(df['amount'])
plt.title('Boxen plot of Amount')
plt.show()
```
