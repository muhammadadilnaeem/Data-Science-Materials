## **Installation For ___SQL___**



To install and set up MySQL on your Windows operating system, follow these steps:

---

### **Step 1: Download MySQL Installer**
1. Go to the [MySQL official website](https://dev.mysql.com/downloads/installer/).
   


2. Select **"MySQL Installer for Windows"**.
3. Choose the appropriate version (e.g., web or full version) and click **Download**.
4. Sign in or click **No thanks, just start my download** to proceed without signing in.

---

### **Step 2: Run the Installer**
1. Locate the downloaded installer file (e.g., `mysql-installer-community-x.x.x.x.msi`) and double-click it to launch.
2. On the setup type screen, choose an installation type:
   - **Developer Default**: Installs all MySQL products.
   - **Server Only**: Installs only the MySQL server.
   - **Custom**: Allows you to choose specific components.



For a beginner, **Developer Default** is recommended.

---

### **Step 3: Install MySQL Products**
1. The installer will check for requirements. Install any missing prerequisites (e.g., Visual Studio Redistributable).
2. Click **Execute** to download and install the selected MySQL components.
3. Once completed, click **Next** to proceed.

















---

### **Step 4: Configure MySQL Server**
1. On the **Type and Networking** page:
   - Select the configuration type (e.g., **Standalone MySQL Server**).
   - Set the port number (default is `3306`).
2. Click **Next**.



3. On the **Authentication Method** page:
   - Choose **Use Strong Password Encryption for Authentication** (recommended).
   - Click **Next**.






4. On the **Accounts and Roles** page:
   - Set a root password (remember this for later use).
   - Optionally, create additional MySQL user accounts.



5. On the **Windows Service** page:
   - Configure MySQL as a Windows Service (default setting).
   - Choose the service startup type (recommended: **Automatic**).

6. Click **Next**, then **Execute** to apply the configuration.























---

### **Step 5: Start MySQL Workbench and Connect**
1. After the setup, the installer will offer to launch **MySQL Workbench** and **MySQL Shell**.
2. Open **MySQL Workbench**.
3. Click on the **+** icon next to "MySQL Connections."
4. Enter:
   - **Connection Name**: Any name you like.
   - **Hostname**: `localhost` (default).
   - **Port**: `3306` (default).
   - **Username**: `root`.
   - Click **Test Connection** and enter the root password to verify the connection.





---

### **Step 6: Verify Installation**
1. Open MySQL Workbench and run a simple query to confirm the server is working:
   ```sql
   SELECT VERSION();
   ```
2. If the query runs successfully, your MySQL installation is complete.
