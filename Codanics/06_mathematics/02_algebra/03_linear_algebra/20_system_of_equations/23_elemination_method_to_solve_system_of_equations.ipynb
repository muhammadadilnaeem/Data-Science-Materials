# 🌟 **Elimination Method to Solve a System of Equations** 🔢📊  

### **Introduction**  
The **elimination method** (also called the **addition method**) is a technique for solving a **system of equations** by eliminating one variable, allowing us to solve for the other. It is **systematic and efficient**, especially for **2 or 3 variables**.  

✔️ **Best for**: Small to medium-sized systems of **linear equations**.  
⚠️ **For large systems**: Matrix-based methods (Gaussian elimination, LU decomposition) are more efficient.  

---

## 🌟 **1. Steps for the Elimination Method**  

1. **Multiply** one or both equations (if needed) to align coefficients for elimination.  
2. **Add or subtract** the equations to eliminate one variable.  
3. **Solve for the remaining variable**.  
4. **Back-substitute** to find the other variable.  
5. **Verify** the solution by plugging values into the original equations.  

---

## 🌟 **2. Example Problem**  

Solve the system:  
$$
\begin{cases}
2x + 3y = 8 \\
x - 4y = -5
\end{cases}
$$  

### **Step 1: Align the Coefficients**  
Multiply the second equation by **2** so that the $ x $-coefficients match:  
$$
\begin{cases}
2x + 3y = 8 \\
2x - 8y = -10
\end{cases}
$$  

---

### **Step 2: Subtract the Equations**  
Eliminate $ x $ by subtracting the second equation from the first:  
$$ 
(2x + 3y) - (2x - 8y) = 8 - (-10) 
$$  
$$ 
2x + 3y - 2x + 8y = 8 + 10 
$$  
$$ 
11y = 18 
$$  
$$ 
y = \frac{18}{11} 
$$  

---

### **Step 3: Solve for $ x $**  
Substituting $ y = \frac{18}{11} $ into the second equation:  
$$ 
x - 4 \left(\frac{18}{11}\right) = -5 
$$  
$$ 
x - \frac{72}{11} = -5 
$$  
$$ 
x = -5 + \frac{72}{11} 
$$  
$$ 
x = \frac{-55 + 72}{11} = \frac{17}{11} 
$$  

### **Solution:**  
$$ x = \frac{17}{11}, \quad y = \frac{18}{11} $$  

---

## 🌟 **3. Python Implementation**
```python
import numpy as np

# Define the coefficient matrix A and result vector b
A = np.array([[2, 3], [1, -4]])  
b = np.array([8, -5])

# Solve using NumPy's built-in linear equation solver
solution = np.linalg.solve(A, b)

# Print the solution
print("Solution (x, y):", solution)
```
