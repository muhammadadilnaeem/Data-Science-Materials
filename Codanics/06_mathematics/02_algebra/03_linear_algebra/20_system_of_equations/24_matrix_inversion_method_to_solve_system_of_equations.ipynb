# 🌟 **Matrix Inversion Method to Solve a System of Equations in Data Science** 🔢📊  

### **Introduction**  
The **Matrix Inversion Method** is a technique for solving **systems of linear equations** using the inverse of the coefficient matrix. It is **efficient for small systems** but can be **computationally expensive** for large datasets.  

✔️ **Best for**: Small systems with **square, non-singular matrices**.  
⚠️ **Not practical**: For large matrices (use Gaussian elimination or LU decomposition).  

---

## 🌟 **1. Understanding the Matrix Inversion Method**  

A system of linear equations can be represented in **matrix form**:  
$$ 
\mathbf{A} \cdot \mathbf{x} = \mathbf{b} 
$$  
where:  
- $ \mathbf{A} $ = **coefficient matrix**  
- $ \mathbf{x} $ = **variable vector**  
- $ \mathbf{b} $ = **result vector**  

### **Solution Using Matrix Inversion**  
If $ \mathbf{A} $ is **invertible**, the solution is given by:  
$$ 
\mathbf{x} = \mathbf{A}^{-1} \cdot \mathbf{b} 
$$  
where $ \mathbf{A}^{-1} $ is the **inverse of matrix $ \mathbf{A} $**.  

---

## 🌟 **2. Example Problem**  

Solve the system:  
$$
\begin{cases}
2x + 3y = 8 \\
x - 4y = -5
\end{cases}
$$  

### **Step 1: Express in Matrix Form**
$$ 
\begin{bmatrix} 
2 & 3 \\ 
1 & -4 
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y 
\end{bmatrix} 
=
\begin{bmatrix} 
8 \\ 
-5 
\end{bmatrix} 
$$  
where:  
- $ \mathbf{A} = \begin{bmatrix} 2 & 3 \\ 1 & -4 \end{bmatrix} $  
- $ \mathbf{b} = \begin{bmatrix} 8 \\ -5 \end{bmatrix} $  

---

### **Step 2: Compute the Inverse of $ \mathbf{A} $**  
For a **2×2 matrix**:  
$$ 
\mathbf{A}^{-1} = \frac{1}{\text{det}(\mathbf{A})} 
\begin{bmatrix} 
d & -b \\ 
-c & a 
\end{bmatrix} 
$$  
where $ \text{det}(\mathbf{A}) = ad - bc $.  

For our example:  
- **Determinant**:  
  $$ \text{det}(\mathbf{A}) = (2)(-4) - (3)(1) = -8 - 3 = -11 $$  

- **Inverse of $ \mathbf{A} $**:  
  $$ 
  \mathbf{A}^{-1} = \frac{1}{-11} 
  \begin{bmatrix} 
  -4 & -3 \\ 
  -1 & 2 
  \end{bmatrix} 
  =
  \begin{bmatrix} 
  4/11 & 3/11 \\ 
  1/11 & -2/11 
  \end{bmatrix} 
  $$  

---

### **Step 3: Compute $ \mathbf{x} = \mathbf{A}^{-1} \cdot \mathbf{b} $**  
$$ 
\begin{bmatrix} 
4/11 & 3/11 \\ 
1/11 & -2/11 
\end{bmatrix} 
\begin{bmatrix} 
8 \\ 
-5 
\end{bmatrix} 
$$  

Multiplying:  
$$ 
x = \frac{4}{11} \times 8 + \frac{3}{11} \times (-5) = \frac{32}{11} - \frac{15}{11} = \frac{17}{11} 
$$  
$$ 
y = \frac{1}{11} \times 8 + \frac{-2}{11} \times (-5) = \frac{8}{11} + \frac{10}{11} = \frac{18}{11} 
$$  

### **Solution:**  
$$ 
x = \frac{17}{11}, \quad y = \frac{18}{11} 
$$  

---

## 🌟 **3. Python Implementation**  

```python
import numpy as np

# Define coefficient matrix A and result vector b
A = np.array([[2, 3], [1, -4]])
b = np.array([8, -5])

# Compute the inverse of A
A_inv = np.linalg.inv(A)

# Solve for x
x = np.dot(A_inv, b)

print("Solution (x, y):", x)
```

✔️ **Output:**  
```
Solution (x, y): [1.54545455 1.63636364]
```
(Equivalent to $ x = 17/11, y $---

## 🌟 **4. Cases in Matrix Inversion Method**  
| **Case** | **Matrix Condition** | **Solution Type** |
|----------|----------------------|----------------|
| **Unique Solution** | $ \text{det}(\mathbf{A}) \neq 0 $ | One solution (consistent & independent) |
| **Infinite Solutions** | Rows of $ \mathbf{A} $ are dependent | Infinite solutions (dependent) |
| **No Solution** | $ \mathbf{A} $ is singular (det = 0) | No solution (inconsistent) |

## 🌟 **5. Applications in Data Science**  

### **1. Linear Regression (Solving for Coefficients)**
The **normal equation** in linear regression is:  
$$ 
\mathbf{\beta} = (\mathbf{X}^\top \mathbf{X})^{-1} \mathbf{X}^\top \mathbf{y} 
$$  
where:
- $ \mathbf{X} $ = Feature matrix.
- $ \mathbf{\beta} $ = Coefficients.
- $ \mathbf{y} $ = Target values.

```python
# Example: Solve Linear Regression Coefficients
X = np.array([[1, 1], [1, 2], [1, 3]])
y = np.array([2, 2.5, 3.5])

# Compute coefficients using the normal equation
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Regression Coefficients:", beta)
```

## 🌟 **6. Advantages & Disadvantages of the Matrix Inversion Method**

| **Feature** | **Pros** | **Cons** |
|------------|---------|---------|
| **Exact Solution** | Works for **small systems** | Computationally expensive for **large matrices** |
| **Efficient for $ 2 \times 2 $ or $ 3 \times 3 $ systems** | Easy to implement using NumPy | Not scalable for high dimensions |
| **Used in Regression & AI** | Used in machine learning & optimization | If $ \mathbf{A} $ is **singular**, no inverse exists |

2. **Solution**:  
   $$ \mathbf{x} = \mathbf{A}^{-1} \cdot \mathbf{b} $$  
3. **Used in**: **Linear regression, AI, and optimization**.  
4. **Limitations**: **Computationally expensive** for large matrices.
