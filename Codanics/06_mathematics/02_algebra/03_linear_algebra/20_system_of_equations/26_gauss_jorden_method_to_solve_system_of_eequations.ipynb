# 🌟 **Gauss-Jordan Elimination Method to Solve a System of Equations in Data Science** 🔢📊  

### **Introduction**  
**Gauss-Jordan Elimination** is an extension of **Gaussian Elimination** that transforms a system of equations into **reduced row echelon form (RREF)**, making it easier to solve for variables. It is **useful for exact solutions** in small to medium-sized systems and forms the basis for **matrix inversion**.  

✔️ **Best for**: Small to medium-sized systems of **linear equations**.  
⚠️ **Not ideal for large systems** (use LU decomposition or iterative methods).  

---

## 🌟 **1. Understanding Gauss-Jordan Elimination**  

A system of equations is represented in **matrix form**:  
$$ 
\mathbf{A} \cdot \mathbf{x} = \mathbf{b} 
$$  
where:  
- $ \mathbf{A} $ = **coefficient matrix**  
- $ \mathbf{x} $ = **variable vector**  
- $ \mathbf{b} $ = **result vector**  

The **goal** of Gauss-Jordan Elimination is to **convert $ \mathbf{A} $ into the identity matrix** $ \mathbf{I} $, allowing us to directly read the solution from $ \mathbf{b} $.

---

## 🌟 **2. Steps for Gauss-Jordan Elimination**

1. **Convert to augmented matrix**: Combine $ \mathbf{A} $ and $ \mathbf{b} $ into one matrix. 
1. **Forward elimination**: Use row operations to create zeros below the main diagonal.  
2. **Backward elimination**: Continue row operations to create zeros above the main diagonal.  
3. **Convert leading coefficients to 1** to get **RREF (Reduced Row Echelon Form)**.  
4. **Extract solutions** from the last column.

---

## 🌟 **3. Example Problem**  

Solve the system:  
$$
\begin{cases}
2x + 3y = 8 \\
x - 4y = -5
\end{cases}
$$  

### **Step 1: Convert to Augmented Matrix**
$$
\left[
\begin{array}{cc|c}
2 & 3 & 8 \\
1 & -4 & -5
\end{array}
\right]
$$  

---

### **Step 2: Forward Elimination**
1. Make the first pivot **1** by swapping rows:
$$
\left[
\begin{array}{cc|c}
1 & -4 & -5 \\
2 & 3 & 8
\end{array}
\right]
$$  

2. **Eliminate** the first column in row 2:
   - Subtract **2 × row 1** from row 2:  
   $$ R_2 \leftarrow R_2 - 2R_1 $$
$$
\left[
\begin{array}{cc|c}
1 & -4 & -5 \\
0 & 11 & 18
\end{array}
\right]
$$  

---

### **Step 3: Backward Elimination**
1. Make the pivot in row 2 a **1** by dividing by 11:  
   $$ R_2 \leftarrow R_2 / 11 $$
$$
\left[
\begin{array}{cc|c}
1 & -4 & -5 \\
0 & 1 & \frac{18}{11}
\end{array}
\right]
$$  

2. **Eliminate** the second column in row 1:
   - Add **4 × row 2** to row 1:  
   $$ R_1 \leftarrow R_1 + 4R_2 $$
$$
\left[
\begin{array}{cc|c}
1 & 0 & \frac{17}{11} \\
0 & 1 & \frac{18}{11}
\end{array}
\right]
$$  

---

### **Step 4: Extract Solution**
Since the matrix is now in **Reduced Row Echelon Form (RREF)**, we directly read:
$$
x = \frac{17}{11}, \quad y = \frac{18}{11}
$$  

---

## 🌟 **4. Python Implementation**
```python
import numpy as np

# Define augmented matrix
A = np.array([[2, 3, 8], [1, -4, -5]], dtype=float)

# Perform Gauss-Jordan elimination
rows, cols = A.shape

for i in range(rows):
    # Make the pivot element 1
    A[i] = A[i] / A[i, i]
    
    for j in range(rows):
        if i != j:
            A[j] = A[j] - A[j, i] * A[i]

# Extract solutions
solution = A[:, -1]

print("Solution (x, y):", solution)
```

✔️ **Output:**  
```
Solution (x, y): [1.54545455 1.63636364]
```
(Equivalent to \( x = 17/11, y = 18/11 \))

## 🌟 **5. Applications in Data Science**  

### **1. Solving Linear Regression Equations**
The **normal equation** in linear regression:

$$ 
\mathbf{\beta} = (\mathbf{X}^\top \mathbf{X})^{-1} \mathbf{X}^\top \mathbf{y} 
$$  

Gauss-Jordan is an alternative to computing $ \mathbf{X}^{-1} $.

```python
# Example: Solve Linear Regression Coefficients
X = np.array([[1, 1], [1, 2], [1, 3]])
y = np.array([2, 2.5, 3.5])

# Compute coefficients using np.linalg.solve
beta = np.linalg.solve(X.T @ X, X.T @ y)
print("Regression Coefficients:", beta)
```

---

### **2. Matrix Inversion**

- If $ \mathbf{A} $ is a **square matrix**, Gauss-Jordan can compute $ \mathbf{A}^{-1} $.

```python
A = np.array([[2, 3], [1, -4]], dtype=float)
A_inv = np.linalg.inv(A)  # Using NumPy for efficiency
print("Inverse of A:\n", A_inv)
```

## 🌟 **6. Advantages & Disadvantages of Gauss-Jordan Elimination**

| **Feature** | **Pros** | **Cons** |
|------------|---------|---------|
| **Direct method** | Provides exact solutions | Computationally expensive for large systems |
| **Simplifies Inversion** | Computes $ \mathbf{A}^{-1} $ | Sensitive to rounding errors |
| **Good for small to medium problems** | Avoids back-substitution | Not ideal for high-dimensional systems |

---

## 🌟 **7. Key Takeaways**
1. **Gauss-Jordan Elimination** transforms the coefficient matrix into **RREF**, allowing direct solution extraction.  
2. **Efficient for small to medium-sized problems** but **not scalable** for high-dimensional datasets.  
3. **Used in matrix inversion, linear regression, and optimization**.  
4. **Python implementation using NumPy** makes solving linear systems easy.
