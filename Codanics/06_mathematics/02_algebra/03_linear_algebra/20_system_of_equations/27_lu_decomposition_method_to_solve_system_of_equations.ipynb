# 🌟 **LU Decomposition Method to Solve a System of Equations in Data Science** 🔢📊  

### **Introduction**  
**LU Decomposition** (or **LU Factorization**) is a method for solving systems of linear equations by **decomposing a matrix** into two triangular matrices:  
- **L (Lower triangular matrix)**  
- **U (Upper triangular matrix)**  

It is **faster than Gaussian elimination** for **multiple right-hand sides** and is widely used in **numerical computing, optimization, and machine learning**.  

✔️ **Best for**: Medium to large-sized systems.  
⚠️ **Not ideal for**: Very large, sparse matrices (use iterative methods).  

---

## 🌟 **1. Understanding LU Decomposition**  

A system of equations in **matrix form**:  
$$ 
\mathbf{A} \cdot \mathbf{x} = \mathbf{b} 
$$  

### **LU Factorization**  
LU decomposition **factorizes** $ \mathbf{A} $ into:  
$$ 
\mathbf{A} = \mathbf{L} \cdot \mathbf{U} 
$$  
where:  
- $ \mathbf{L} $ is a **lower triangular matrix** (diagonal elements = 1).  
- $ \mathbf{U} $ is an **upper triangular matrix**.  

Once $ \mathbf{A} $ is decomposed, we solve in **two steps**:  
1. **Solve $ \mathbf{L} \cdot \mathbf{y} = \mathbf{b} $ (Forward Substitution)**  
2. **Solve $ \mathbf{U} \cdot \mathbf{x} = \mathbf{y} $ (Backward Substitution)**  

LU decomposition is **more efficient** than direct matrix inversion for **large systems**.

---

## 🌟 **2. Example Problem**  

Solve the system:  
$$
\begin{cases}
2x + 3y = 8 \\
x - 4y = -5
\end{cases}
$$  

### **Step 1: Convert to Matrix Form**  
$$ 
\begin{bmatrix} 
2 & 3 \\ 
1 & -4 
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y 
\end{bmatrix} 
=
\begin{bmatrix} 
8 \\ 
-5 
\end{bmatrix} 
$$  
where:  
- $ \mathbf{A} = \begin{bmatrix} 2 & 3 \\ 1 & -4 \end{bmatrix} $  
- $ \mathbf{b} = \begin{bmatrix} 8 \\ -5 \end{bmatrix} $

---

### **Step 2: Compute LU Decomposition**  

We want to factorize $ \mathbf{A} $ into:  
$$ 
\mathbf{A} = \mathbf{L} \cdot \mathbf{U} 
$$  
where:  
- $ \mathbf{L} = \begin{bmatrix} 1 & 0 \\ l_{21} & 1 \end{bmatrix} $  
- $ \mathbf{U} = \begin{bmatrix} u_{11} & u_{12} \\ 0 & u_{22} \end{bmatrix} $  

Using **Gaussian elimination**, we find:  
$$ 
\mathbf{L} = \begin{bmatrix} 1 & 0 \\ 0.5 & 1 \end{bmatrix}, \quad  
\mathbf{U} = \begin{bmatrix} 2 & 3 \\ 0 & -5.5 \end{bmatrix} 
$$

---

### **Step 3: Solve $ \mathbf{L} \cdot \mathbf{y} = \mathbf{b} $ (Forward Substitution)**  
$$ 
\begin{bmatrix} 
1 & 0 \\ 
0.5 & 1 
\end{bmatrix} 
\begin{bmatrix} 
y_1 \\ 
y_2 
\end{bmatrix} 
=
\begin{bmatrix} 
8 \\ 
-5 
\end{bmatrix} 
$$  

Solving:  
- $ y_1 = 8 $  
- $ y_2 = -5 - 0.5(8) = -9 $  

So,  
$$ y = \begin{bmatrix} 8 \\ -9 \end{bmatrix} $$  

---

### **Step 4: Solve $ \mathbf{U} \cdot \mathbf{x} = \mathbf{y} $ (Backward Substitution)**  
$$ 
\begin{bmatrix} 
2 & 3 \\ 
0 & -5.5 
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y 
\end{bmatrix} 
=
\begin{bmatrix} 
8 \\ 
-9 
\end{bmatrix} 
$$  

Solving:  
- $ y = -9 / -5.5 = \frac{18}{11} $  
- $ x = \frac{8 - 3(18/11)}{2} = \frac{17}{11} $  

### **Solution:**  
$$ x = \frac{17}{11}, \quad y = \frac{18}{11} $$

---

## 🌟 **3. Python Implementation**  

### **Using SciPy's LU Decomposition**
```python
import numpy as np
from scipy.linalg import lu, lu_solve, lu_factor

# Define coefficient matrix A and result vector b
A = np.array([[2, 3], [1, -4]], dtype=float)
b = np.array([8, -5], dtype=float)

# Perform LU decomposition
LU, piv = lu_factor(A)

# Solve for x using forward & backward substitution
x = lu_solve((LU, piv), b)

print("Solution (x, y):", x)
```

✔️ **Output:**  
```
Solution (x, y): [1.54545455 1.63636364]
```

(Equivalent to $ x = \frac{17}{11}, y = \frac{18}{11} $)

## 🌟 **4. Applications in Data Science**  

### **1. Solving Linear Regression Equations**
For **linear regression**, we solve:  
$$ 
\mathbf{\beta} = (\mathbf{X}^\top \mathbf{X})^{-1} \mathbf{X}^\top \mathbf{y} 
$$  
LU decomposition can efficiently solve **$ (\mathbf{X}^\top \mathbf{X}) \cdot \mathbf{\beta} = \mathbf{X}^\top \mathbf{y} $**.

---

### **2. Matrix Inversion**
If $ \mathbf{A} $ is a **square matrix**, LU decomposition helps compute $ \mathbf{A}^{-1} $.

```python
A_inv = np.linalg.inv(A)  # Using NumPy
print("Inverse of A:\n", A_inv)
```

## 🌟 **5. Advantages & Disadvantages of LU Decomposition**

| **Feature** | **Pros** | **Cons** |
|------------|---------|---------|
| **Efficient for large systems** | Faster than direct Gaussian elimination | Requires **pivoting** for numerical stability |
| **Used in regression & optimization** | Handles multiple right-hand sides efficiently | Not ideal for sparse matrices (use iterative methods) |
| **Avoids matrix inversion** | More stable than direct inversion | Computationally expensive for very large systems |
