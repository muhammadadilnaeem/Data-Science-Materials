# 🌟 **Gaussian Elimination Method to Solve Systems of Equations in Data Science** 🔢📊  

### **Introduction**  
**Gaussian Elimination** is a systematic method for solving systems of linear equations by transforming them into an **upper triangular matrix**, then using **back-substitution** to find the solutions. It is **efficient for moderate-sized systems** and is the basis for advanced numerical methods like **LU decomposition**.  

✔️ **Best for**: Medium-sized systems ($ n < 1000 $).  
⚠️ **Not ideal for large-scale problems** (use LU decomposition or iterative methods).  

---

## 🌟 **1. Understanding Gaussian Elimination**  

A system of equations can be represented in **matrix form**:  
$$ 
\mathbf{A} \cdot \mathbf{x} = \mathbf{b} 
$$  
where:  
- $ \mathbf{A} $ = **coefficient matrix**  
- $ \mathbf{x} $ = **variable vector**  
- $ \mathbf{b} $ = **result vector**  

The **goal** of Gaussian elimination is to **convert $ \mathbf{A} $ into an upper triangular matrix**, then solve for variables using **back-substitution**.


-----

## 🌟 **2. Steps for Gaussian Elimination**

1. **Convert to augmented matrix**: Combine $ \mathbf{A} $ and $ \mathbf{b} $ into one matrix.  
2. **Forward elimination**: Use row operations to create zeros below the main diagonal.  
3. **Back-substitution**: Solve for variables starting from the last row.

---

## 🌟 **3. Example Problem**  

Solve the system:  
$$
\begin{cases}
2x + 3y = 8 \\
x - 4y = -5
\end{cases}
$$  

### **Step 1: Convert to Augmented Matrix**
$$
\left[
\begin{array}{cc|c}
2 & 3 & 8 \\
1 & -4 & -5
\end{array}
\right]
$$  

---

### **Step 2: Forward Elimination**
**Make the first column’s second row zero**:  
- Multiply row 2 by 2 and subtract row 1:
$$
R_2 \leftarrow R_2 - \frac{1}{2} R_1
$$  

New matrix:
$$
\left[
\begin{array}{cc|c}
2 & 3 & 8 \\
0 & -5.5 & -9
\end{array}
\right]
$$  

---

### **Step 3: Back-Substitution**  
1. Solve for $ y $ from row 2:  
   $$ -5.5y = -9 $$  
   $$ y = \frac{-9}{-5.5} = \frac{18}{11} $$  

2. Substitute $ y $ into row 1:  
   $$ 2x + 3 \left(\frac{18}{11}\right) = 8 $$  
   $$ 2x + \frac{54}{11} = 8 $$  
   $$ 2x = 8 - \frac{54}{11} $$  
   $$ 2x = \frac{88}{11} - \frac{54}{11} = \frac{34}{11} $$  
   $$ x = \frac{34}{22} = \frac{17}{11} $$  

### **Solution:**  
$$ x = \frac{17}{11}, \quad y = \frac{18}{11} $$

---

## 🌟 **4. Python Implementation**
```python
import numpy as np

# Define coefficient matrix A and result vector b
A = np.array([[2, 3], [1, -4]], dtype=float)
b = np.array([8, -5], dtype=float)

# Solve using NumPy's built-in function
solution = np.linalg.solve(A, b)

print("Solution (x, y):", solution)
```

✔️ **Output:**  
```
Solution (x, y): [1.54545455 1.63636364]
```
(Equivalent to \( x = 17/11, y = 18/11 \))

## 🌟 **5. Applications in Data Science**  

### **1. Linear Regression (Solving for Coefficients)**
The **normal equation** in linear regression:  
$$ 
\mathbf{\beta} = (\mathbf{X}^\top \mathbf{X})^{-1} \mathbf{X}^\top \mathbf{y} 
$$  
Gaussian elimination helps solve this efficiently.

```python
# Solve Linear Regression Coefficients
X = np.array([[1, 1], [1, 2], [1, 3]])
y = np.array([2, 2.5, 3.5])

# Compute coefficients using np.linalg.solve
beta = np.linalg.solve(X.T @ X, X.T @ y)
print("Regression Coefficients:", beta)
```

## 🌟 **6. Advantages & Disadvantages of Gaussian Elimination**

| **Feature** | **Pros** | **Cons** |
|-------------|----------|----------|
| **Direct method** | Exact solution | Computationally expensive for very large systems |
| **Handles any square system** | Works for dependent & independent variables | Pivoting required for numerical stability |
| **Good for moderate-size problems** | Works well for $ n < 1000 $ | Not ideal for sparse matrices |
