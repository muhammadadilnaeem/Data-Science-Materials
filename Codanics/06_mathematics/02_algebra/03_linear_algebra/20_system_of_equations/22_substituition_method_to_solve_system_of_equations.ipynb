# **🌟 Substitution Method to Solve a System of Equations** 🔢📊  

### **Introduction**  
The **substitution method** is a step-by-step approach to solving a **system of equations** by expressing one variable in terms of another and substituting it into the second equation.  

✔️ **Best for**: Small systems (2 variables, 2 equations).  
⚠️ **Not practical**: For large systems (use matrix methods instead).  

---

## 🌟 **1. Steps for the Substitution Method**

1. **Solve for one variable in terms of the other** in one equation.  
2. **Substitute** this expression into the second equation.  
3. **Solve for the remaining variable**.  
4. **Back-substitute** to find the other variable.  
5. **Check the solution** by plugging values into the original equations.  

---

## 🌟 **2. Example Problem**

Solve the system:  
$$
\begin{cases}
2x + 3y = 8 \\
x - 4y = -5
\end{cases}
$$  

### **Step 1: Solve for one variable in terms of the other**  
Solve the second equation for $ x $:  
$$ x = 4y - 5 $$
  
### **Step 2: Substitute into the first equation**  
Replace $ x $ in $ 2x + 3y = 8 $:  
$$ 2(4y - 5) + 3y = 8 $$  
$$ 8y - 10 + 3y = 8 $$  
$$ 11y = 18 $$  
$$ y = \frac{18}{11} $$  

### **Step 3: Solve for $ x $**  
Substituting $ y = \frac{18}{11} $ into $ x = 4y - 5 $:  
$$ x = 4 \times \frac{18}{11} - 5 $$  
$$ x = \frac{72}{11} - \frac{55}{11} $$  
$$ x = \frac{17}{11} $$  

### **Solution:**  
$$ x = \frac{17}{11}, \quad y = \frac{18}{11} $$  

---

## 🌟 **3. Python Implementation**
```python
# Define the equations
def solve_by_substitution():
    # Step 1: Solve for x in terms of y
    def equation1(y):
        return 4 * y - 5  # x = 4y - 5

    # Step 2: Solve for y
    y = 18 / 11
    x = equation1(y)

    # Output the solution
    return x, y

# Get the solution
solution = solve_by_substitution()
print("Solution (x, y):", solution)
```
