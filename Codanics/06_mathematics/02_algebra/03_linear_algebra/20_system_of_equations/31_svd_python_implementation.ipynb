## **Solve the system:**  

$$
\begin{cases}
2x + 3y = 8 \\
5x + 4y = 20
\end{cases}
$$  

### **Step 1: Convert to Matrix Form**  
$$ 
\begin{bmatrix} 
2 & 3 \\ 
5 & 4 
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y 
\end{bmatrix} 
=
\begin{bmatrix} 
8 \\ 
20 
\end{bmatrix} 
$$  
where:  
- $ \mathbf{A} = \begin{bmatrix} 2 & 3 \\ 5 & 4 \end{bmatrix} $  
- $ \mathbf{b} = \begin{bmatrix} 8 \\ 20 \end{bmatrix} $


```python
# import libraries
import os
import numpy as np
```

```python
# create a coefficent matrix A
A = np.array([[2,3],[5,4]]) # 2x2 matrix
print(A)
```

```python
# create a coefficent matrix B
B = np.array([8,20]) # 1x1 matrix
print(B)
```

```python
# solve for x
x = np.linalg.solve(A,B) 
x
```
