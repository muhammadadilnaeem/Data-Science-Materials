### 🌟 **Percent and Percentages** 🔢📊  

**Percent** and **percentages** are fundamental concepts in mathematics that describe parts of a whole in terms of 100. They are widely used in data science to express proportions, analyze datasets, and interpret statistical results.  

---

### 🌟 **What is a Percent?**

The word **percent** means "per hundred" and is denoted by the symbol **%**. It is a way of expressing a fraction or ratio where the denominator is 100.  

- **Definition**:  
  $$ 
  \text{Percent} = \frac{\text{Part}}{\text{Whole}} \times 100 
  $$  

- **Example**:  
  If a student scores 80 out of 100 in a test, their score as a percent is:  
  $$ 
  \text{Percent} = \frac{80}{100} \times 100 = 80\% 
  $$  

---

### 🌟 **What is a Percentage?**

A **percentage** represents the portion of a whole expressed as a fraction of 100. It is often used to compare relative sizes or proportions in datasets.  

- **Usage**: Percentages help quantify changes, distributions, and comparisons in data.  

- **Example**:  
  If a company’s sales increased from \$500 to \$750, the percentage increase is:  
  $$ 
  \text{Percentage Increase} = \frac{\text{Change}}{\text{Original Value}} \times 100 
  $$  
  $$ 
  \text{Percentage Increase} = \frac{750 - 500}{500} \times 100 = 50\% 
  $$  

---

### 🌟 **Key Types of Percentage Calculations**

#### **1. Percent of a Number**  
- To find \( p\% \) of a number \( N \):  
  $$ 
  \text{Result} = \frac{p}{100} \times N 
  $$  

- **Example**:  
  What is \( 20\% \) of 150?  
  $$ 
  \text{Result} = \frac{20}{100} \times 150 = 30 
  $$  

---

#### **2. Percentage Change**  
- To calculate the percentage change between two values:  
  $$ 
  \text{Percentage Change} = \frac{\text{New Value} - \text{Old Value}}{\text{Old Value}} \times 100 
  $$  

- **Examples**:  
  - **Increase**: From 50 to 75:  
    $$ 
    \text{Percentage Increase} = \frac{75 - 50}{50} \times 100 = 50\% 
    $$  
  - **Decrease**: From 100 to 80:  
    $$ 
    \text{Percentage Decrease} = \frac{100 - 80}{100} \times 100 = 20\% 
    $$  

---

#### **3. Converting Fractions and Decimals to Percentages**  
- Convert a **fraction** to a percentage:  
  $$ 
  \text{Percentage} = \frac{\text{Numerator}}{\text{Denominator}} \times 100 
  $$  
  Example: \( \frac{3}{5} \):  
  $$ 
  \text{Percentage} = \frac{3}{5} \times 100 = 60\% 
  $$  

- Convert a **decimal** to a percentage:  
  $$ 
  \text{Percentage} = \text{Decimal} \times 100 
  $$  
  Example: \( 0.75 \):  
  $$ 
  \text{Percentage} = 0.75 \times 100 = 75\% 
  $$  

---

#### **4. Reverse Percentage**  
- To find the original value from a percentage:  
  $$ 
  \text{Original Value} = \frac{\text{Part}}{\text{Percentage}} \times 100 
  $$  

- **Example**: If \( 20\% \) of a number is 50, what is the original number?  
  $$ 
  \text{Original Value} = \frac{50}{20} \times 100 = 250 
  $$  

---

### 🌟 **Applications of Percentages in Data Science**

Percentages are widely used in **data science** for analyzing datasets, interpreting trends, and building models.  

#### **1. Descriptive Statistics**  
Percentages summarize distributions and proportions in data.  
- **Example**:  
  In a survey of 1,000 people, if 600 prefer product A:  
  $$ 
  \text{Percentage of People Preferring A} = \frac{600}{1000} \times 100 = 60\% 
  $$  

---

#### **2. Probability and Classification Metrics**  
Percentages are used in evaluating classification models in machine learning:  
- **Accuracy**:  
  $$ 
  \text{Accuracy} = \frac{\text{Correct Predictions}}{\text{Total Predictions}} \times 100 
  $$  
- **Precision, Recall, and F1-Score**: Also expressed as percentages to evaluate model performance.  

---

#### **3. Normalization and Scaling**  
- Percentages are used in scaling features to relative proportions.  
  - Example: Sales as a percentage of total revenue.  

---

#### **4. Data Visualizations**  
Percentages are displayed in pie charts, bar charts, and stacked charts to show proportions.  

- **Example (Pie Chart)**:  
  - \( 60\% \) for A, \( 30\% \) for B, and \( 10\% \) for C.  

---

#### **5. Financial Analysis**  
Percentages quantify changes in revenue, profit, and other metrics.  
- **Example**: Profit margin:  
  $$ 
  \text{Profit Margin} = \frac{\text{Net Profit}}{\text{Revenue}} \times 100 
  $$  

---

#### **6. A/B Testing**  
Percentages are used to compare groups in experiments, such as click-through rates (CTR):  
- CTR for Group A:  
  $$ 
  \text{CTR} = \frac{\text{Clicks}}{\text{Impressions}} \times 100 
  $$  

---

#### **7. Missing Data Analysis**  
Percentages show the proportion of missing or invalid data.  
- **Example**: If 200 entries are missing in a dataset of 1,000 records:  
  $$ 
  \text{Missing Data Percentage} = \frac{200}{1000} \times 100 = 20\% 
  $$  

---

### 🌟 **Practical Example: Percentages in Python**

```python
# Example: Calculating Percentage
total = 1000
males = 450

# Percentage of males
percent_males = (males / total) * 100
print(f"Percentage of Males: {percent_males:.2f}%")  # Output: 45.00%

# Percentage increase
old_value = 200
new_value = 250
percent_increase = ((new_value - old_value) / old_value) * 100
print(f"Percentage Increase: {percent_increase:.2f}%")  # Output: 25.00%
```
