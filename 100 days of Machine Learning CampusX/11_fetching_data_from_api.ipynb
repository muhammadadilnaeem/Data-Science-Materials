### **Import Libraries**

````python
import os
import requests
import pandas as pd
````

### **Fetch Data From API**

- To fetch only `page 1` data.

````python

# Send a GET request to the API endpoint to fetch the top-rated movies
response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=***********************&language=en-US&page=1")

# Convert the JSON response to a DataFrame and select specific columns
df = pd.DataFrame(response.json()["results"])[["id", "title", "overview", "release_date", "popularity", "vote_average", "vote_count"]]

# Display the first few rows of the DataFrame
df.head()
````

- To fetch `500` pages data.

````python
# Create an empty list to collect DataFrames
df_list = []

# Loop through pages 1 to 500 to fetch top-rated movies
for page_no in range(1, 501):
    # Send a GET request to the API endpoint for each page
    response = requests.get(f"https://api.themoviedb.org/3/movie/top_rated?api_key=***********************&language=en-US&page={page_no}")
    
    # Check if the request was successful
    if response.status_code == 200:
        # Convert the JSON response to a DataFrame and select specific columns
        temporary_df_movies = pd.DataFrame(response.json().get("results", []))[["id", "title", "overview", "release_date", "popularity", "vote_average", "vote_count"]]
        
        # Append the temporary DataFrame to the list
        df_list.append(temporary_df_movies)
    else:
        print(f"Failed to fetch data for page {page_no}, Status Code: {response.status_code}")

# Concatenate all DataFrames into a single DataFrame
the_df_movies = pd.concat(df_list, ignore_index=True)

# Display the final DataFrame
the_df_movies.head()
````
