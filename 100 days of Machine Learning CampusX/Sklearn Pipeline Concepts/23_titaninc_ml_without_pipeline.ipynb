# ***`Predict Without Pipeline`***

---

- The main goal of this notebook is to implement the `scikit-learn pipeline` without worring about the accuracy to build better concepts.

### **Import Libraries**

```python
import os
import pickle
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    OneHotEncoder,
    LabelEncoder,
)

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

import warnings
warnings.filterwarnings("ignore")
```

### **Load the Dataset**

```python
df = pd.read_csv(r"https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day29-sklearn-pipelines/train.csv")

df.sample(10)
```

### **Drop Unnecessory Columns**

```python
df.drop(columns=['PassengerId','Name','Ticket','Cabin'],inplace=True)
```

```python
df.head()
```

### **Train test Split the Data**

```python
X_train, X_test, y_train, y_test = train_test_split(
    df.drop(columns=['Survived']),
    df['Survived'],
    test_size=0.2,
    random_state=42,
)
```

```python
X_train.head()
```

```python
y_train.head()
```

### **Imputing Missing Values**

```python
# Importing SimpleImputer from scikit-learn for handling missing values in the dataset.

# Create an instance of SimpleImputer for imputing missing values in the 'Age' column.
si_age = SimpleImputer()

# Create an instance of SimpleImputer for imputing missing values in the 'Embarked' column.
# This imputer uses the most frequent value in the column for replacement.
si_embarked = SimpleImputer(strategy='most_frequent')

# Fit the SimpleImputer on the 'Age' column of the training set and transform it.
# This replaces missing values in 'Age' with the mean (default strategy).
X_train_age = si_age.fit_transform(X_train[['Age']])

# Fit the SimpleImputer on the 'Embarked' column of the training set and transform it.
# This replaces missing values in 'Embarked' with the most frequent value.
X_train_embarked = si_embarked.fit_transform(X_train[['Embarked']])

# Transform the 'Age' column of the test set using the fitted imputer from the training set.
# This ensures consistent imputation across training and testing data.
X_test_age = si_age.transform(X_test[['Age']])

# Transform the 'Embarked' column of the test set using the fitted imputer from the training set.
# It replaces missing values with the most frequent value established in the training data.
X_test_embarked = si_embarked.transform(X_test[['Embarked']])
```

```python
# Importing OneHotEncoder from scikit-learn for converting categorical variables into a one-hot encoded format.

# Create an instance of OneHotEncoder for 'Sex' column, allowing for unknown categories to be ignored.
# Use 'sparse_output' instead of 'sparse' to avoid the error.
ohe_sex = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

# Create another instance of OneHotEncoder for 'Embarked' column with the same settings.
ohe_embarked = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

# Fit the OneHotEncoder on the 'Sex' column of the training set and transform it.
X_train_sex = ohe_sex.fit_transform(X_train[['Sex']])

# Fit the OneHotEncoder on the 'Embarked' column of the training set (note: corrected variable name).
X_train_embarked = ohe_embarked.fit_transform(X_train_embarked)

# Transform the 'Sex' column of the test set using the fitted encoder from the training set.
X_test_sex = ohe_sex.transform(X_test[['Sex']])

# Transform the 'Embarked' column of the test set using the fitted encoder from the training set.
X_test_embarked = ohe_embarked.transform(X_test_embarked)
```

```python
X_train_sex[:6]
```

```python
X_train_embarked[:6]
```

```python
# Drop the specified columns ('Sex', 'Age', 'Embarked') from the training set.
# This creates a new DataFrame, 'X_train_remaining', containing only the remaining features.
X_train_remaining = X_train.drop(columns=['Sex', 'Age', 'Embarked'])

# Drop the specified columns ('Sex', 'Age', 'Embarked') from the test set.
# This creates a new DataFrame, 'X_test_remaining', containing only the remaining features.
X_test_remaining = X_test.drop(columns=['Sex', 'Age', 'Embarked'])
```

```python
# Concatenate the remaining features from the training set with the transformed 'Age', 'Sex', and 'Embarked' features.
# The result is a single NumPy array, 'X_train_transformed', containing all relevant features for the model.
X_train_transformed = np.concatenate((X_train_remaining, X_train_age, X_train_sex, X_train_embarked), axis=1)

# Concatenate the remaining features from the test set with the transformed 'Age', 'Sex', and 'Embarked' features.
# The result is a single NumPy array, 'X_test_transformed', containing all relevant features for the model.
X_test_transformed = np.concatenate((X_test_remaining, X_test_age, X_test_sex, X_test_embarked), axis=1)
```

### **Applying Decision Tree Classifier**

```python
# Importing the DecisionTreeClassifier from scikit-learn to create a decision tree model.
clf = DecisionTreeClassifier()

# Fit the decision tree classifier on the transformed training data 'X_train_transformed' and the target variable 'y_train'.
clf.fit(X_train_transformed, y_train)
```

```python
# Use the trained decision tree classifier to make predictions on the transformed test data 'X_test_transformed'.
y_pred = clf.predict(X_test_transformed)

# Output the predicted values for the test set.
y_pred  # This will display the predicted class labels for the test data.
```

```python
# Calculate the accuracy of the predictions by comparing the true labels in 'X_test' with the predicted labels 'y_pred'.
# The accuracy score represents the proportion of correct predictions made by the classifier.
accuracy = accuracy_score(y_test, y_pred)

# Output the accuracy score.
accuracy  # This will display the accuracy of the model on the test set.
```

### **Exporting Models**

```python
# Importing the pickle module for serializing and saving Python objects to disk.

# Save the fitted OneHotEncoder for the 'Sex' column to a file named 'ohe_sex.pkl' in the 'models' directory.
# The 'wb' mode indicates that the file is opened for writing in binary format.
pickle.dump(ohe_sex, open("models/ohe_sex.pkl", "wb"))

# Save the fitted OneHotEncoder for the 'Embarked' column to a file named 'ohe_embarked.pkl' in the 'models' directory.
# This allows for later use without needing to refit the encoder.
pickle.dump(ohe_embarked, open("models/ohe_embarked.pkl", "wb"))

# Save the trained decision tree classifier to a file named 'clf.pkl' in the 'models' directory.
# This enables loading the trained model later for making predictions without retraining.
pickle.dump(clf, open("models/clf.pkl", "wb"))
```

### **Loading Models**

```python
ohe_sex = pickle.load(open("models/ohe_sex.pkl", "rb"))
ohe_embarked = pickle.load(open("models/ohe_embarked.pkl", "rb"))
clf = pickle.load(open("models/clf.pkl", "rb"))
```
