## ✅ **What is Softmax Regression?**  
📌 **Softmax Regression** (also known as **Multinomial Logistic Regression**) is an extension of **Logistic Regression** that handles **multi-class classification problems** where the target variable has **3 or more classes**.  

✔️ Used for **multi-class classification** tasks.  
✔️ Outputs **probabilities** for each class.  
✔️ Uses the **Softmax function** to convert raw scores (logits) into probabilities that sum to **1**.  

---

## 📊 **1. How Does Softmax Regression Work?**  

### 🔹 **Step 1: Compute Linear Model (Logits)**
Softmax Regression calculates a **linear combination** of input features for each class:

$
z_k = \theta_{k0} + \theta_{k1}X_1 + \theta_{k2}X_2 + ... + \theta_{kn}X_n
$

Where:  
- $ z_k $ → The **logit (score)** for class $ k $.  
- $ \theta_k $ → Coefficients (weights) for class $ k $.  

---

### 🔹 **Step 2: Apply the Softmax Function**  
The **Softmax Function** converts these logits into probabilities for each class:

$
P(y = k | X) = \frac{e^{z_k}}{\sum_{j=1}^{K} e^{z_j}}
$

Where:  
- $ P(y = k | X) $ → Probability of class $ k $ given $ X $.  
- $ e^{z_k} $ → Exponentiation to ensure non-negative values.  
- $ \sum_{j=1}^{K} e^{z_j} $ → Ensures probabilities sum to **1**.  

---

### 🔹 **Step 3: Prediction**  
- The class with the **highest probability** is the predicted class:

$
\hat{y} = \arg\max_k P(y = k | X)
$

---

### 🔹 **Step 4: Cross-Entropy Loss Function**  
Softmax Regression minimizes the **Cross-Entropy Loss** to improve model performance:  

$
J(\theta) = -\frac{1}{n} \sum_{i=1}^{n} \sum_{k=1}^{K} y_{ik} \log(P(y = k | X_i))
$

Where:  
- $ y_{ik} $ = 1 if sample $ i $ belongs to class $ k $, otherwise 0.  
- $ P(y = k | X_i) $ → Predicted probability for class $ k $.

---

## 🔍 **2. Example Use Cases for Softmax Regression**  

✅ **Image Classification** → Identifying objects as **cat, dog, or bird**.  
✅ **Sentiment Analysis** → Predicting sentiments like **positive, neutral, or negative**.  
✅ **Handwritten Digit Recognition** → Classifying digits (0–9).  
✅ **Disease Diagnosis** → Classifying patient risk as **low, medium, or high**.

## 🏹 **3. Implementing Softmax Regression in Python**  

### 📁 **Dataset: Iris Flower Classification (3 Classes - Setosa, Versicolor, Virginica)**  


```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import warnings
warnings.filterwarnings("ignore")

# Load Dataset
from sklearn.datasets import load_iris
data = load_iris()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target  # Target has 3 classes: 0 (Setosa), 1 (Versicolor), 2 (Virginica)

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train Softmax Regression Model
softmax_reg = LogisticRegression(multi_class='multinomial', solver='lbfgs', max_iter=1000)
softmax_reg.fit(X_train, y_train)

# Predictions
y_pred = softmax_reg.predict(X_test)

# Model Evaluation
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```


📌 **Observations:**  
✔ **High Accuracy (0.98)** → The model performs well.  
✔ **Precision & Recall** scores are high for all classes.  
✔ The model predicts each flower species with excellent precision.

## 📊 **5. Visualizing Softmax Probabilities**  

The **Softmax function** outputs **probability scores** for each class. Visualizing these scores can reveal model confidence.

```python
import matplotlib.pyplot as plt

# Predict Probability Scores
y_prob = softmax_reg.predict_proba(X_test)

# Plot Probabilities for 5 Random Samples
plt.figure(figsize=(8, 5))
for i in range(5):
    plt.bar(data.target_names, y_prob[i], label=f"Sample {i+1}")
    plt.xlabel("Classes")
    plt.ylabel("Probability")
    plt.title(f"Probability Distribution (Sample {i+1})")
    plt.show()
```


📌 **Observation:** Each bar shows the probability distribution across the 3 classes — the highest bar corresponds to the predicted class.

## 📋 **6. Key Differences Between Logistic Regression & Softmax Regression**  

| **Aspect**            | **Logistic Regression** | **Softmax Regression** |
|------------------------|------------------------|------------------------|
| **Type of Task**         | **Binary Classification** (2 classes) | **Multi-Class Classification** (3 or more classes) |
| **Output**                | Probability for **one class** | Probability for **each class** |
| **Decision Boundary**      | Linear | Linear or Complex (depends on data) |
| **Activation Function**     | **Sigmoid** | **Softmax** |


## 🔥 **7. When to Use Softmax Regression**  

✅ **Best for Multi-Class Classification** tasks.  
✅ When classes are **mutually exclusive** (each sample belongs to only one class).  
✅ Useful for problems with **interpretable probability scores**.  

---

## 🏆 **8. Key Takeaways**  

✔️ **Softmax Regression (Multinomial Logistic Regression)** is ideal for **multi-class classification**.  
✔️ Uses the **Softmax function** to output probabilities for each class.  
✔️ Optimized using **Cross-Entropy Loss** and **Gradient Descent**.  
✔️ Performs best when classes are **mutually exclusive**.  
✔️ Suitable for datasets like **Iris, MNIST**, and **Sentiment Analysis**.  

💡 **Pro Tip:** If your dataset has **high-dimensional features**, consider adding **L1 (Lasso)** or **L2 (Ridge)** regularization for better generalization.

## ***`Further Practice`***

```python
# import libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report,confusion_matrix
import warnings
warnings.filterwarnings("ignore")
```

```python
# load the iris dataset from the seaborn
df = sns.load_dataset("iris")
df.head()
```

```python
# encode the categorical column
encoder = LabelEncoder()

df['species'] = encoder.fit_transform(df['species'])

df.sample(6)
```

```python
# select 3 columns for further analysis
df = df[['sepal_length','petal_length','species']]
```

```python
X = df.iloc[:, 0:2]
y = df.iloc[:, -1]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
clf = LogisticRegression(multi_class="multinomial")
```

```python
clf.fit(X_train,y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
pd.DataFrame(confusion_matrix(y_test,y_pred))
```

```python
# prediction
query = np.array([[3.4,2.7]])
clf.predict_proba(query)
```

```python
clf.predict(query)
```

```python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(X.values, y.values, clf, legend=2)

# Adding axes annotations
plt.xlabel('sepal length [cm]')
plt.xlabel('petal length [cm]')
plt.title('Softmax on Iris')

plt.show()
```
