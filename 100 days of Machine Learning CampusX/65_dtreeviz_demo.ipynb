````python
import graphviz.backend as be  
from sklearn.datasets import load_iris 
from sklearn.tree import plot_tree, DecisionTreeClassifier,DecisionTreeRegressor
from sklearn.model_selection import train_test_split
import dtreeviz
from dtreeviz.trees import *  
from IPython.display import Image, display_svg, SVG  

import warnings
warnings.filterwarnings("ignore")
````

````python
# Load the Iris dataset
iris = load_iris()
X_train = iris.data  # Features
y_train = iris.target  # Target labels

# Train model
# clas = DecisionTreeClassifier(max_depth=2)
clas = DecisionTreeClassifier()
clas.fit(X_train, y_train)
````

````python
plot_tree(clas)
````

### **Classification**

````python
# Assuming 'clas' is your trained DecisionTreeClassifier model
viz = dtreeviz.model(
    clas,
    X_train,
    y_train,
    target_name='species',
    feature_names=iris.feature_names,
    class_names=iris.target_names
)

# Display the visualization with increased plot size
viz.view(scale=2.0)  # Adjust scale for larger output
````

````python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
import numpy as np
import dtreeviz

# Load the Iris dataset
iris = load_iris()
X_train = iris.data  # Features
y_train = iris.target  # Target labels

# Train model
clas = DecisionTreeClassifier()
clas.fit(X_train, y_train)

# Randomly select an instance from the dataset
X = iris.data[np.random.randint(0, len(iris.data)), :]

# Assuming 'clas' is your trained DecisionTreeClassifier model
viz = dtreeviz.model(
    clas,
    X_train,
    y_train,
    target_name='species',
    feature_names=iris.feature_names,
    class_names=iris.target_names
)

# Display the visualization with increased plot size
viz.view(scale=2.0, x=X)  # Pass the instance 'X' here using the 'x' parameter
````

````python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
import numpy as np
import dtreeviz

# Load the Iris dataset
iris = load_iris()
X_train = iris.data  # Features
y_train = iris.target  # Target labels

# Train model
clas = DecisionTreeClassifier()
clas.fit(X_train, y_train)

# Randomly select an instance from the dataset
X = iris.data[np.random.randint(0, len(iris.data)), :]

# Assuming 'clas' is your trained DecisionTreeClassifier model
viz = dtreeviz.model(
    clas,
    X_train,
    y_train,
    target_name='species',
    feature_names=iris.feature_names,
    class_names=iris.target_names
)

# Display the visualization with increased plot size
viz.view(x=X,show_node_labels=True,orientation="LR",
         show_just_path=True)  # Pass the instance 'X' here using the 'x' parameter
````
