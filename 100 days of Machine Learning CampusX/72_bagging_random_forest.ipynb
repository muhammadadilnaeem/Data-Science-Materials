# ***`Bagging Vs Random Forest`***

Here's a tabular comparison of Bagging and Random Forest:

| Feature                | Bagging                               | Random Forest                          |
|------------------------|---------------------------------------|---------------------------------------|
| **Definition**         | An ensemble method that combines multiple models to improve accuracy. | A specific type of bagging that uses decision trees as base learners. |
| **Algorithm Type**     | General ensemble method.              | Ensemble of decision trees.           |
| **Base Learners**      | Can use any model (e.g., trees, SVM).| Primarily decision trees.             |
| **Data Sampling**      | Bootstrapping (sampling with replacement). | Bootstrapping + random feature selection. |
| **Feature Selection**   | All features are considered for each model. | Random subset of features used for each split. |
| **Variance Reduction**  | Reduces variance by averaging predictions. | Further reduces variance through randomness in feature selection. |
| **Overfitting**        | Less prone to overfitting than single models, but can still overfit. | Generally less prone to overfitting due to randomness. |
| **Interpretability**   | More interpretable if simple models are used. | Less interpretable due to complexity of multiple trees. |
| **Performance**        | Good performance, especially with high variance models. | Often better performance than bagging due to additional randomness. |
| **Use Cases**          | Suitable for various algorithms and datasets. | Preferred for classification and regression tasks with complex data. |

This comparison highlights the key differences and similarities between Bagging and Random Forest.

## **Further Practice**

#### **Bagging**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.ensemble import BaggingClassifier,RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt


import warnings
warnings.filterwarnings('ignore')
```

```python
X,y = make_classification(n_features=5, n_redundant=0, n_informative=5,n_clusters_per_class=1)
```

```python
df = pd.DataFrame(X,columns=['col1','col2','col3','col4','col5'])
df['target'] = y
print(df.shape)
df.head()
```

```python
bag = BaggingClassifier(max_features=2)
```

```python
bag.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
plt.figure(figsize=(12,12))
plot_tree(bag.estimators_[0])
plt.show()
```

- This proved that column splitting occurs before the tree formation in bagging.

##### **RandomForest**

```python
rf = RandomForestClassifier(max_features=2)
```

```python
rf.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
plt.figure(figsize=(12,12))
plot_tree(rf.estimators_[4])
plt.show()
```

- This shows node level spliting , that's why it has more reandomness and differenrece betweeen base models.
