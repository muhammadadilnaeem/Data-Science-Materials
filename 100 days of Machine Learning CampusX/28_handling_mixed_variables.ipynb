```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day33-handling-mixed-variables/titanic.csv")
```

```python
df.head()
```

```python
df['number'].unique()
```

```python
fig = df['number'].value_counts().plot.bar()
fig.set_title("Passangers Traveling With")
plt.show()
```

```python
# Convert the 'number' column in the DataFrame 'df' to a numeric type
# This will extract the numerical part of the 'number' column
# - 'errors='coerce'' will replace any non-convertible values with NaN
# - 'downcast='integer'' attempts to downcast the resulting numeric values to the smallest integer type
df['number_numerical'] = pd.to_numeric(df['number'], errors='coerce', downcast='integer')
```

```python
# Extract the categorical part of the 'number' column in the DataFrame 'df'
# Create a new column 'number_categorical' based on the condition
# - If 'number_numerical' is null (NaN), take the original 'number' value
# - If 'number_numerical' is not null, assign NaN to 'number_categorical'
df['number_categorical'] = np.where(df['number_numerical'].isnull(), df['number'], np.nan)
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['Ticket'].unique()
```

```python
# Extract the numeric part from the 'Cabin' column in the DataFrame 'df'
# The regex pattern "(\d+)" captures one or more digits
# The result is stored in a new column 'cabin_num'
df['cabin_num'] = df['Cabin'].str.extract("(\d+)")

# Extract the first character from the 'Cabin' column to get the categorical part
# This is stored in a new column 'cabin_cat'
df['cabin_cat'] = df['Cabin'].str[0]

df.head()
```

```python
df['cabin_cat'].value_counts().plot(kind='bar')
```

```python
# extract the last bit of ticket as number
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],
                                   errors='coerce',
                                   downcast='integer')

# extract the first part of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,
                              df['ticket_cat'])

df.head(20)
```
