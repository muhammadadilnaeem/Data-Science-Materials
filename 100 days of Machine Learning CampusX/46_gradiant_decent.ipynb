# ***`Gradiant Decent Basic Implementation`***

---

#### **Import Libraries**

```python
import os 
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

#### **Make a Sample Dataset**

```python
# Generate a synthetic regression dataset with 4 samples and 1 informative feature, 
# including noise for variability in the target values.
X, y = make_regression(
    n_samples=4,        # Number of samples to generate
    n_features=1,       # Number of features (independent variables)
    n_informative=1,    # Number of informative features contributing to the output
    n_targets=1,        # Number of target variables
    noise=80,           # Standard deviation of the noise added to the output
    random_state=13,    # Seed for reproducibility of the random output
)
```

#### **Scatter Plot of Data**

```python
plt.scatter(X,y)
plt.show()
```

#### **Use Linear Regression on Data**

```python
lr_reg = LinearRegression()
lr_reg.fit(X,y)
```

```python
# check coefficent /  slope
lr_reg.coef_
```

```python
# check intercept
lr_reg.intercept_
```

```python
# plot the result of the model
plt.scatter(X,y)
plt.plot(X,lr_reg.predict(X),color="red")
plt.show()
```

#### **Apply the Gradiant Descent**

```python
# Lets apply Gradient Descent assuming slope is constant m = 78.35
# and let's assume the starting value for intercept b = 100
y_pred = ((78.35 * X) + 100).reshape(4)
```

```python
# Plot the result of the model
plt.scatter(X, y, color='blue', label='Data Points')  # Scatter plot for actual data
plt.plot(X, lr_reg.predict(X), color='red', label='OLS Prediction')  # OLS regression line
plt.plot(X, y_pred, color='green', label='b = 0 Line')  # Reference line at y = 0

# Adding titles and labels
plt.title('Linear Regression Model Results')  # Title of the plot
plt.xlabel('Feature (X)')  # X-axis label
plt.ylabel('Target (y)')   # Y-axis label

# Adding grid for better readability
plt.grid(True)

# Displaying the legend
plt.legend()

# Show the plot
plt.show()
```

```python
# Calculate the slope of the loss function (gradient) for a linear regression model
# using the current values of the slope (m) and intercept (b).
m = 78.35  # Current slope of the regression line
b = 100    # Current intercept of the regression line

loss_slope = -2 * np.sum(y - m * X.ravel() - b)  # Compute the gradient of the loss
loss_slope  # Output the calculated slope
```

```python
# Lets take learning rate = 0.1
lr = 0.1

step_size = loss_slope * lr
step_size
```

```python
# Calculating the new intercept
b = b - step_size
b
```

```python
# Assuming X, y, reg, and y_pred are already defined

y_pred1 = ((78.35 * X) + b).reshape(4)

# Create the scatter plot and add regression lines
plt.scatter(X, y, color='blue', label='Data Points')  # Actual data points
plt.plot(X, lr_reg.predict(X), color='red', label='OLS Prediction')  # OLS regression line
plt.plot(X, y_pred1, color='#00a65a', label='b = {}'.format(b))  # Line for specific intercept b
plt.plot(X, y_pred, color='#A3E4D7', label='b = 0')  # Reference line at b = 0

# Adding titles and labels
plt.title('Linear Regression Model with Different Intercepts')  # Title of the plot
plt.xlabel('Feature (X)')  # X-axis label
plt.ylabel('Target (y)')   # Y-axis label

# Adding grid for better readability
plt.grid(True)

# Displaying the legend
plt.legend()

# Show the plot
plt.show()
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
# Assuming X, y, reg, y_pred, and b are already defined

y_pred2 = ((78.35 * X) + b).reshape(4)

# Create the scatter plot and add regression lines
plt.scatter(X, y, color='blue', label='Data Points', s=50, alpha=0.7)  # Actual data points with size and transparency
plt.plot(X, lr_reg.predict(X), color='red', label='OLS Prediction', linewidth=2)  # OLS regression line
plt.plot(X, y_pred2, color='#00a65a', label='b = {}'.format(b), linestyle='--', linewidth=2)  # Line for specific intercept b
plt.plot(X, y_pred1, color='#A3E4D7', label='b = {}'.format(b), linestyle=':', linewidth=2)  # Another line for specific intercept b
plt.plot(X, y_pred, color='#A3E4D7', label='b = 0', linestyle='-', linewidth=2)  # Reference line at b = 0

# Adding titles and labels
plt.title('Linear Regression Model with Different Intercepts')  # Title of the plot
plt.xlabel('Feature (X)')  # X-axis label
plt.ylabel('Target (y)')   # Y-axis label

# Adding grid for better readability
plt.grid(True)

# Displaying the legend
plt.legend()

# Show the plot
plt.show()
```

```python
# Assuming X, y, reg, y_pred, y_pred1, y_pred2, and b are already defined

y_pred3 = ((78.35 * X) + b).reshape(4)

# Create a larger figure for better visibility
plt.figure(figsize=(15, 15))

# Scatter plot of the actual data points
plt.scatter(X, y, color='blue', label='Data Points', s=100, alpha=0.6)  # Increased size and transparency

# OLS regression line
plt.plot(X, lr_reg.predict(X), color='red', label='OLS Prediction', linewidth=2)

# Lines for different intercepts
plt.plot(X, y_pred3, color='#00a65a', label='b = {}'.format(b), linestyle='--', linewidth=2)
plt.plot(X, y_pred2, color='#A3E4D7', label='b = {}'.format(b), linestyle=':', linewidth=2)
plt.plot(X, y_pred1, color='#A3E4D7', label='b = {}'.format(b), linestyle='-', linewidth=2)
plt.plot(X, y_pred, color='#A3E4D7', label='b = 0', linestyle='-', linewidth=2)

# Adding titles and labels
plt.title('Comparison of Linear Regression Predictions with Different Intercepts', fontsize=24)  # Title of the plot
plt.xlabel('Feature (X)', fontsize=18)  # X-axis label
plt.ylabel('Target (y)', fontsize=18)   # Y-axis label

# Adding grid for better readability
plt.grid(True)

# Displaying the legend with increased font size
plt.legend(fontsize=14)

# Show the plot
plt.show()
```

```python
b = -100
m = 78.35
lr = 0.01

epochs = 100

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*X.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * X + b

  plt.plot(X,y_pred)

plt.scatter(X,y)
```

```python
# Initial parameters for the linear regression model
b = -100      # Initial intercept
m = 78.35     # Initial slope
lr = 0.01     # Learning rate
epochs = 100  # Number of iterations for gradient descent

# Create a figure for plotting
plt.figure(figsize=(10, 6))

# Perform gradient descent to update the intercept (b)
for i in range(epochs):
    # Calculate the slope of the loss function (gradient)
    loss_slope = -2 * np.sum(y - m * X.ravel() - b)
    
    # Update the intercept (b) using the learning rate and loss slope
    b = b - (lr * loss_slope)

    # Calculate the predicted values based on the current slope (m) and intercept (b)
    y_pred = m * X + b

    # Plot the current prediction line
    plt.plot(X, y_pred, color='orange', alpha=0.1)  # Use a light color for intermediate lines

# Scatter plot of the actual data points
plt.scatter(X, y, color='blue', label='Data Points', s=100, alpha=0.6)  # Increased size and transparency

# Final prediction line after all epochs
plt.plot(X, y_pred, color='red', label='Final Prediction', linewidth=2)  # Highlight final prediction line

# Adding titles and labels
plt.title('Gradient Descent for Linear Regression', fontsize=20)  # Title of the plot
plt.xlabel('Feature (X)', fontsize=16)  # X-axis label
plt.ylabel('Target (y)', fontsize=16)   # Y-axis label

# Adding grid for better readability
plt.grid(True)

# Displaying the legend
plt.legend(fontsize=14)

# Show the plot
plt.show()
```
