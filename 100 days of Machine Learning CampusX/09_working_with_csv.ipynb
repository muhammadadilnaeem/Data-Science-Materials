# ***`Working With CSV Files`***

### **Import pandas** 


````python
import pandas as pd
````

### **Load `local csv` data**

````python
df = pd.read_csv("./data/aug_train.csv")
df.head()
````

### **`index_col` paramerter**

- Sometimes you need to set some column from your data as index.

````python
df = pd.read_csv("./data/aug_train.csv",index_col="enrollee_id")
df.head()
````

### **`header` paramerter**

- Sometimes you see that your data column names become first row of data and your columns names are like `Unnamed: 0`,`Unnamed: 1` etc.

- If we want our first row of data to be header then we need to use `header=1` parameter.

```python
df = pd.read_csv("test.csv",header=1)
df.head()
```

### **`usecols` paramerter**

- Sometimes you need only few columns for analysis.
- You can specify them while importing data and rest of the columns will be removed. 

````python
df = pd.read_csv("./data/aug_train.csv",usecols=["enrollee_id","gender","education_level"])
df.head()
````

### **`skiprows/nrows` paramerter**

- Sometimes you need remove few rows for analysis.
- You can specify them while importing data and that rows will be removed from rest of the data.

````python
# df = pd.read_csv("./data/aug_train.csv",skiprows=[0,1])
df = pd.read_csv("./data/aug_train.csv",nrows=10) # specify number of rows
df.head()
````

### **`encoding` paramerter**


- Sometimes you need to declare encoding of your data. 
- You need to find the encoding.

````python
df = pd.read_csv("./data/zomato.csv",encoding="latin-1") 
df.head()
````

### **`on_bad_lines='skip'` skip bad lines**

- Sometimes your data contain somes ***rows*** that are disordered, ike your data has 10 rows but some rows has 12 values in this case use `on_bad_lines='skip'` 

````python
df = pd.read_csv("./data/BX-Books.csv", sep=";", encoding="latin-1", on_bad_lines='skip')
df.head()
````

### **`parse_dates=['column_name']`Handling Dates**
- You can change datatype of datetime columns on importing the dataset using `parse_dates=['column_name']`.

````python
df = pd.read_csv(r"E:\Data Science Materials\Codanics\practice_dataset\payments.csv",parse_dates=['paymentDate'])
df.info()
````

### **`na_values=['column_name']` parameter**

- We want to declare different types of `Null` values using 
-  `na_values=['column_name']`.

````python
df = pd.read_csv("./data/aug_train.csv",nrows=10,na_values=['Male'])
df.head()
````

### **Load `csv` data from `URL`**

````python
# Importing required libraries
import requests  # To make HTTP requests
from io import StringIO  # To handle in-memory text streams
import pandas as pd  # Importing pandas for data manipulation

# URL of the raw CSV file
url = "https://raw.githubusercontent.com/cs109/2014_data/refs/heads/master/countries.csv"

# Custom headers to mimic a browser request (some servers block default requests)
headers = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"
}

# Sending a GET request to the URL with custom headers
req = requests.get(url, headers=headers)

# Converting the response text into a file-like object
data = StringIO(req.text)

# Reading the CSV data using pandas
df = pd.read_csv(data)  # Reading the CSV into a DataFrame

# Displaying the DataFrame
df.head()  # Shows the first 5 rows of the DataFrame
````
