# 🔎 ***`Precision, Recall, and F1 Score in Classification Metrics`***  

### ✅ **Why Are These Metrics Important?**  
📌 While **accuracy** is useful, it can be **misleading** for imbalanced datasets. Metrics like **Precision**, **Recall**, and **F1 Score** provide deeper insights into model performance, especially in cases where False Positives (FP) or False Negatives (FN) are critical.  

---

## 📊 **1. Understanding Confusion Matrix**  

A **confusion matrix** helps visualize model predictions by comparing **actual** and **predicted** values.

|                 | **Predicted: Positive** | **Predicted: Negative** |
|-----------------|------------------------|------------------------|
| **Actual: Positive** | **True Positive (TP)** | **False Negative (FN)** |
| **Actual: Negative** | **False Positive (FP)** | **True Negative (TN)** |

---

## 📌 **2. Precision (Positive Predictive Value - PPV)**  

### 🔹 **What is Precision?**  
**Precision** measures how many of the predicted **positive cases** are actually **positive**.  

### 📋 **Formula:**  
$
Precision = \frac{TP}{TP + FP}
$

### 🔍 **When to Focus on Precision?**  
✅ When **`False Positives (FP)`** are costly.  
✅ Examples:  
- **Spam detection** → Minimizing marking legitimate emails as spam.  
- **Fraud detection** → Avoid incorrectly flagging valid transactions.  

### 📌 **Python Code for Precision:**


```python
from sklearn.metrics import precision_score
precision = precision_score(y_test, y_pred)
print("Precision:", precision)
```


---

## 📌 **3. Recall (Sensitivity or True Positive Rate - TPR)**  

### 🔹 **What is Recall?**  
**Recall** measures how many **actual positive cases** were correctly predicted.  

### 📋 **Formula:**  
$
Recall = \frac{TP}{TP + FN}
$

### 🔍 **When to Focus on Recall?**  
✅ When **`False Negatives (FN)`** are costly.  
✅ Examples:  
- **Cancer diagnosis** → Missing a sick patient can be dangerous.  
- **Fraud detection** → Better to mistakenly flag a transaction than miss a fraudulent one.  

### 📌 **Python Code for Recall:**

```python
from sklearn.metrics import recall_score
recall = recall_score(y_test, y_pred)
print("Recall:", recall)
```


---

## 📌 **4. F1 Score (Harmonic Mean of Precision & Recall)**  

### 🔹 **What is the F1 Score?**  
**F1 Score** combines **Precision** and **Recall** into one metric to balance both.  

### 📋 **Formula:**  
$
F1 = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}
$

### 🔍 **When to Focus on F1 Score?**  
✅ When you need a **balance** between Precision and Recall.  
✅ Best suited for **imbalanced datasets**.  

### 📌 **Python Code for F1 Score:**

```python
from sklearn.metrics import f1_score
f1 = f1_score(y_test, y_pred)
print("F1 Score:", f1)
```



---

## 📋 **5. Key Differences Between Precision, Recall & F1 Score**  

| Aspect             | **Precision** | **Recall** | **F1 Score** |
|-------------------|:--------------|:------------|:--------------|
| **Definition**         | Measures the accuracy of **positive predictions**. | Measures the model’s ability to **detect positive cases**. | Balances Precision and Recall. |
| **Focuses On**          | Reducing **False Positives (FP)**. | Reducing **False Negatives (FN)**. | Achieving a trade-off between Precision and Recall. |
| **Best Use Case** | Spam detection, fraud detection, etc. | Medical diagnosis, cancer detection, etc. | Imbalanced datasets with both FP and FN concerns. |
| **Value Range**       | 0 to 1 (Higher = Better) | 0 to 1 (Higher = Better) | 0 to 1 (Higher = Better) |



## 🔥 **6. Practical Example with Python Code**  

### 📁 **Dataset: Diabetes Prediction**  

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score, f1_score, classification_report

# Sample Dataset
data = {
    'Glucose': [148, 85, 183, 89, 137, 116, 78, 115, 197, 125],
    'BMI': [33.6, 26.6, 23.3, 28.1, 43.1, 25.6, 31.0, 35.3, 30.5, 34.3],
    'Age': [50, 31, 32, 21, 45, 29, 41, 28, 53, 54],
    'Outcome': [1, 0, 1, 0, 1, 0, 0, 1, 1, 0]  # Binary target
}
df = pd.DataFrame(data)

# Split Data
X = df[['Glucose', 'BMI', 'Age']]
y = df['Outcome']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train Logistic Regression Model
model = LogisticRegression()
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Evaluation
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

## 🚨 **7. Choosing the Right Metric for Your Problem**  

| **Scenario**           | **Best Metric** |
|------------------------|-----------------|
| **Balanced Dataset**     | **Accuracy** |
| **Imbalanced Dataset**   | **F1 Score** |
| **Minimize False Positives (FP)** | **Precision** |
| **Minimize False Negatives (FN)** | **Recall** |
| **Maximize Overall Model Performance** | **AUC-ROC** |

---

## 🏆 **8. Key Takeaways**  

✅ **Precision** → Focuses on reducing **False Positives**.  
✅ **Recall** → Focuses on reducing **False Negatives**.  
✅ **F1 Score** → Best for **imbalanced datasets** or when you need a balance between **Precision** and **Recall**.  
✅ **Use Precision for spam detection**, and **Recall for medical diagnosis**.  

💡 **Pro Tip:** For **imbalanced datasets**, rely on **F1 Score**, **Precision**, or **Recall** — not just **Accuracy**.
