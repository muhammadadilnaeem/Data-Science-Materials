#### **Import Libraries**

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

#### **Load the dataset**

```python
df = pd.read_csv(r"E:\Data Science Materials\100 days of Machine Learning CampusX\data\titanic_train.csv")

df.head()
```

```python
df.info()
```

- Change datatypes of `Survived` and `Pclass` to category.

```python
df["Survived"] = df["Survived"].astype('category')
df["Pclass"] = df["Pclass"].astype('category')
```

### **1. Categorical Data**

#### **a. Countplot**

- For Categorical Variables

```python
sns.countplot(x=df['Pclass'])
plt.show()
```

```python
sns.countplot(x=df['Survived'])
plt.show()
```

```python
sns.countplot(x=df['Sex'])
plt.show()
```

```python
sns.countplot(x=df['Embarked'])
plt.show()
```

#### **b. Piechart**

- For Categorical Variables

```python
df['Pclass'].value_counts().plot(kind="pie",autopct='%.2f')
plt.show()
```

```python
df['Sex'].value_counts().plot(kind="pie",autopct='%.2f')
plt.show()
```

### **2. Numerical Data**

#### **a. Histogram**


```python
# plt.hist(df['Age'])
sns.histplot(data=df,x="Age",bins=15)
plt.show()
```

```python
sns.histplot(data=df,x="Age",bins=15,kde=True)
plt.show()
```

#### **b. boxplot**


```python
sns.boxplot(df['Fare'])
plt.show()
```

```python
sns.boxplot(df['Age'])
plt.show()
```
