### **Working With Dates**

```python
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

#### **1. Extract Year**

```python
date['date_year'] = date['date'].dt.year

date.head()
```

#### **2. Extract Month**

```python
date['date_month'] = date['date'].dt.month

date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()

date.head()
```

#### **3. Extract Day**

```python
date['date_day'] = date['date'].dt.day

date.head()
```

```python
# day of week
date['date_day_of_week'] = date['date'].dt.dayofweek

date.head()
```

```python
# day of week name
date['date_day_of_week_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# is weekend?

date['date_is_weekend'] = np.where(date['date_day_of_week_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

#### **4. Extract Week of the Year**

```python
# Extract the week number from the 'date' column and create a new column 'date_week'
date['date_week'] = date['date'].dt.isocalendar().week

# Drop unnecessary columns ('product_id', 'city_id', 'orders') from the DataFrame
# Display the first few rows of the updated DataFrame
date.drop(columns=['product_id', 'city_id', 'orders']).head()
```

#### **5. Extract quarter of the Year**

```python
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).head()
```

#### **5. Extract Smester of the Year**

```python
# Create a new column 'semester' in the DataFrame 'date'
# Assign a value of 1 if 'quarter' is either 1 or 2 (indicating the first semester)
# Assign a value of 2 if 'quarter' is 3 or 4 (indicating the second semester)
date['semester'] = np.where(date['quarter'].isin([1, 2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).head()
```

#### **6. Time Elapsed Between 2 Dates**

```python
today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
from dateutil.relativedelta import relativedelta

# Calculate the number of months passed
months_passed = date['date'].apply(lambda x: relativedelta(today, x).years * 12 + relativedelta(today, x).months)

months_passed
```

### **Time**

```python
time.head()
```

```python
time.info()
```

```python
time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

```python
time['hour']= time['date'].dt.hour
time['minute']= time['date'].dt.minute
time['second']= time['date'].dt.second

time.head()
```

#### **Extract the time part**

```python
time['time'] = time['date'].dt.time

time.head()
```

```python
# Create a new column with the datetime part
time['datetime'] = time['date']

# Calculate the time difference
today - time['datetime']
```

```python
# in seconds

(today - time['date'])/np.timedelta64(1,'s')
```

```python
# in minutes

(today - time['date'])/np.timedelta64(1,'m')
```

```python
# in hours

(today - time['date'])/np.timedelta64(1,'h')
```
