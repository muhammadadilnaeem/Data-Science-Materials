````python
import pandas as pd
from ydata_profiling import ProfileReport

import warnings
warnings.filterwarnings("ignore")

# Load dataset
df = pd.read_csv(r"E:\Data Science Materials\100 days of Machine Learning CampusX\data\titanic_train.csv")

# Generate profile report
profile = ProfileReport(df, title='Pandas Profiling Titanic EDA Report', explorative=True)

# Save report to HTML
profile.to_file('titanic_report.html')
````
