#### **`OneHotEncoding Using Scikit-Learn`**

```python
# split the data into train and test set

# This function is used to split the dataset into training and testing subsets.

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    df.iloc[:, 0:4],   # Features: select the first four columns of the DataFrame 'df'
    df.iloc[:, -1],    # Target variable: select the last column of 'df'
    test_size=0.2,     # Specify that 20% of the data should be used for testing
)
```

```python
# Importing the OneHotEncoder from a library (usually scikit-learn)
# This encoder is used to convert categorical variables into a format that can be provided to ML algorithms.

# Create an instance of OneHotEncoder
ohe = OneHotEncoder(drop='first') # drop first category of encoded columns

# Fit the OneHotEncoder on the training data and transform the 'fuel' and 'owner' columns
# This will convert these categorical variables into one-hot encoded format for the training set
X_train_new = ohe.fit_transform(X_train[['fuel', 'owner']]).toarray()

# Transform the test data using the previously fitted OneHotEncoder
# This applies the same one-hot encoding to the test set without fitting again
X_test_new = ohe.transform(X_test[['fuel', 'owner']]).toarray()
```

### ***OneHotEncoding with Top Categories***

```python
df['brand'].value_counts()
```

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold = 100
```

```python
# Identify the indices of elements in the 'counts' Series
# that have values less than or equal to a specified 'threshold'.
# These indices will be stored in the variable 'replacement'.
replacement = counts[counts <= threshold].index
```

```python
# Replace occurrences of brands in the 'replacement' list with the label 'uncommon' 
# in the 'brand' column of the DataFrame 'df'.
# Then, create dummy variables for the 'brand' column, converting it into a one-hot encoded format.

pd.get_dummies(df['brand'].replace(replacement, 'uncommon'))
```
