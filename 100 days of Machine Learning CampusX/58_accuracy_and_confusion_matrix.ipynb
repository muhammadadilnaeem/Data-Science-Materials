# 📊 ***`Classification Metrics in Machine Learning`***  

### ✅ **Why Are Classification Metrics Important?**  
In **classification models**, evaluating performance using the right metrics is crucial. Relying solely on **accuracy** can be misleading, especially with **imbalanced datasets**.  

---

## 📌 **1. Confusion Matrix**  

The **confusion matrix** is a table that shows the performance of a classification model by comparing **actual vs. predicted values**.

### 🔹 **Confusion Matrix Structure**

|               | **Predicted: 0** | **Predicted: 1** |
|----------------|-----------------|-----------------|
| **Actual: 0** | **True Negative (TN)** | **False Positive (FP)** |
| **Actual: 1** | **False Negative (FN)** | **True Positive (TP)** |

![Confusion Matrix](https://cdn.prod.website-files.com/660ef16a9e0687d9cc27474a/662c42677529a0f4e97e4f43_644bedb136edb30fc02b0753_classification_metrics_img_source_grey.001-min.png)

---

### 🔹 **Key Terms in a Confusion Matrix**  

| **Term** | **Meaning** | **Error Type** |
|------------|---------------|---------------|
| **True Positive (TP)** | Model correctly predicted **positive class** | ✅ Correct |
| **True Negative (TN)** | Model correctly predicted **negative class** | ✅ Correct |
| **False Positive (FP)** | Model incorrectly predicted **positive** instead of negative | ❌ **Type I Error** |
| **False Negative (FN)** | Model incorrectly predicted **negative** instead of positive | ❌ **Type II Error** |

---

## 📌 **2. Type I and Type II Errors**  

| **Error Type** | **Meaning** | **Impact** | **Example** |
|----------------|--------------|------------|------------|
| **Type I Error (False Positive)** | Predicting **positive** when it's actually **negative** | **False Alarm** | Diagnosing a healthy patient as sick |
| **Type II Error (False Negative)** | Predicting **negative** when it's actually **positive** | **Missed Detection** | Diagnosing a sick patient as healthy |

---

### ⚠️ **When to Focus on Type I vs Type II Error**  
✅ **Type I Error → Minimize when false alarms are costly.**  
✅ **Type II Error → Minimize when missing positive cases is dangerous.**  

| **Scenario**         | **Focus On** |
|----------------------|----------------|
| **Spam Email Filtering** | **Type I Error** (Avoid marking legitimate emails as spam) |
| **Medical Diagnosis**     | **Type II Error** (Avoid missing sick patients) |
| **Fraud Detection**       | **Type I Error** (Better to wrongly flag suspicious activity than miss fraud) |

---

## 📌 **3. Classification Metrics**  

### 🔹 **3.1. Accuracy**  
- Measures the **percentage of correct predictions**.  
- **Formula:**  
$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$  

✅ **Best for balanced datasets.**  
❌ **Misleading for imbalanced datasets.**  


```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```



---

### 🔹 **3.2. Precision (Positive Predictive Value)**  
- Measures how many predicted **positives** are actually **positive**.  
- **Formula:**  
$
Precision = \frac{TP}{TP + FP}
$  

✅ Important when **False Positives** must be minimized.  
✅ Example: **Spam detection**, where marking a real email as spam is costly.  


```python
from sklearn.metrics import precision_score
precision = precision_score(y_test, y_pred)
print("Precision:", precision)
```


---

### 🔹 **3.3. Recall (Sensitivity or True Positive Rate)**  
- Measures how many **actual positives** were correctly predicted.  
- **Formula:**  
$
Recall = \frac{TP}{TP + FN}
$  

✅ Important when **False Negatives** must be minimized.  
✅ Example: **Cancer detection**, where missing a sick patient is dangerous.  

```python
from sklearn.metrics import recall_score
recall = recall_score(y_test, y_pred)
print("Recall:", recall)
```


---

### 🔹 **3.4. F1 Score (Harmonic Mean of Precision & Recall)**  
- Combines **Precision** and **Recall** into a single score.  
- **Formula:**  
$
F1 = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}
$  

✅ Best for **imbalanced datasets**.  
✅ Ideal when both **False Positives** and **False Negatives** are costly.  

```python
from sklearn.metrics import f1_score
f1 = f1_score(y_test, y_pred)
print("F1 Score:", f1)
```



---

### 🔹 **3.5. Specificity (True Negative Rate)**  
- Measures the ability to correctly identify **negative cases**.  
- **Formula:**  
$
Specificity = \frac{TN}{TN + FP}
$  

✅ Useful in **fraud detection** and **medical testing** where **false alarms** must be reduced.  

---

### 🔹 **3.6. ROC Curve (Receiver Operating Characteristic Curve)**  
- Visualizes the **trade-off between True Positive Rate (Recall) and False Positive Rate**.  
- **AUC (Area Under Curve)** → Measures overall model performance.  

✅ **AUC = 1** → Perfect model.  
✅ **AUC = 0.5** → Random guess.  

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Probability Predictions
y_prob = log_reg.predict_proba(X_test_scaled)[:, 1]

# Compute ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_prob)

# Plot ROC Curve
plt.plot(fpr, tpr, label='ROC Curve (AUC = {:.2f})'.format(roc_auc_score(y_test, y_prob)))
plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()

# AUC Score
auc_score = roc_auc_score(y_test, y_prob)
print("AUC Score:", auc_score)
```

---

## 📌 **4. Choosing the Right Metric for Your Problem**  

| **Scenario** | **Best Metric** |
|---------------|-----------------|
| **Balanced Dataset** | **Accuracy** |
| **Imbalanced Dataset** | **F1 Score** |
| **Minimize False Positives** | **Precision** |
| **Minimize False Negatives** | **Recall** |
| **Maximize Overall Performance** | **AUC-ROC Score** |

---

## 🏆 **5. Key Takeaways**  

✅ **Accuracy** → Best for **balanced datasets**.  
✅ **Precision** → Focus on minimizing **False Positives**.  
✅ **Recall** → Focus on minimizing **False Negatives**.  
✅ **F1 Score** → Ideal for **imbalanced datasets**.  
✅ **AUC-ROC Score** → Best for evaluating **overall model performance**.  

💡 **Pro Tip:** For imbalanced datasets, always use **Precision, Recall**, or **F1 Score** instead of Accuracy.
