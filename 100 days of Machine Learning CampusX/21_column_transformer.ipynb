# 🔄 ***`ColumnTransformer in Python for Machine Learning & Data Science`*** 🚀  

## ✅ **What is ColumnTransformer?**  

**`ColumnTransformer`** is a powerful tool from `sklearn.compose` that allows you to **apply different transformations to different columns** in a dataset. This is particularly useful in **Machine Learning (ML) pipelines** when dealing with **both numerical and categorical data**.

---

## 🎯 **Why Use ColumnTransformer?**  

✔ **Handles mixed data types (numerical & categorical)** in one step.  
✔ **Automates feature engineering**—scaling, encoding, and imputing together.  
✔ **Prevents data leakage**—applies transformations only on training data.  
✔ **Works efficiently in ML Pipelines**—combines preprocessing & model training.  

---

## 📊 **1. Example Dataset**  



```python

import pandas as pd

# Sample Data
df = pd.DataFrame({
    'Age': [25, 30, 35, 40, 50],
    'Salary': [40000, 50000, 60000, 80000, 100000],
    'City': ['New York', 'London', 'Paris', 'Berlin', 'Tokyo'],
    'Purchased': ['No', 'Yes', 'No', 'Yes', 'No']
})

print(df)
```

---

## 🔄 **2. Applying ColumnTransformer for Data Preprocessing**  

We will:  
✅ **Standardize Numerical Features** → `Age`, `Salary` (**StandardScaler**)  
✅ **One-Hot Encode Categorical Features** → `City` (**OneHotEncoder**)  
✅ **Label Encode the Target Column** → `Purchased` (**LabelEncoder**)  

---

## 🛠 **3. Implementing ColumnTransformer in Python**  

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder

# Define column types
numerical_features = ['Age', 'Salary']
categorical_features = ['City']

# Define the ColumnTransformer
preprocessor = ColumnTransformer([
    ('num_scaler', StandardScaler(), numerical_features),
    ('cat_encoder', OneHotEncoder(), categorical_features)
])

# Apply transformations
X = preprocessor.fit_transform(df.drop(columns=['Purchased']))

# Encode target variable
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(df['Purchased'])

# Convert to DataFrame for readability
import numpy as np
X_transformed = pd.DataFrame(X, columns=preprocessor.get_feature_names_out())
print("\nTransformed Features:\n", X_transformed)
```

---

```python

from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")

# Create a new preprocessor with the correct column names
preprocessor = ColumnTransformer([
    ('num_scaler', StandardScaler(), ['Age', 'Salary']),
    ('cat_encoder', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'), ['City'])
])

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['Purchased']), y, test_size=0.2, random_state=42)

# Define ML Pipeline
pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42))
])

# Train model
pipeline.fit(X_train, y_train)

# Evaluate model
accuracy = pipeline.score(X_test, y_test)
print("Model Accuracy:", accuracy)
```

## 🔍 **6. When to Use ColumnTransformer?**  

| **Scenario**                           | **Use ColumnTransformer?** |
|-----------------------------------------|----------------------------|
| **Mix of numerical & categorical data** | ✅ Yes                     |
| **Feature scaling & encoding needed**   | ✅ Yes                     |
| **Preprocessing large datasets**        | ✅ Yes                     |
| **Text/NLP tasks**                      | ❌ No (Use TF-IDF Vectorizer) |

---

## 🎯 **7. Key Takeaways**  

✔ **ColumnTransformer simplifies preprocessing of mixed data types.**  
✔ **Prevents data leakage by applying transformations only to training data.**  
✔ **Works seamlessly in ML Pipelines, automating feature engineering.**  
✔ **Use with distance-based models (KNN, SVM) for better performance.**  

💡 **Pro Tip:** Always check transformed feature names using `get_feature_names_out()`.

#### **Import Libraries**

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    LabelEncoder,
    OneHotEncoder,
    OrdinalEncoder,
)

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")
```

#### **Load the Dataset**

```python
df = pd.read_csv(r"https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day28-column-transformer/covid_toy.csv")

df.sample(10)
```

```python
# check gender values count
df['gender'].value_counts()
```

```python
# check cough values count
df['cough'].value_counts()
```

```python
# check city values count
df['city'].value_counts()
```

- `gender` and `city` is `Nominal`.
- `cough` is `ordinal`.

```python
# check missing values
df.isnull().sum()
```

```python
# This function is used to split the dataset into training and testing subsets.

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    df.drop(columns=['has_covid']),   # Features
    df['has_covid'],    # Target variable
    test_size=0.2,     # Specify that 20% of the data should be used for testing
)
```

```python
X_train.sample(10)
```

### ***`ColumnTransformer`***

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1', SimpleImputer(), ['fever']),
    ('tnf2', OrdinalEncoder(categories=[['Mild','Strong']]), ['cough']),
    ('tnf3', OneHotEncoder(drop='first'), ['gender','city'])
], remainder='passthrough')
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```
