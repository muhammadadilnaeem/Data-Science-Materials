# 📊 ***`Regression Metrics in Machine Learning for Data Science`*** 🚀  

In **Machine Learning (ML)**, **Regression Metrics** help evaluate how well a regression model predicts continuous values. The most commonly used metrics are:  

✅ **Mean Squared Error (MSE)**  
✅ **Mean Absolute Error (MAE)**  
✅ **Root Mean Squared Error (RMSE)**  
✅ **R² Score (Coefficient of Determination)**  
✅ **Adjusted R² Score**  

---

## 📌 **1. Why Are Regression Metrics Important?**  

✔ **Assess Model Accuracy** → Helps determine how close predictions are to actual values.  
✔ **Compare Different Models** → Choose the best regression model.  
✔ **Detect Overfitting or Underfitting** → Ensures good generalization.  
✔ **Optimize Hyperparameters** → Fine-tune models for better performance.  

---

## 🏹 **2. Mean Absolute Error (MAE)**  

📌 **Definition:**  
- Measures the **average absolute difference** between actual values (**$Y$**) and predicted values (**$\hat{Y}$**).  
- Less sensitive to outliers compared to MSE.  

📌 **Formula:**  

$
MAE = \frac{1}{n} \sum_{i=1}^{n} |Y_i - \hat{Y}_i|
$

📌 **Example in Python:**  

```python
from sklearn.metrics import mean_absolute_error

y_actual = [50, 60, 70, 80, 90]
y_predicted = [52, 58, 68, 85, 88]

# Compute MAE
mae = mean_absolute_error(y_actual, y_predicted)
print("Mean Absolute Error (MAE):", mae)
```

📌 **When to Use?**  
- If you need an **easily interpretable metric**.  
- **Best when outliers are not a concern.**  

---

## 🎯 **3. Mean Squared Error (MSE)**  

📌 **Definition:**  
- Measures the **average squared difference** between actual and predicted values.  
- **More sensitive to outliers** (squaring amplifies large errors).  

📌 **Formula:**  

$
MSE = \frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2
$

📌 **Example in Python:**  

```python
from sklearn.metrics import mean_squared_error

# Compute MSE
mse = mean_squared_error(y_actual, y_predicted)
print("Mean Squared Error (MSE):", mse)
```


📌 **When to Use?**  
- If you want **larger errors to be penalized** more.  
- **Not ideal when outliers are present**.  

---

## 📏 **4. Root Mean Squared Error (RMSE)**  

📌 **Definition:**  
- **Square root of MSE**, making it easier to interpret in the same unit as the target variable.  
- **Still sensitive to outliers**, but more interpretable than MSE.  

📌 **Formula:**
  
$
RMSE = \sqrt{MSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2}
$

📌 **Example in Python:**  

```python
import numpy as np

# Compute RMSE
rmse = np.sqrt(mse)
print("Root Mean Squared Error (RMSE):", rmse)
```


📌 **When to Use?**  
- **Preferred when large errors should be penalized** but results need to be in the same unit as the target variable.  
- Used in **finance, stock price prediction, and forecasting models**.  

---

## 📈 **5. R² Score (Coefficient of Determination)**  

📌 **Definition:**  
- Measures the **proportion of variance** explained by the model.  
- **R² = 1** → Perfect fit.  
- **R² = 0** → Model is no better than the mean.  
- **R² < 0** → Model performs worse than simply using the mean.  

📌 **Formula:**  

$
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
$

Where:  
- **$ SS_{res} $** = Sum of squared residuals = $ \sum (Y_i - \hat{Y}_i)^2 $  
- **$ SS_{tot} $** = Total sum of squares = $ \sum (Y_i - \bar{Y})^2 $  

📌 **Example in Python:**  

```python
from sklearn.metrics import r2_score

# Compute R² Score
r2 = r2_score(y_actual, y_predicted)
print("R² Score:", r2)
```

📌 **When to Use?**  
- If you want to measure **how much variance the model explains**.  
- **Best for model comparison.**  

---

## 📊 **6. Adjusted R² Score**  

📌 **Problem with R² Score:**  
- R² **always increases** when adding more features, even if they are useless.  

📌 **Solution:**  

- **Adjusted R² Score** accounts for the number of predictors $p$ and prevents overfitting.  

📌 **Formula:**  

$
Adjusted\ R^2 = 1 - \left( \frac{(1 - R^2)(n - 1)}{n - p - 1} \right)
$

📌 **Example in Python:**  

```python
# Function to compute Adjusted R²
def adjusted_r2(r2, n, p):
    return 1 - ((1 - r2) * (n - 1) / (n - p - 1))

# Number of data points (n) and features (p)
n = len(y_actual)
p = 2  # Assume 2 predictors in the model

# Compute Adjusted R² Score
adj_r2 = adjusted_r2(r2, n, p)
print("Adjusted R² Score:", adj_r2)
```


📌 **When to Use?**  
- **When comparing multiple regression models** with different numbers of features.  
- Helps **avoid overfitting** in **multiple linear regression**.  

---

## 🎯 **7. When to Use Each Metric?**  

| **Metric** | **Best When…** | **Avoid When…** |
|------------|-----------------|-----------------|
| **MAE (Mean Absolute Error)** | Need an **easy-to-understand metric** | Large errors should be penalized more |
| **MSE (Mean Squared Error)** | Need to **penalize large errors** | Outliers exist |
| **RMSE (Root Mean Squared Error)** | Need a **unit-consistent metric** | Outliers exist |
| **R² Score** | Want to **measure explained variance** | Overfitting is a concern |
| **Adjusted R² Score** | Comparing models with **different features** | Only one predictor is used |

---

## 🏆 **8. Key Takeaways**  

✔ **MAE is the most interpretable** but treats all errors equally.  
✔ **MSE penalizes large errors more**, making it useful for financial applications.  
✔ **RMSE is useful when units should match the target variable.**  
✔ **R² Score explains variance**, but **Adjusted R² prevents overfitting**.  
✔ **Always check multiple metrics** before finalizing a model.  

💡 **Pro Tip:** Always **scale data** before applying regression models to avoid biased results.
