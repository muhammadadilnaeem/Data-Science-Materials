# **📉 Loss Functions in Binary Classification**

In binary classification, **loss functions** guide how models learn by quantifying “how wrong” the predictions are.

We'll explore three commonly used loss functions:

| Loss Function           | Best For                       | Model Types             |
|------------------------|--------------------------------|-------------------------|
| 🤖 **Perceptron Loss**     | Linearly separable data          | Perceptron              |
| 📏 **Hinge Loss**          | Margin-based classification      | SVMs (Support Vector Machines) |
| 🔐 **Binary Cross Entropy** | Probabilistic output             | Logistic Regression, Neural Nets |

---

## 🔹 **1. Perceptron Loss Function**

### 📐 Formula:

$$
L(\mathbf{w}) =
\begin{cases}
0 & \text{if } y(\mathbf{w} \cdot \mathbf{x} + b) > 0 \\
- y(\mathbf{w} \cdot \mathbf{x} + b) & \text{otherwise}
\end{cases}
$$

Where:
- $y \in \{-1, +1\}$ is the true label  
- $\mathbf{w} \cdot \mathbf{x} + b$ is the raw output  

### ✅ Pros:
- Simple and fast for linearly separable data

### ❌ Cons:
- Not differentiable ➡️ can’t use with gradient descent  
- Doesn't work well for non-linear problems

---

## 🧱 **2. Hinge Loss (Used in SVMs)**

### 📐 Formula:

$$
L = \max(0, 1 - y \cdot (\mathbf{w} \cdot \mathbf{x} + b))
$$

Where:
- $y \in \{-1, +1\}$  
- The loss is **0** if the prediction is at least 1 unit away from the decision boundary on the correct side.

### ✨ Intuition:
- Encourages not just **correct classification**, but also a **margin of confidence**.

### ✅ Pros:
- Supports large margin classification  
- Useful when we want more robust generalization

### ❌ Cons:
- Doesn’t give probabilistic outputs  
- Only works with binary labels $\{-1, +1\}$

---

## 🔐 **3. Binary Cross-Entropy (Log Loss)**

### 📐 Formula:

For $y \in \{0, 1\}$ and prediction $\hat{y}$:

$$
L = - \left[ y \cdot \log(\hat{y}) + (1 - y) \cdot \log(1 - \hat{y}) \right]
$$

Where $\hat{y} \in [0, 1]$ is the predicted probability.

### 🔎 Used in:
- **Logistic Regression**
- **Neural Networks** with **Sigmoid activation**

### ✨ Intuition:
- Penalizes incorrect predictions **based on confidence**
- A confident wrong prediction gets a **very high loss**

### ✅ Pros:
- Works well with probabilistic models  
- Differentiable — perfect for backpropagation

### ❌ Cons:
- Sensitive to **imbalanced data**
- Can be harsh on overly confident wrong predictions

# 🧩 Summary Table:

| Feature                    | **Perceptron Loss**       | **Hinge Loss**                | **Binary Cross Entropy**           |
|---------------------------|---------------------------|-------------------------------|------------------------------------|
| 🔢 Output Type            | Hard (0 or 1)             | Margin-based (not prob.)      | Probabilistic (0 to 1)             |
| 🧠 Differentiable         | ❌                         | ✅ (but not smooth)            | ✅ Smooth & differentiable         |
| 📊 Use Case              | Simple linear models       | Support Vector Machines       | Logistic Regression, Neural Nets  |
| 🧮 Formula Basis          | Sign of prediction         | Margin & classification       | Probabilistic error                |
| 🚫 Limitation             | Only linearly separable    | No probabilistic outputs       | Sensitive to confident mistakes    |

---

### 🔁 Visual Analogy:

```
Prediction Score ─────────▶  
|                     |                  |                   
| 🟢 Perceptron:      | Only cares if right or wrong  
| 🟡 Hinge:           | Cares how confidently right you are  
| 🔴 Cross Entropy:   | Cares how confident you are — especially if wrong  
```
