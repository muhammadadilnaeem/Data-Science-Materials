# **🧠 What is MLP (Multi-Layer Perceptron)?**

An **MLP** is a type of **feedforward neural network** consisting of:

1. **Input layer**  
2. One or more **hidden layers**  
3. An **output layer**  

It uses **non-linear activation functions** and is trained using **backpropagation**. 🔄

---

# 🔢 **MLP Notation Guide**

Let’s break it down layer-by-layer using proper symbols and equations.

---

## 📥 **1. Input Layer**

Let:
- $\mathbf{x} \in \mathbb{R}^n$ = input vector with $n$ features  
- $x_i$ = $i^{th}$ feature of input

Example: $\mathbf{x} = [x_1, x_2, \dots, x_n]^T$

---

## 🧱 **2. Hidden Layers**

Let’s define:
- $L$ = number of layers (excluding input layer)  
- $l$ = index of current layer ($l = 1, 2, \dots, L$)  
- $a^{[l]}$ = activations (outputs) of layer $l$  
- $z^{[l]}$ = linear output before activation in layer $l$

### 📐 Forward Propagation:

Each layer performs:

#### **Step 1: Weighted Sum + Bias**
$$
z^{[l]} = \mathbf{W}^{[l]} a^{[l-1]} + \mathbf{b}^{[l]}
$$

#### **Step 2: Activation**
$$
a^{[l]} = f^{[l]}(z^{[l]})
$$

Where:
- $\mathbf{W}^{[l]}$ = weight matrix of layer $l$  
- $\mathbf{b}^{[l]}$ = bias vector  
- $f^{[l]}$ = activation function (ReLU, Sigmoid, Tanh, etc.)  

---

## 🎯 **3. Output Layer**

Let:
- $a^{[L]}$ = final output of the MLP  
- For binary classification, often:
  $$
  a^{[L]} = \sigma(z^{[L]})
  $$
  Where $\sigma$ is the sigmoid function

---

## **🔁 Full MLP Flow:**

For a 3-layer MLP (1 hidden layer):

### **🔹 Layer 1 (Hidden Layer):**
$$
z^{[1]} = \mathbf{W}^{[1]} \mathbf{x} + \mathbf{b}^{[1]} \\
a^{[1]} = f^{[1]}(z^{[1]})
$$

### **🔹 Layer 2 (Output Layer):**
$$
z^{[2]} = \mathbf{W}^{[2]} a^{[1]} + \mathbf{b}^{[2]} \\
a^{[2]} = f^{[2]}(z^{[2]})
$$

---

## **🧮 Notation Summary Table:**

| Symbol               | Meaning                                     |
|----------------------|---------------------------------------------|
| $\mathbf{x}$         | Input vector                                |
| $L$                  | Number of layers (excluding input)          |
| $z^{[l]}$            | Pre-activation (weighted sum) at layer $l$  |
| $a^{[l]}$            | Activation/output at layer $l$              |
| $\mathbf{W}^{[l]}$   | Weight matrix at layer $l$                   |
| $\mathbf{b}^{[l]}$   | Bias vector at layer $l$                     |
| $f^{[l]}(\cdot)$     | Activation function of layer $l$            |

---

## **🔌 Common Activation Functions:**

| Function | Formula                                | Range        | Use Case                       |
|----------|----------------------------------------|--------------|-------------------------------|
| ReLU     | $\max(0, z)$                           | $[0, \infty)$ | Hidden layers (default)       |
| Sigmoid  | $\frac{1}{1 + e^{-z}}$                 | $(0, 1)$      | Binary classification outputs |
| Tanh     | $\frac{e^z - e^{-z}}{e^z + e^{-z}}$     | $(-1, 1)$     | Hidden layers (less common)   |
| Softmax  | $\frac{e^{z_i}}{\sum e^{z_j}}$         | $(0, 1)$      | Multi-class outputs           |

---

## **🔄 Training Process (in short):**

1. **Forward pass** using the above equations  
2. Compute **loss** (e.g., Binary Cross-Entropy)  
3. **Backpropagation** to compute gradients  
4. Update weights using **gradient descent**  

---

## **🎨 Visual Overview:**

```
Input (x1, x2, ..., xn)
        ↓
   [ Weighted Sum ]
        ↓
  [ Activation f() ]
        ↓
   Hidden Layer(s)
        ↓
   Output Layer → Prediction (ŷ)
```
